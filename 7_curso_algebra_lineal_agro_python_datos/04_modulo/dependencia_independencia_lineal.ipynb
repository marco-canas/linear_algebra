# **Geometría de la Información: Dependencia e Independencia Lineal como Filtro de Redundancia en la Optimización de Datos Agropecuarios**

Para generar esta imagen introductoria con **Plotly**, utilizaremos una visualización en 3D que ilustre la diferencia fundamental: dos vectores que forman un plano (**Independencia Lineal**) y un tercer vector que "colapsa" en ese mismo plano o sobre uno de los vectores (**Dependencia Lineal**).



En el contexto agropecuario, esto representa cómo una variable (como la humedad del suelo) podría ser simplemente una combinación lineal de otras, no aportando información nueva al modelo de Machine Learning.



````python
import plotly.graph_objects as go
import numpy as np

# Definición de vectores
# v1 y v2 son linealmente independientes (forman la base de un plano)
# v3 es linealmente dependiente (v3 = v1 + v2)
v1 = np.array([1, 0, 0.5])
v2 = np.array([0, 1, 0.8])
v3 = v1 + v2  # Combinación lineal (Dependencia)

fig = go.Figure()

# Vector 1: Nutrientes Suelo A
fig.add_trace(go.Scatter3d(x=[0, v1[0]], y=[0, v1[1]], z=[0, v1[2]],
    mode='lines+markers+text', name='Vector v1 (Independiente)',
    line=dict(color='blue', width=6), text=["", "v1"]))

# Vector 2: Nutrientes Suelo B
fig.add_trace(go.Scatter3d(x=[0, v2[0]], y=[0, v2[1]], z=[0, v2[2]],
    mode='lines+markers+text', name='Vector v2 (Independiente)',
    line=dict(color='green', width=6), text=["", "v2"]))

# Vector 3: Redundancia (Dependencia Lineal)
fig.add_trace(go.Scatter3d(x=[0, v3[0]], y=[0, v3[1]], z=[0, v3[2]],
    mode='lines+markers+text', name='Vector v3 (v1 + v2: Dependiente)',
    line=dict(color='red', width=6, dash='dash'), text=["", "v3 = v1+v2"]))

# Configuración del diseño
fig.update_layout(
    title="Visualización de Dependencia Lineal en el Espacio de Características",
    scene=dict(
        xaxis_title='Variable X (Nitrógeno)',
        yaxis_title='Variable Y (Fósforo)',
        zaxis_title='Variable Z (Rendimiento)',
        camera=dict(eye=dict(x=1.5, y=1.5, z=0.8))
    ),
    margin=dict(l=0, r=0, b=0, t=40),
    legend=dict(yanchor="top", y=0.99, xanchor="left", x=0.01)
)

fig.show()
fig.write_image(r"C:\\Users\\marco\\Downloads\\dependencia_lineal.png")
````

# 8. Generación de Datasets Artificiales (Python)

A continuación, generamos los datos que emulan estas situaciones para demostrar cómo la **Independencia Lineal** permite la reducción dimensional.


````python
import pandas as pd
import numpy as np

# Configuración de reproducibilidad
np.random.seed(42)
n_muestras = 100

# --- DATASET SITUACIÓN A (Agricultura de Precisión) ---
humedad = np.random.uniform(20, 45, n_muestras)  # X1
evapotransp = np.random.uniform(2, 8, n_muestras) # X2
# X3 es linealmente dependiente: 0.5*X1 + 2*X2 + Ruido mínimo
deficit_vapor = 0.5 * humedad + 2 * evapotransp

df_agro = pd.DataFrame({
    'Humedad_Suelo': humedad,
    'Evapotranspiracion': evapotransp,
    'Deficit_Vapor': deficit_vapor
})

# --- DATASET SITUACIÓN B (Administración de Empresas) ---
costo_fert = np.random.normal(5000, 500, n_muestras) # C1
costo_mano_obra = np.random.normal(3000, 300, n_muestras) # C2
# Gasto_Total es una combinación lineal exacta: C1 + C2
gasto_total = costo_fert + costo_mano_obra 

df_admin = pd.DataFrame({
    'Costo_Fertilizantes': costo_fert,
    'Costo_Mano_Obra': costo_mano_obra,
    'Gasto_Total_Redundante': gasto_total
})

# --- VERIFICACIÓN DE INDEPENDENCIA LINEAL ---
def verificar_independencia(df, nombre):
    matriz = df.values
    rango = np.linalg.matrix_rank(matriz)
    n_columnas = matriz.shape[1]
    
    print(f"--- Análisis para: {nombre} ---")
    print(f"Número de variables: {n_columnas}")
    print(f"Rango de la matriz: {rango}")
    if rango < n_columnas:
        print("Resultado: Hay Dependencia Lineal (Redundancia detectada).")
    else:
        print("Resultado: Las variables son Linealmente Independientes.")
    print("-" * 40)

verificar_independencia(df_agro, "Dataset Agropecuario")
verificar_independencia(df_admin, "Dataset Administrativo")
````


# **Interpretación para Ciencia de Datos:**

En ambos casos, el **Rango** de la matriz es menor que el número de columnas (Rango = 2, Columnas = 3). Esto indica que podemos reducir el problema de 3 dimensiones a 2 sin perder un solo bit de información relevante. Esto mejora el **desempeño predictivo** al eliminar el ruido y la redundancia que confunden a los pesos del modelo de Deep Learning.

# 9. Definiciones Asociadas al Módulo (Lenguaje $\LaTeX$)

En el contexto del Álgebra Lineal aplicada a la Ciencia de Datos, definimos la independencia de nuestras variables (vectores de características) de la siguiente manera:



# **Definición 1: Combinación Lineal**

Sea un conjunto de vectores $\{v_1, v_2, \dots, v_n\}$ en $\mathbb{R}^m$. Una **combinación lineal** de estos vectores es un vector de la forma:


$$w = c_1v_1 + c_2v_2 + \dots + c_nv_n$$


donde $c_i \in \mathbb{R}$ son escalares (pesos o coeficientes). En Machine Learning, estos $c_i$ representan a menudo los parámetros o coeficientes de importancia de una variable.



### **Definición 2: Independencia Lineal**

Se dice que un conjunto de vectores $\{v_1, v_2, \dots, v_n\}$ es **linealmente independiente** si la única solución a la ecuación vectorial:


$$c_1v_1 + c_2v_2 + \dots + c_nv_n = \mathbf{0}$$


es la solución trivial:


$$c_1 = c_2 = \dots = c_n = 0$$


Si existe al menos un $c_i \neq 0$ que satisfaga la ecuación, el conjunto es **linealmente dependiente**.

### **Definición 3: Rango de una Matriz ($rank$)**

El rango de una matriz $A \in \mathbb{R}^{m \times n}$, denotado como $rank(A)$, es el número máximo de columnas (o filas) linealmente independientes.

> **Nota para Data Science:** Si en un dataset de $n$ variables el $rank(A) < n$, tenemos redundancia y la matriz de covarianza no será invertible, causando fallos en algoritmos de optimización.

# 10. Ejemplos de Aplicación (Basados en las situaciones del punto 7)

## **Ejemplo 1: Caso Agropecuario (Variables de Suelo)**

Consideremos los vectores de las mediciones de los tres primeros lotes para las variables del **Dataset A**:

* $v_1$ (Humedad): $\begin{bmatrix} 25 \\ 30 \end{bmatrix}$
* $v_2$ (Evapotranspiración): $\begin{bmatrix} 4 \\ 5 \end{bmatrix}$
* $v_3$ (Déficit Vapor): $\begin{bmatrix} 20.5 \\ 25 \end{bmatrix}$



Si observamos que $v_3 = 0.5v_1 + 2v_2$, entonces existen escalares $c_1=0.5, c_2=2, c_3=-1$ tales que:


$$0.5 \begin{bmatrix} 25 \\ 30 \end{bmatrix} + 2 \begin{bmatrix} 4 \\ 5 \end{bmatrix} - 1 \begin{bmatrix} 20.5 \\ 25 \end{bmatrix} = \begin{bmatrix} 0 \\ 0 \end{bmatrix}$$




Como existen coeficientes no nulos, las variables son **linealmente dependientes**. El sensor de Déficit de Vapor no aporta información nueva que no pueda inferirse de los otros dos.



# **Ejemplo 2: Caso Administrativo (Costos)**

En el **Dataset B**, definimos el Gasto Total como $G_T = C_1 + C_2$.
En términos matriciales, si nuestra matriz de datos $X$ es:


$$X = \begin{bmatrix} C_1 & C_2 & C_1+C_2 \end{bmatrix}$$


Cualquier intento de calcular la inversa de $(X^T X)$ (necesaria para la Regresión Lineal clásica) fallará porque el determinante será cero. El sistema tiene infinitas soluciones para los coeficientes de predicción, lo que hace que el modelo sea **ininterpretable**.



## 11. Teoremas del Álgebra Lineal para Ciencia de Datos

Estos teoremas son los que permiten que algoritmos como **PCA (Principal Component Analysis)** o la **Regresión Ridge** funcionen:

1. **Teorema de la Existencia de Dependencia:** Un conjunto de $n$ vectores en $\mathbb{R}^m$ es siempre linealmente dependiente si $n > m$. (Si tienes más variables que registros, tus datos son inherentemente dependientes).
2. **Teorema del Rango y Nulidad:** Establece la relación entre las dimensiones del espacio de entrada y la información "perdida" por la dependencia lineal.
3. **Teorema de la Invertibilidad:** Una matriz cuadrada $A$ es invertible si y solo si sus columnas son linealmente independientes.

# Comparación de desempeño predictivo de una red neuronal con dataset de predictores dependientes y dataset de predictores independientes. 

Este es el núcleo práctico del Módulo 4. En este script, utilizaremos **TensorFlow/Keras** para construir una red neuronal sencilla y compararemos dos escenarios:



1. **Modelo Redundante:** Utiliza todas las variables, incluyendo aquellas linealmente dependientes (que generan ruido y confusión en el aprendizaje de pesos).
2. **Modelo Optimizado:** Utiliza solo las variables linealmente independientes, aplicando el criterio de rango de matriz.


````python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# 1. Generación de Datos Artificiales (Contexto Agropecuario)
np.random.seed(42)
n_muestras = 2000

# Variables Independientes (Información real)
humedad = np.random.uniform(20, 45, n_muestras) # x1
temperatura = np.random.uniform(15, 35, n_muestras) # x2

# Variables Dependientes (Redundancia Lineal)
# x3 y x4 no aportan información nueva, son combinaciones de x1 y x2
var_redundante_1 = 0.7 * humedad + 0.3 * temperatura 
var_redundante_2 = 1.5 * humedad - 0.5 * temperatura

# Variable Objetivo (Rendimiento de cosecha en Ton/Ha)
# El rendimiento real solo depende de la humedad y temperatura + ruido
rendimiento = (0.5 * humedad) + (0.8 * temperatura) + np.random.normal(0, 1, n_muestras)

# Crear Datasets
X_completo = np.column_stack([humedad, temperatura, var_redundante_1, var_redundante_2])
X_reducido = np.column_stack([humedad, temperatura])
y = rendimiento

# 2. Análisis de Rango (Criterio de Independencia Lineal)
rango_completo = np.linalg.matrix_rank(X_completo)
print(f"Dataset Completo: 4 columnas, pero Rango = {rango_completo}")
print(f"Dataset Reducido: 2 columnas, Rango = {np.linalg.matrix_rank(X_reducido)}")

# 3. Preprocesamiento
scaler = StandardScaler()
X_comp_scaled = scaler.fit_transform(X_completo)
X_red_scaled = scaler.fit_transform(X_reducido)

def entrenar_modelo(X_train, y_train, nombre):
    model = tf.keras.Sequential([
        layers.Input(shape=(X_train.shape[1],)),
        layers.Dense(16, activation='relu'),
        layers.Dense(8, activation='relu'),
        layers.Dense(1)
    ])
    model.compile(optimizer='adam', loss='mse', metrics=['mae'])
    
    print(f"\n--- Entrenando Modelo: {nombre} ---")
    history = model.fit(X_train, y_train, epochs=50, batch_size=32, verbose=0, validation_split=0.2)
    return history

# Dividir datos
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(X_comp_scaled, y, test_size=0.2)
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(X_red_scaled, y, test_size=0.2)

# 4. Ejecución de la Comparativa
historia_completa = entrenar_modelo(X_train_c, y_train_c, "Con Redundancia (4 variables)")
historia_reducida = entrenar_modelo(X_train_r, y_train_r, "Reducido por Independencia (2 variables)")

# 5. Visualización de Desempeño
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(historia_completa.history['val_loss'], label='Val Loss (Redundante)', color='red')
plt.plot(historia_reducida.history['val_loss'], label='Val Loss (Independiente)', color='green')
plt.title('Error Cuadrático Medio (MSE) en Validación')
plt.xlabel('Epochs')
plt.ylabel('MSE')
plt.legend()

plt.subplot(1, 2, 2)
plt.bar(['Redundante', 'Independiente'], 
        [historia_completa.history['val_mae'][-1], historia_reducida.history['val_mae'][-1]],
        color=['red', 'green'])
plt.title('Error Medio Absoluto (MAE) Final')
plt.ylabel('Error en Ton/Ha')

plt.tight_layout()
plt.savefig(r"C:\\Users\\marco\\Downloads\\comparativa_modelos_redundancia.png")
plt.show()

print("\nConclusión:")
print("El modelo reducido suele converger más rápido o mantener una estabilidad similar con menos costo computacional.")
````


# **Puntos clave de este script para tu clase:**

1. **Criterio de Rango:** El script demuestra que aunque el `X_completo` tiene 4 columnas, su rango es 2. Esto es la prueba matemática de que hay 2 variables que no sirven para el aprendizaje.
2. **Eficiencia en Deep Learning:** Al reducir el número de entradas, la red neuronal tiene menos pesos que optimizar ($w_i$). Esto reduce el riesgo de **overfitting** (sobreajuste) y hace que el modelo sea más fácil de desplegar en dispositivos con pocos recursos (como sensores en campo o apps móviles).
3. **Interpretabilidad:** En administración de empresas, es más valioso saber que el rendimiento depende de la "Humedad" que de una "Variable X" que es una mezcla confusa de varias mediciones.

# Script de Python: El Peligro de la Redundancia con Datos Ruidosos


````python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# 1. Generación de Datos con Ruido en la Redundancia
np.random.seed(42)
n_muestras = 2500

# Variables Originales (Limpias)
humedad = np.random.uniform(20, 45, n_muestras)
temperatura = np.random.uniform(15, 35, n_muestras)

# Variables Redundantes con RUIDO (Simulando sensores defectuosos)
# Matemáticamente deberían ser dependientes, pero el ruido las "ensucia"
ruido_sensor = np.random.normal(0, 5, n_muestras) # Ruido fuerte
var_redundante_ruidosa = 0.5 * humedad + 1.2 * temperatura + ruido_sensor

# Variable Objetivo (Solo depende de la realidad, no del ruido del sensor)
rendimiento = (0.6 * humedad) + (0.9 * temperatura) + np.random.normal(0, 1, n_muestras)

# Datasets
X_ruidoso = np.column_stack([humedad, temperatura, var_redundante_ruidosa])
X_limpio = np.column_stack([humedad, temperatura])
y = rendimiento

# 2. Preprocesamiento
scaler = StandardScaler()
X_ruidoso_scaled = scaler.fit_transform(X_ruidoso)
X_limpio_scaled = scaler.fit_transform(X_limpio)

# Dividir datos
X_train_ruid, X_test_ruid, y_train_ruid, y_test_ruid = train_test_split(X_ruidoso_scaled, y, test_size=0.2)
X_train_limp, X_test_limp, y_train_limp, y_test_limp = train_test_split(X_limpio_scaled, y, test_size=0.2)

# 3. Modelo de Red Neuronal
def build_model(input_shape):
    model = tf.keras.Sequential([
        layers.Dense(32, activation='relu', input_shape=(input_shape,)),
        layers.Dense(16, activation='relu'),
        layers.Dense(1)
    ])
    model.compile(optimizer='adam', loss='mse')
    return model

model_ruid = build_model(3)
model_limp = build_model(2)

# 4. Entrenamiento
print("Entrenando modelo con redundancia ruidosa...")
history_ruid = model_ruid.fit(X_train_ruid, y_train_ruid, epochs=60, verbose=0, validation_data=(X_test_ruid, y_test_ruid))

print("Entrenando modelo limpio (Independencia Lineal)...")
history_limp = model_limp.fit(X_train_limp, y_train_limp, epochs=60, verbose=0, validation_data=(X_test_limp, y_test_limp))

# 5. Visualización de la "Trampa" de la Redundancia
plt.figure(figsize=(10, 6))
plt.plot(history_ruid.history['val_loss'], label='Error con Variable Redundante Ruidosa', color='red', lw=2)
plt.plot(history_limp.history['val_loss'], label='Error con Variables Independientes Limpias', color='green', lw=2)
plt.yscale('log')
plt.title('Impacto de la Dependencia Lineal Ruidosa en la Predicción')
plt.xlabel('Épocas')
plt.ylabel('MSE (Escala Logarítmica)')
plt.legend()
plt.grid(True, which="both", ls="-")
plt.savefig(r"C:\\Users\\marco\\Downloads\\trampa_redundancia_ruidosa.png")
plt.show()
````


# ¿Qué lección sacamos de este gráfico para los alumnos?

2. **El Principio de Parsimonia (Navaja de Ockham):** "La explicación más simple suele ser la correcta". Si puedes explicar el rendimiento con 2 variables independientes, añadir una tercera que depende de ellas solo introduce vías para que el error entre al sistema.
3. **La Realidad Administrativa:** En una empresa, si tomas decisiones basadas en un reporte que suma otros dos (dependencia), y ese reporte tiene un error de captura, duplicarás el error en tu análisis. El álgebra lineal te dice: **"Quédate solo con las fuentes originales de información"**.
