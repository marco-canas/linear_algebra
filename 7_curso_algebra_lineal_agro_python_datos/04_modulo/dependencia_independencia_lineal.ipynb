# **Geometría de la Información: Dependencia e Independencia Lineal como Filtro de Redundancia en la Optimización de Datos Agropecuarios**

Para generar esta imagen introductoria con **Plotly**, utilizaremos una visualización en 3D que ilustre la diferencia fundamental: dos vectores que forman un plano (**Independencia Lineal**) y un tercer vector que "colapsa" en ese mismo plano o sobre uno de los vectores (**Dependencia Lineal**).



En el contexto agropecuario, esto representa cómo una variable (como la humedad del suelo) podría ser simplemente una combinación lineal de otras, no aportando información nueva al modelo de Machine Learning.



````python
import plotly.graph_objects as go
import numpy as np

# Definición de vectores
# v1 y v2 son linealmente independientes (forman la base de un plano)
# v3 es linealmente dependiente (v3 = v1 + v2)
v1 = np.array([1, 0, 0.5])
v2 = np.array([0, 1, 0.8])
v3 = v1 + v2  # Combinación lineal (Dependencia)

fig = go.Figure()

# Vector 1: Nutrientes Suelo A
fig.add_trace(go.Scatter3d(x=[0, v1[0]], y=[0, v1[1]], z=[0, v1[2]],
    mode='lines+markers+text', name='Vector v1 (Independiente)',
    line=dict(color='blue', width=6), text=["", "v1"]))

# Vector 2: Nutrientes Suelo B
fig.add_trace(go.Scatter3d(x=[0, v2[0]], y=[0, v2[1]], z=[0, v2[2]],
    mode='lines+markers+text', name='Vector v2 (Independiente)',
    line=dict(color='green', width=6), text=["", "v2"]))

# Vector 3: Redundancia (Dependencia Lineal)
fig.add_trace(go.Scatter3d(x=[0, v3[0]], y=[0, v3[1]], z=[0, v3[2]],
    mode='lines+markers+text', name='Vector v3 (v1 + v2: Dependiente)',
    line=dict(color='red', width=6, dash='dash'), text=["", "v3 = v1+v2"]))

# Configuración del diseño
fig.update_layout(
    title="Visualización de Dependencia Lineal en el Espacio de Características",
    scene=dict(
        xaxis_title='Variable X (Nitrógeno)',
        yaxis_title='Variable Y (Fósforo)',
        zaxis_title='Variable Z (Rendimiento)',
        camera=dict(eye=dict(x=1.5, y=1.5, z=0.8))
    ),
    margin=dict(l=0, r=0, b=0, t=40),
    legend=dict(yanchor="top", y=0.99, xanchor="left", x=0.01)
)

fig.show()
fig.write_image(r"C:\\Users\\marco\\Downloads\\dependencia_lineal.png")
````

# 8. Generación de Datasets Artificiales (Python)

A continuación, generamos los datos que emulan estas situaciones para demostrar cómo la **Independencia Lineal** permite la reducción dimensional.


Este es un excelente escenario para aplicar **reducción de dimensionalidad**. En tu código, la variable `Deficit_Vapor` es una combinación lineal casi perfecta de las otras dos. Al agregar la variable objetivo `Peso_Frutos`, crearemos una relación que dependa de la información real (Humedad y Evapotranspiración) para observar cómo la redundancia afecta al modelo.

Aquí tienes el código completado:

```python


````python
import pandas as pd
import numpy as np

# Configuración de reproducibilidad
np.random.seed(42)
n_muestras = 100

# --- DATASET SITUACIÓN A (Agricultura de Precisión) ---
humedad = np.random.uniform(20, 45, n_muestras)  # X1
evapotransp = np.random.uniform(2, 8, n_muestras) # X2

# X3 es linealmente dependiente: 0.5*X1 + 2*X2
# (Representa una redundancia que no aporta información nueva al sistema)
deficit_vapor = 0.5 * humedad + 2 * evapotransp + np.random.normal(0, scale = 0.5, size = n_muestras)  # Agregamos un poco de ruido

# --- VARIABLE OBJETIVO: Peso de Frutos (Gramos) ---
# El peso aumenta con la humedad óptima y disminuye si la evapotranspiración es extrema.
# Usamos solo X1 y X2 para definir la lógica real.
peso_frutos = (10 * humedad) - (15 * evapotransp) + np.random.normal(0, 5, n_muestras)

df_agro = pd.DataFrame({
    'Humedad_Suelo': humedad,
    'Evapotranspiracion': evapotransp,
    'Deficit_Vapor': deficit_vapor,
    'Peso_Frutos_g': peso_frutos  # Nuestra Variable Objetivo (Target)
})

print("Dataset con redundancia lineal:")
df_agro
````


# ¿Por qué esto es clave para tu clase de Álgebra Lineal?

1. **Redundancia (Dependencia Lineal):** Como `Deficit_Vapor` se calcula directamente de las otras dos, si intentas resolver un sistema de ecuaciones para encontrar la importancia de cada variable, obtendrás una **matriz singular** o mal condicionada.


2. **Rango de la Matriz:** Aunque tengas 3 columnas predictoras, el **rango** de la matriz de características es **2**. Esto significa que solo hay dos "direcciones" independientes de información.


3. **Reducción de Atributos:** En una práctica de Data Science, el paso siguiente sería eliminar `Deficit_Vapor`. Al ser una combinación lineal, el modelo de predicción no pierde precisión al quitarla, pero gana en estabilidad numérica y simplicidad (Navaja de Ockham).



¿Te gustaría que calculáramos la matriz de correlación para demostrar matemáticamente esta redundancia antes de entrenar un modelo?

¡Excelente decisión! Calcular la **Matriz de Correlación** es el paso lógico para confirmar la sospecha de dependencia lineal antes de limpiar los datos. En Álgebra Lineal, esto se relaciona con la idea de que si dos vectores (columnas) son linealmente dependientes, su correlación será 1 o -1.

Aquí tienes el código para visualizar la redundancia y validar matemáticamente por qué `Deficit_Vapor` debe ser eliminada:



````python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Calculamos la matriz de correlación
# El método .corr() de pandas usa el coeficiente de Pearson por defecto
matriz_corr = df_agro.corr()

# 2. Visualización con un Heatmap (Mapa de Calor)
plt.figure(figsize=(8, 6))
sns.heatmap(matriz_corr, annot=True, cmap='RdYlGn', center=0)
plt.title('Matriz de Correlación: Identificación de Redundancia')
plt.savefig(r"C:\\Users\\marco\\Downloads\\matriz_correlacion.png")  # Guardar la figura
plt.show()

# 3. Verificación de Independencia Lineal con el Rango
matriz_predictores = df_agro[['Humedad_Suelo', 'Evapotranspiracion', 'Deficit_Vapor']].values
rango = np.linalg.matrix_rank(matriz_predictores)

print(f"Rango de la matriz de predictores: {rango}")
print("Nota: El rango es 2, pero tenemos 3 columnas. Una es redundante.")
````


# ¿Qué nos dice este análisis?

1. **Correlación Perfecta:** Verás que `Deficit_Vapor` tiene una correlación muy alta (casi 1) con las otras variables. En términos de datos, esto se llama **Multicolinealidad**.
2. **Rango vs. Columnas:** El hecho de que el rango sea 2 teniendo 3 columnas confirma que el conjunto de predictores es **linealmente dependiente**.
3. **Acción en Data Science:** * Si dejas las 3 variables, un modelo de Regresión Lineal podría fallar al intentar calcular $(X^T X)^{-1}$, ya que la matriz $X^T X$ no sería invertible (su determinante sería cercano a 0).
* Al eliminar `Deficit_Vapor`, el rango sigue siendo 2 pero ahora coincide con el número de columnas. ¡Tu matriz ahora es de **rango completo** y el modelo será estable!



Con esto, ya tienes el dataset listo y "limpio" bajo los principios del álgebra. ¿Te gustaría que apliquemos una **Regresión Lineal** simple para comparar qué pasa con los coeficientes si dejamos o quitamos la variable redundante?

¡Excelente ejercicio! Este es el momento donde la teoría del **Álgebra Lineal** se encuentra con la práctica del **Machine Learning**.

Vamos a realizar un experimento comparativo. Entrenaremos dos modelos de regresión:

1. **Modelo A (Con redundancia):** Usando `Humedad`, `Evapotranspiración` y `Deficit_Vapor`.
2. **Modelo B (Sin redundancia):** Usando solo `Humedad` y `Evapotranspiración`.



````python
from sklearn.linear_model import LinearRegression

# Preparación de datos
X_redundante = df_agro[['Humedad_Suelo', 'Evapotranspiracion', 'Deficit_Vapor']]
X_limpio = df_agro[['Humedad_Suelo', 'Evapotranspiracion']]
y = df_agro['Peso_Frutos_g']

# Modelo A: Con todas las variables
modelo_a = LinearRegression()
modelo_a.fit(X_redundante, y)

# Modelo B: Sin la variable linealmente dependiente
modelo_b = LinearRegression()
modelo_b.fit(X_limpio, y)

print("--- RESULTADOS ---")
print(f"Precisión Modelo A (R2): {modelo_a.score(X_redundante, y):.4f}")
print(f"Precisión Modelo B (R2): {modelo_b.score(X_limpio, y):.4f}")
print("\nCoeficientes Modelo A:", modelo_a.coef_)
print("Coeficientes Modelo B:", modelo_b.coef_)
````


# Análisis de los resultados:

1. **Misma Precisión ($R^2$):** Verás que ambos modelos tienen exactamente el mismo desempeño. Esto demuestra matemáticamente que la variable `Deficit_Vapor` **no aportaba información nueva** al sistema; el espacio generado por los predictores es el mismo.
2. **Inestabilidad de Coeficientes:** En el Modelo A, los coeficientes pueden volverse erráticos o muy grandes. Esto sucede porque, al haber dependencia lineal, hay infinitas combinaciones de pesos que pueden dar el mismo resultado (el sistema es indeterminado).

### Interpretación desde el Álgebra Lineal

Cuando las columnas de tu matriz de datos $X$ son linealmente dependientes, la matriz $X^T X$ es **singular** (su determinante es cero).

* En teoría, la solución de mínimos cuadrados $\hat{\beta} = (X^T X)^{-1} X^T y$ **no existe** porque no se puede invertir la matriz.
* En la práctica, las librerías de Python usan aproximaciones (como la pseudoinversa de Moore-Penrose), pero el modelo resultante es menos interpretable y más sensible al ruido.

**Conclusión para tu formación:** En agricultura de precisión o cualquier área de datos, siempre debes buscar que tu matriz de diseño sea de **rango completo**. Si una variable es combinación lineal de otras, elimínala para que tu modelo sea más robusto y fácil de explicar.

````python

# --- DATASET SITUACIÓN B (Administración de Empresas) ---
costo_fert = np.random.normal(5000, 500, n_muestras) # C1
costo_mano_obra = np.random.normal(3000, 300, n_muestras) # C2
# Gasto_Total es una combinación lineal exacta: C1 + C2
gasto_total = costo_fert + costo_mano_obra 

df_admin = pd.DataFrame({
    'Costo_Fertilizantes': costo_fert,
    'Costo_Mano_Obra': costo_mano_obra,
    'Gasto_Total_Redundante': gasto_total
})

# --- VERIFICACIÓN DE INDEPENDENCIA LINEAL ---
def verificar_independencia(df, nombre):
    matriz = df.values
    rango = np.linalg.matrix_rank(matriz)
    n_columnas = matriz.shape[1]
    
    print(f"--- Análisis para: {nombre} ---")
    print(f"Número de variables: {n_columnas}")
    print(f"Rango de la matriz: {rango}")
    if rango < n_columnas:
        print("Resultado: Hay Dependencia Lineal (Redundancia detectada).")
    else:
        print("Resultado: Las variables son Linealmente Independientes.")
    print("-" * 40)

verificar_independencia(df_agro, "Dataset Agropecuario")
verificar_independencia(df_admin, "Dataset Administrativo")
````

# 9. Definiciones Asociadas al Módulo (Lenguaje $\LaTeX$)

En el contexto del Álgebra Lineal aplicada a la Ciencia de Datos, definimos la independencia de nuestras variables (vectores de características) de la siguiente manera:



# **Definición 1: Combinación Lineal**

Sea un conjunto de vectores $\{v_1, v_2, \dots, v_n\}$ en $\mathbb{R}^m$. Una **combinación lineal** de estos vectores es un vector de la forma:


$$w = c_1v_1 + c_2v_2 + \dots + c_nv_n$$


donde $c_i \in \mathbb{R}$ son escalares (pesos o coeficientes). En Machine Learning, estos $c_i$ representan a menudo los parámetros o coeficientes de importancia de una variable.



### **Definición 2: Independencia Lineal**

Se dice que un conjunto de vectores $\{v_1, v_2, \dots, v_n\}$ es **linealmente independiente** si la única solución a la ecuación vectorial:


$$c_1v_1 + c_2v_2 + \dots + c_nv_n = \mathbf{0}$$


es la solución trivial:


$$c_1 = c_2 = \dots = c_n = 0$$


Si existe al menos un $c_i \neq 0$ que satisfaga la ecuación, el conjunto es **linealmente dependiente**.

### **Definición 3: Rango de una Matriz ($rank$)**

El rango de una matriz $A \in \mathbb{R}^{m \times n}$, denotado como $rank(A)$, es el número máximo de columnas (o filas) linealmente independientes.

> **Nota para Data Science:** Si en un dataset de $n$ variables el $rank(A) < n$, tenemos redundancia y la matriz de covarianza no será invertible, causando fallos en algoritmos de optimización.

## 11. Teoremas del Álgebra Lineal para Ciencia de Datos

Estos teoremas son los que permiten que algoritmos como **PCA (Principal Component Analysis)** o la **Regresión Ridge** funcionen:

1. **Teorema de la Existencia de Dependencia:** Un conjunto de $n$ vectores en $\mathbb{R}^m$ es siempre linealmente dependiente si $n > m$. (Si tienes más variables que registros, tus datos son inherentemente dependientes).
2. **Teorema del Rango y Nulidad:** Establece la relación entre las dimensiones del espacio de entrada y la información "perdida" por la dependencia lineal.
3. **Teorema de la Invertibilidad:** Una matriz cuadrada $A$ es invertible si y solo si sus columnas son linealmente independientes.

### 1. Generación del Dataset con `numpy`

Aquí hay un detalle técnico importante: el **Costo Total** ($V_4 = V_1 \times V_2 - V_3$) implica una multiplicación entre dos variables ($V_1$ y $V_2$). En álgebra lineal, esto se considera una operación no lineal. Sin embargo, si fijamos una de las variables o si los datos resultan en una estructura específica, el rango nos dirá cuánta "libertad" real tienen los datos.

```python


````python
import numpy as np
import pandas as pd

# Configuración
np.random.seed(10)
n_registros = 50

# Generación de variables independientes
v1_unidades = np.random.randint(10, 100, n_registros)  # Unidades
v2_precio = np.random.uniform(5.0, 50.0, n_registros)  # Precio unitario
v3_descuento = np.random.uniform(0, 10, n_registros)   # Descuento

# Variable dependiente (Costo Total)
v4_costo_total = (v1_unidades * v2_precio) - v3_descuento

# Crear el DataFrame
df_inventario = pd.DataFrame({
    'Unidades': v1_unidades,
    'Precio_Unitario': v2_precio,
    'Descuento': v3_descuento,
    'Costo_Total': v4_costo_total
})

print(df_inventario.head())
````


# 2. Demostración mediante el Rango

Para saber si el conjunto es linealmente independiente, calculamos el rango de la matriz de datos.



````python
# Convertimos el dataframe a una matriz (numpy array)
matriz_inventario = df_inventario.values

# Calculamos el rango
rango = np.linalg.matrix_rank(matriz_inventario)

print(f"\nNúmero de columnas: {matriz_inventario.shape[1]}")
print(f"Rango de la matriz: {rango}")
````


# Análisis del Resultado:

* Si el **Rango = 4**, el conjunto es **linealmente independiente**. Esto ocurrirá aquí porque la operación $V_1 \times V_2$ no es una combinación lineal (no es una simple suma de columnas multiplicadas por escalares).
* Aunque $V_4$ depende de las otras, su relación es **multiplicativa**. El álgebra lineal "busca" sumas y restas.

### 3. Pregunta de Reflexión: ¿Qué columna eliminar para una Red Neuronal?

A pesar de que el rango sea 4, desde la perspectiva de **Data Science**, tenemos una "redundancia lógica".

**¿Qué columna eliminar?**
Yo eliminaría **`Costo_Total`** (o en su defecto, una de las componentes si el costo es lo que queremos predecir).

**¿Por qué?**

1. **Redundancia de Información:** El `Costo_Total` es una función determinística de las otras tres. Una Red Neuronal moderna puede aprender la operación de multiplicación, pero darle el resultado final (`Costo_Total`) le quita la necesidad de entender la relación entre precio y volumen.
2. **Fuga de Datos (Data Leakage):** Si tu objetivo es predecir, por ejemplo, la demanda futura, y dejas el `Costo_Total` como entrada, podrías estar dándole al modelo "la respuesta" de forma indirecta, lo que infla artificialmente la precisión durante el entrenamiento pero falla en la vida real.
3. **Lógica del Negocio:** Las unidades y el precio son las variables "causales". El costo es el "efecto". Para un diagnóstico de inventario, te interesa saber *cuánto* compras y a *qué precio*, ya que son las palancas que puedes mover.

---

**Conclusión:** En este caso, el álgebra lineal te diría que los vectores son independientes (porque el rango es completo), pero tu criterio de **Data Scientist** te dice que hay redundancia lógica. ¡Esa es la diferencia clave entre un matemático puro y uno aplicado al agro o los suministros!

````python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# 1. Generación de Datos Artificiales (Contexto Agropecuario)
np.random.seed(42)
n_muestras = 2000

# Variables Independientes (Información real)
humedad = np.random.uniform(20, 45, n_muestras) # x1
temperatura = np.random.uniform(15, 35, n_muestras) # x2

# Variables Dependientes (Redundancia Lineal)
# x3 y x4 no aportan información nueva, son combinaciones de x1 y x2
var_redundante_1 = 0.7 * humedad + 0.3 * temperatura 
var_redundante_2 = 1.5 * humedad - 0.5 * temperatura

# Variable Objetivo (Rendimiento de cosecha en Ton/Ha)
# El rendimiento real solo depende de la humedad y temperatura + ruido
rendimiento = (0.5 * humedad) + (0.8 * temperatura) + np.random.normal(0, 1, n_muestras)

# Crear Datasets
X_completo = np.column_stack([humedad, temperatura, var_redundante_1, var_redundante_2])
X_reducido = np.column_stack([humedad, temperatura])
y = rendimiento

# 2. Análisis de Rango (Criterio de Independencia Lineal)
rango_completo = np.linalg.matrix_rank(X_completo)
print(f"Dataset Completo: 4 columnas, pero Rango = {rango_completo}")
print(f"Dataset Reducido: 2 columnas, Rango = {np.linalg.matrix_rank(X_reducido)}")

# 3. Preprocesamiento
scaler = StandardScaler()
X_comp_scaled = scaler.fit_transform(X_completo)
X_red_scaled = scaler.fit_transform(X_reducido)

def entrenar_modelo(X_train, y_train, nombre):
    model = tf.keras.Sequential([
        layers.Input(shape=(X_train.shape[1],)),
        layers.Dense(16, activation='relu'),
        layers.Dense(8, activation='relu'),
        layers.Dense(1)
    ])
    model.compile(optimizer='adam', loss='mse', metrics=['mae'])
    
    print(f"\n--- Entrenando Modelo: {nombre} ---")
    history = model.fit(X_train, y_train, epochs=50, batch_size=32, verbose=0, validation_split=0.2)
    return history

# Dividir datos
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(X_comp_scaled, y, test_size=0.2)
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(X_red_scaled, y, test_size=0.2)

# 4. Ejecución de la Comparativa
historia_completa = entrenar_modelo(X_train_c, y_train_c, "Con Redundancia (4 variables)")
historia_reducida = entrenar_modelo(X_train_r, y_train_r, "Reducido por Independencia (2 variables)")

# 5. Visualización de Desempeño
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(historia_completa.history['val_loss'], label='Val Loss (Redundante)', color='red')
plt.plot(historia_reducida.history['val_loss'], label='Val Loss (Independiente)', color='green')
plt.title('Error Cuadrático Medio (MSE) en Validación')
plt.xlabel('Epochs')
plt.ylabel('MSE')
plt.legend()

plt.subplot(1, 2, 2)
plt.bar(['Redundante', 'Independiente'], 
        [historia_completa.history['val_mae'][-1], historia_reducida.history['val_mae'][-1]],
        color=['red', 'green'])
plt.title('Error Medio Absoluto (MAE) Final')
plt.ylabel('Error en Ton/Ha')

plt.tight_layout()
plt.savefig(r"C:\\Users\\marco\\Downloads\\comparativa_modelos_redundancia.png")
plt.show()

print("\nConclusión:")
print("El modelo reducido suele converger más rápido o mantener una estabilidad similar con menos costo computacional.")
````

# Script de Python: El Peligro de la Redundancia con Datos Ruidosos


````python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# 1. Generación de Datos con Ruido en la Redundancia
np.random.seed(42)
n_muestras = 2500

# Variables Originales (Limpias)
humedad = np.random.uniform(20, 45, n_muestras)
temperatura = np.random.uniform(15, 35, n_muestras)

# Variables Redundantes con RUIDO (Simulando sensores defectuosos)
# Matemáticamente deberían ser dependientes, pero el ruido las "ensucia"
ruido_sensor = np.random.normal(0, 5, n_muestras) # Ruido fuerte
var_redundante_ruidosa = 0.5 * humedad + 1.2 * temperatura + ruido_sensor

# Variable Objetivo (Solo depende de la realidad, no del ruido del sensor)
rendimiento = (0.6 * humedad) + (0.9 * temperatura) + np.random.normal(0, 1, n_muestras)

# Datasets
X_ruidoso = np.column_stack([humedad, temperatura, var_redundante_ruidosa])
X_limpio = np.column_stack([humedad, temperatura])
y = rendimiento

# 2. Preprocesamiento
scaler = StandardScaler()
X_ruidoso_scaled = scaler.fit_transform(X_ruidoso)
X_limpio_scaled = scaler.fit_transform(X_limpio)

# Dividir datos
X_train_ruid, X_test_ruid, y_train_ruid, y_test_ruid = train_test_split(X_ruidoso_scaled, y, test_size=0.2)
X_train_limp, X_test_limp, y_train_limp, y_test_limp = train_test_split(X_limpio_scaled, y, test_size=0.2)

# 3. Modelo de Red Neuronal
def build_model(input_shape):
    model = tf.keras.Sequential([
        layers.Dense(32, activation='relu', input_shape=(input_shape,)),
        layers.Dense(16, activation='relu'),
        layers.Dense(1)
    ])
    model.compile(optimizer='adam', loss='mse')
    return model

model_ruid = build_model(3)
model_limp = build_model(2)

# 4. Entrenamiento
print("Entrenando modelo con redundancia ruidosa...")
history_ruid = model_ruid.fit(X_train_ruid, y_train_ruid, epochs=60, verbose=0, validation_data=(X_test_ruid, y_test_ruid))

print("Entrenando modelo limpio (Independencia Lineal)...")
history_limp = model_limp.fit(X_train_limp, y_train_limp, epochs=60, verbose=0, validation_data=(X_test_limp, y_test_limp))

# 5. Visualización de la "Trampa" de la Redundancia
plt.figure(figsize=(10, 6))
plt.plot(history_ruid.history['val_loss'], label='Error con Variable Redundante Ruidosa', color='red', lw=2)
plt.plot(history_limp.history['val_loss'], label='Error con Variables Independientes Limpias', color='green', lw=2)
plt.yscale('log')
plt.title('Impacto de la Dependencia Lineal Ruidosa en la Predicción')
plt.xlabel('Épocas')
plt.ylabel('MSE (Escala Logarítmica)')
plt.legend()
plt.grid(True, which="both", ls="-")
plt.savefig(r"C:\\Users\\marco\\Downloads\\trampa_redundancia_ruidosa.png")
plt.show()
````


# ¿Qué lección sacamos de este gráfico para los alumnos?

1. **La Inestabilidad de la Dependencia:** Verás que la curva roja (redundante) suele ser más errática y termina en un error más alto. Esto sucede porque la Red Neuronal está intentando encontrar una relación lógica en una variable que es, en esencia, una copia mal hecha de las otras.
2. **El Principio de Parsimonia (Navaja de Ockham):** "La explicación más simple suele ser la correcta". Si puedes explicar el rendimiento con 2 variables independientes, añadir una tercera que depende de ellas solo introduce vías para que el error entre al sistema.
3. **La Realidad Administrativa:** En una empresa, si tomas decisiones basadas en un reporte que suma otros dos (dependencia), y ese reporte tiene un error de captura, duplicarás el error en tu análisis. El álgebra lineal te dice: **"Quédate solo con las fuentes originales de información"**.

Con este experimento, la barra del MAE del modelo redundante ahora debería ser **más alta**, demostrando por qué la independencia lineal es el primer paso de cualquier limpieza de datos seria.
