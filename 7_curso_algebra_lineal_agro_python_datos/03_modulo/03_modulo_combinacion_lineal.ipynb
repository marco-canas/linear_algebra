```python
import plotly.graph_objects as go
import numpy as np

# 1. Definición de Vectores Base (Insumos/Características)
# Supongamos: v1 = [Nitrógeno=2, Fósforo=1, RendimientoBase=0]
#            v2 = [Nitrógeno=1, Fósforo=3, RendimientoBase=0]
# El rendimiento base es 0 para que vivan en el "piso" R2 primero.
v1 = np.array([2, 1, 0])
v2 = np.array([1, 3, 0])

# 2. Definición de Escalares (Decisión de Administración)
c1 = 1.5
c2 = 2.0

# 3. Cálculo de la Combinación Lineal (Resultado Productivo)
# y = c1*v1 + c2*v2
# y = 1.5*[2,1,0] + 2.0*[1,3,0] = [3,1.5,0] + [2,6,0] = [5, 7.5, 0]
y = c1 * v1 + c2 * v2

# 4. Generación del Plano (Espacio Generado - Span)
# Creamos una malla de escalares para visualizar el plano generado por v1 y v2
s = np.linspace(-1, 4, 10)
t = np.linspace(-1, 4, 10)
S, T = np.meshgrid(s, t)

# El plano es S*v1 + T*v2
plane_x = S * v1[0] + T * v2[0]
plane_y = S * v1[1] + T * v2[1]
plane_z = S * v1[2] + T * v2[2] # En este caso simple, z será siempre 0.

# 5. Creación de la Figura Plotly
fig = go.Figure()

# -- A. Dibujar el Espacio Generado (Plano translúcido) --
fig.add_trace(go.Surface(x=plane_x, y=plane_y, z=plane_z,
                         colorscale='Greens', opacity=0.3,
                         showscale=False, name='Espacio Generado (Span)'))

# -- B. Dibujar Vectores Base (Conjunto Generador) --
# Vector 1
fig.add_trace(go.Scatter3d(x=[0, v1[0]], y=[0, v1[1]], z=[0, v1[2]],
                           mode='lines+markers+text',
                           line=dict(color='blue', width=6),
                           marker=dict(size=4),
                           text=["", "<b>v₁</b> (Insumo A)"], textposition="top center",
                           name='Vector Generador v₁'))
# Vector 2
fig.add_trace(go.Scatter3d(x=[0, v2[0]], y=[0, v2[1]], z=[0, v2[2]],
                           mode='lines+markers+text',
                           line=dict(color='blue', width=6),
                           marker=dict(size=4),
                           text=["", "<b>v₂</b> (Insumo B)"], textposition="top center",
                           name='Vector Generador v₂'))

# -- C. Dibujar la Combinación Lineal (Resultado) --
fig.add_trace(go.Scatter3d(x=[0, y[0]], y=[0, y[1]], z=[0, y[2]],
                           mode='lines+markers+text',
                           line=dict(color='red', width=8, dash='dash'),
                           marker=dict(size=6, symbol='diamond'),
                           text=["", f"<b>y</b> = {c1}v₁ + {c2}v₂<br>(Resultado)"],
                           textposition="bottom right",
                           name='Combinación Lineal y'))

# 6. Configuración del Diseño (Layout) - Estilo Profesional
fig.update_layout(
    title=dict(
        text="Visualización de Combinación Lineal y Espacio Generado<br>"
             "<sup>Contexto: Optimización de Insumos Agropecuarios (v₁, v₂) para obtener un Resultado (y)</sup>",
        x=0.5, y=0.95
    ),
    scene=dict(
        xaxis_title='Característica 1 (ej. Nitrógeno)',
        yaxis_title='Característica 2 (ej. Fósforo)',
        zaxis_title='Eje Algebraico (z=0)',
        camera=dict(eye=dict(x=1.5, y=1.5, z=0.8)) # Perspectiva inicial
    ),
    margin=dict(l=0, r=0, b=0, t=60),
    legend=dict(yanchor="top", y=0.99, xanchor="left", x=0.01),
    template="plotly_white"
)

# 7. Mostrar la imagen (En un Jupyter Notebook)
fig.show()
fig.write_image(r"C:\Users\marco\Downloads\módulo3_combinacion_lineal.png", scale=2)  # Para exportar como imagen de alta calidad

# Para exportar como HTML estático (para GitHub pages o YouTube descripciones):
# fig.write_html("módulo3_combinacion_lineal.html")
```

En el estudio del Álgebra Lineal, la **combinación lineal** es la operación fundamental que permite "construir" nuevos elementos a partir de otros preexistentes.

Comprender el **Espacio Generado** ($Span$) es entender el alcance de nuestro modelo: si un problema del mundo real se encuentra fuera del espacio generado por nuestras variables, nuestro modelo jamás podrá resolverlo.

## 9. Definiciones Asociadas al Módulo 3

Utilizaremos el lenguaje **LaTeX** para definir con precisión los conceptos que rigen el comportamiento de nuestros datasets.



### A. Combinación Lineal
Sea $V$ un espacio vectorial sobre un cuerpo $K$ (generalmente $\mathbb{R}$). Un vector $v \in V$ es una **combinación lineal** de los vectores $v_1, v_2, \dots, v_n \in V$ si existen escalares $c_1, c_2, \dots, c_n \in K$ tales que:

$$
v = c_1v_1 + c_2v_2 + \dots + c_nv_n = \sum_{i=1}^{n} c_iv_i
$$

En nuestro contexto:
* $v_i$ son las columnas de nuestros datasets (composición de fertilizantes o fondos).
* $c_i$ son las cantidades a aplicar o invertir (nuestras variables de decisión).



### B. Conjunto Finito Generador
Se dice que un conjunto de vectores $S = \{v_1, v_2, \dots, v_n\}$ es un **conjunto generador** de un espacio vectorial $V$ si todo vector en $V$ puede escribirse como una combinación lineal de los elementos de $S$. Es decir:

$$V = \left\{ \sum_{i=1}^{n} c_iv_i \mid c_i \in K \right\}$$



# C. Espacio Generado ($Span$)
El **espacio generado** por un conjunto de vectores $S$, denotado como $gen(S)$ o $Span(S)$, es el conjunto de todas las combinaciones lineales posibles de los vectores en $S$. Matemáticamente:

$$
gen\ (v_1, v_2, \dots, v_n) = \{ v \in V : v = c_1v_1 + \dots + c_nv_n \}
$$

> **Nota para Data Science:** Si nuestro dataset tiene 10 variables, pero 3 de ellas son combinaciones lineales de las otras, el *Span* de nuestros datos tiene una dimensión menor a 10. Esto es la base de la **Reducción de Dimensionalidad**.

## 11. Teoremas

Para que un ingeniero o administrador pueda confiar en sus modelos, debe conocer estos fundamentos:



### Teorema A: Existencia de la Combinación Lineal
Sea $A$ una matriz cuyas columnas son los vectores $\{v_1, v_2, \dots, v_n\}$ y sea $\mathbf{b}$ un vector de resultados. El vector $\mathbf{b}$ es una combinación lineal de las columnas de $A$ **si y solo si** el sistema de ecuaciones $A\mathbf{x} = \mathbf{b}$ es consistente (tiene al menos una solución).



### Teorema B: El Espacio Generado y el Rango
Sea $S = \{v_1, v_2, \dots, v_n\}$ un conjunto de vectores en $\mathbb{R}^m$. El espacio generado $Span(S)$ es igual a todo $\mathbb{R}^m$ si y solo si el rango de la matriz formada por estos vectores es igual a $m$ ($Rank(A) = m$).

> **Implicación en Data Science:** Si el rango de tu matriz de características (features) es menor al número de variables, tienes **multicolinealidad**. Esto significa que algunas variables son redundantes y podrías simplificar tu modelo sin perder información (Reducción de dimensionalidad).




## 12. Aplicación de Teoremas a los Datasets

### Caso 1: Consistencia en el Cultivo de Cacao
Utilizando el **Teorema A**, para saber si podemos satisfacer los requerimientos del suelo, construimos la matriz de fertilizantes $A_{agro}$ y verificamos si el vector objetivo $\mathbf{b}$ "vive" en su espacio columna.

En Python, esto se verifica calculando el rango de la matriz de coeficientes y comparándolo con el rango de la matriz aumentada $[A|\mathbf{b}]$. Si son iguales, la combinación lineal existe.





### Caso 2: Cobertura del Mercado en Administración
Utilizando el **Teorema B**, la administradora debe verificar si sus tres fondos ($A, B, C$) generan todo el espacio de riesgo $\mathbb{R}^3$ (Bioinsumos, Logística, Exportación). 
* Si $Rank(A_{admin}) = 3$, ella puede replicar **cualquier** índice objetivo que se le presente. 
* Si $Rank(A_{admin}) < 3$, hay perfiles de riesgo que sus fondos no pueden alcanzar.

## 13. Solución a las Situaciones Problemáticas

A continuación, resolvemos matemáticamente ambos casos usando los datos generados en el ítem 8 mediante `numpy`.



```python
import numpy as np

# --- SOLUCIÓN SITUACIÓN 1 (AGRO) --- # AC = b
A_agro = np.array([[0.2, 0.1, 0.1],
                   [0.1, 0.3, 0.1],
                   [0.1, 0.2, 0.4]])
b_agro = np.array([25, 15, 20])

# Resolver para c1, c2, c3 (litros de cada fertilizante)
sol_agro = np.linalg.solve(A_agro, b_agro)
sol_agro 
```

```python

print(f"Solución Agropecuaria:")
print(f"Litros F1: {sol_agro[0]:.2f}, F2: {sol_agro[1]:.2f}, F3: {sol_agro[2]:.2f}")
```

```python

# --- SOLUCIÓN SITUACIÓN 2 (ADMIN) ---
A_admin = np.array([[0.6, 0.1, 0.3],
                    [0.2, 0.7, 0.3],
                    [0.2, 0.2, 0.4]])
b_admin = np.array([0.4, 0.3, 0.3])

# Resolver para determinar pesos de inversión
sol_admin = np.linalg.solve(A_admin, b_admin)

print(f"\nSolución Administrativa (Pesos del Portafolio):")
print(f"Inversión Fondo A: {sol_admin[0]*100:.1f}%")
print(f"Inversión Fondo B: {sol_admin[1]*100:.1f}%")
print(f"Inversión Fondo C: {sol_admin[2]*100:.1f}%")
```

**Resultado del análisis:**
1. **Agro:** Existe una combinación única. Se requieren cantidades específicas de cada producto para no desperdiciar nutrientes ni dejar el suelo pobre.
2. **Admin:** El índice objetivo se logra invirtiendo proporciones exactas de cada fondo, demostrando que el índice objetivo pertenece al espacio generado por los activos disponibles.

# Estado del arte con citas



El uso de la tecnología en la enseñanza de las matemáticas encuentra uno de sus pilares en **Papert (1980)**, quien afirmaba que *“los niños no aprenden sobre la computadora, aprenden con la computadora”* (p. 6). 

En su visión, la programación se convierte en un medio para la construcción activa de significados matemáticos, idea que se enlaza con la epistemología genética de **Piaget (1970)**, quien sostuvo que *“el conocimiento se construye a través de la acción”* (p. 15), y con la perspectiva de **Vygotsky (1978)**, al señalar que *“toda función en el desarrollo cultural del niño aparece dos veces: primero en el plano social y después en el individual”* (p. 57). 

Estos enfoques coinciden en otorgar a las herramientas culturales un papel esencial en la mediación del aprendizaje.



Posteriormente, **Wing (2006)** acuñó el término pensamiento computacional, definiéndolo como *“una habilidad fundamental para todos, no solo para los científicos de la computación”* (p. 33). 

Esta autora subrayó que debía enseñarse de manera transversal, en paralelo con la lectura, la escritura y la matemática. 

En la misma línea, el **NCTM (2000)** destacó que *“la tecnología es esencial en la enseñanza y el aprendizaje de las matemáticas; influye en la matemática que se enseña y mejora el aprendizaje de los estudiantes”* (p. 24).



Desde la didáctica de las matemáticas, **Duval (2006)** mostró que *“la comprensión matemática requiere la coordinación de al menos dos registros de representación diferentes”* (p. 107), lo que legitima la incorporación de registros digitales como LaTeX, Markdown y Python. 

Por su parte, **Godino (2017)** planteó que *“el significado de un objeto matemático está constituido por el sistema de prácticas en el que interviene”* (p. 7), con lo cual la práctica digital se convierte en parte constitutiva del aprendizaje matemático.



Finalmente, organismos internacionales como la **ACM y el IEEE (2013)** enfatizan que *“las habilidades computacionales son esenciales para el ejercicio profesional en cualquier disciplina”* (p. 12). 

De esta manera, se consolida un consenso: la enseñanza universitaria de las matemáticas debe integrar herramientas contemporáneas como **Jupyter Notebooks, Python, LaTeX, Markdown e inglés**, no para competir con la enseñanza tradicional, sino para **fortalecer la comprensión matemática, potenciar la comunicación científica y responder a las demandas del mundo laboral**.
