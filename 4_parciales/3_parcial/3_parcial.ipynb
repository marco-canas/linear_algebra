# Escalonamiento y reducción de las matrices del parcial  3 



```python
from sympy import Matrix, Rational

def obtener_matriz_rref():
    print("--- Calculadora de Matriz Escalonada Reducida (RREF) ---")
    
    try:
        # 1. Preguntar el orden de la matriz
        filas = int(input("Introduce el número de filas: "))
        columnas = int(input("Introduce el número de columnas: "))
        
        elementos = []
        
        # 2. Preguntar las componentes
        print(f"\nIntroduce las componentes fila por fila:")
        for i in range(filas):
            fila_actual = []
            for j in range(columnas):
                valor = input(f"Elemento [{i+1}][{j+1}]: ")
                # Convertimos la entrada a Rational para mantener precisión total
                fila_actual.append(Rational(valor))
            elementos.append(fila_actual)
        
        # 3. Crear la matriz de SymPy
        A = Matrix(elementos)
        
        print("\nMatriz original:")
        print(A)
        
        # 4. Calcular la RREF
        # rref() devuelve una tupla: (matriz_rref, indices_pivote)
        matriz_rref, pivotes = A.rref()
        
        print("\nForma escalonada reducida (RREF):")
        return matriz_rref

    except ValueError:
        return "Error: Por favor, introduce números válidos (enteros, decimales o fracciones como '1/3')."
    except Exception as e:
        return f"Ocurrió un error: {e}"
```

```python
obtener_matriz_rref()
```
