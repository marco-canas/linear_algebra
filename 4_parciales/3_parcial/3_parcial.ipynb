# Escalonamiento y reducción de las matrices del parcial  3 



```python
from sympy import Matrix, Rational

def analizar_matriz(datos_matriz):
    """
    Recibe una lista de listas, calcula la RREF, el Espacio Nulo
    y el Espacio Columna.
    """
    try:
        # 1. Crear la matriz de SymPy
        elementos_rational = [[Rational(celda) for celda in fila] for fila in datos_matriz]
        A = Matrix(elementos_rational)
        
        print("--- Análisis de Matriz ---")
        print(f"Matriz original ({A.rows}x{A.cols}):")
        if 'display' in globals(): display(A) 
        else: print(A)
        
        # 2. Calcular la RREF y obtener índices de pivotes
        matriz_rref, pivotes = A.rref()
        
        # 3. Calcular el Espacio Nulo (Nullspace)
        espacio_nulo = A.nullspace()
        
        # 4. Calcular el Espacio Columna (Column Space)
        # Se obtienen las columnas originales basándose en los índices de pivote
        espacio_columna = [A.col(i) for i in pivotes]
        
        return {
            "rref": matriz_rref,
            "pivotes": pivotes,
            "espacio_nulo": espacio_nulo,
            "espacio_columna": espacio_columna
        }

    except Exception as e:
        return f"Ocurrió un error al procesar la matriz: {e}"

# --- EJEMPLO DE USO ---
mi_matriz = [
    [0, 2, -3, 1, 2],
    [0, -2, 3, 3, 1],
    [0, 4, -6, 6, 7]
]

resultado = analizar_matriz(mi_matriz)

if isinstance(resultado, dict):
    print("\n1. Forma Escalonada Reducida (RREF):")
    print(resultado["rref"])
    
    print("\n2. Base del Espacio Nulo (Nullspace):")
    if not resultado["espacio_nulo"]:
        print("El espacio nulo contiene solo el vector cero.")
    else:
        for i, v in enumerate(resultado["espacio_nulo"]):
            print(f"v_{i+1}: {v}")

    print("\n3. Base del Espacio Columna (Column Space):")
    for i, v in enumerate(resultado["espacio_columna"]):
        print(f"c_{i+1}: {v}")
else:
    print(resultado)
```

```python
from sympy import Matrix, Rational

class Matriz:
    def __init__(self, datos):
        """Inicializa la matriz y precalcula su forma escalonada."""
        elementos_rational = [[Rational(celda) for celda in fila] for fila in datos]
        self.A = Matrix(elementos_rational)
        # Precalculamos la RREF y los índices de los pivotes
        self._rref_matrix, self._pivotes = self.A.rref()

    def rref(self):
        """Retorna la Forma Escalonada Reducida por Filas (RREF)."""
        return self._rref_matrix

    def fila(self):
        """Retorna la base para el espacio fila."""
        # Se toman las filas no nulas de la RREF
        return [self._rref_matrix.row(i) for i in range(self.rango())]

    def columna(self):
        """Retorna la base para el espacio columna usando las columnas originales."""
        return [self.A.col(i) for i in self._pivotes]

    def rango(self):
        """Retorna el rango (número de pivotes)."""
        return len(self._pivotes)

    def nulidad(self):
        """Retorna la nulidad (dimensión del espacio nulo)."""
        return self.A.cols - self.rango()

    def nulo(self):
        """Retorna la base para el espacio nulo."""
        return self.A.nullspace()

    def __str__(self):
        return f"Matriz Original ({self.A.rows}x{self.A.cols}):\n{self.A}"
```

```python

# --- PRUEBA DE LA CLASE CON TUS DATOS ---

datos = [
    [1,1,-1,2],   
    [1,2,2,-3], 
    [2,3,1,1]]

# Crear instancia del objeto
m = Matriz(datos)

print(m)
print("-" * 30)
print(f"Forma Escalonada Reducida (RREF):")
print(m.rref())

print(f"\n Rango: {m.rango()}")
print(f" Nulidad: {m.nulidad()}")

print("\n Base Espacio Fila (Filas base):")
for v in m.fila(): print(v)

print("\n Base Espacio Columna (Columnas base):")
for v in m.columna(): print(v)

print("\n Base Espacio Nulo (Vectores nulos):")
base_nula = m.nulo()
if not base_nula:
    print("El espacio nulo contiene solo el vector cero (Nulidad 0).")
else:
    for v in base_nula: print(v)
```
