# 4. 

Sean $x = (1, 2, 4)$ y $y = (-3,2,0)$, dos vectores de $\mathbb{R}^{3}$ y sea 


$$
W = \{ u \in \mathbb{R}^{3} \mid u = \alpha x + \beta y, \ \alpha, \beta \in \mathbb{R} \}
$$  

Muestre que $W$ es subespacio de $\mathbb{R}^{3}$. 



```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Vectores del problema
x = np.array([1,2,4])
y = np.array([-3,2,0])

# Crear figura con dos gráficos
fig = make_subplots(
    rows=1, cols=2,
    specs=[[{'type':'scene'}, {'type':'scene'}]],
    subplot_titles=[
        "Vectores $x$ y $y$ en $\mathbb{R}^3$",
        "Subespacio $W = span\{x,y\}$"
    ]
)

# -------- función para dibujar vectores con flecha ----------
def vector_con_flecha(v, nombre):

    line = go.Scatter3d(
        x=[0,v[0]],
        y=[0,v[1]],
        z=[0,v[2]],
        mode="lines",
        line=dict(width=6),
        name=nombre
    )

    cone = go.Cone(
        x=[v[0]],
        y=[v[1]],
        z=[v[2]],
        u=[v[0]],
        v=[v[1]],
        w=[v[2]],
        sizemode="absolute",
        sizeref=0.4,
        showscale=False,
        name=nombre
    )

    return line, cone


# --------- ejes coordenados ----------
def ejes():

    L = 5

    eje_x = go.Scatter3d(x=[-L,L], y=[0,0], z=[0,0], mode='lines', name='x')
    eje_y = go.Scatter3d(x=[0,0], y=[-L,L], z=[0,0], mode='lines', name='y')
    eje_z = go.Scatter3d(x=[0,0], y=[0,0], z=[-L,L], mode='lines', name='z')

    return [eje_x,eje_y,eje_z]


# -------- gráfico 1 ----------
for eje in ejes():
    fig.add_trace(eje,row=1,col=1)

for vec,name in [(x,"x"),(y,"y")]:
    l,c = vector_con_flecha(vec,name)
    fig.add_trace(l,row=1,col=1)
    fig.add_trace(c,row=1,col=1)


# -------- plano generado por x,y ----------
a = np.linspace(-2,2,10)
b = np.linspace(-2,2,10)

A,B = np.meshgrid(a,b)

X = A*x[0] + B*y[0]
Y = A*x[1] + B*y[1]
Z = A*x[2] + B*y[2]

plane = go.Surface(x=X,y=Y,z=Z, opacity=0.5, showscale=False, name="Plano W")


# -------- gráfico 2 ----------
for eje in ejes():
    fig.add_trace(eje,row=1,col=2)

for vec,name in [(x,"x"),(y,"y")]:
    l,c = vector_con_flecha(vec,name)
    fig.add_trace(l,row=1,col=2)
    fig.add_trace(c,row=1,col=2)

fig.add_trace(plane,row=1,col=2)


# layout
fig.update_layout(
    height=600,
    width=1100,
    title="Subespacio generado por $x$ y $y$"
)

fig.show()
fig.write_image(r"C:\Users\marco\Downloads\espacio_generado_por_dos_vectores.png")
```

## Interpretación matemática

Sean
$x = (1,2,4)$,
$y = (-3,2,0)$.

El conjunto

$$
W = \{ u \in \mathbb{R}^{3} \mid u = \alpha x + \beta y,\ \alpha,\beta \in \mathbb{R} \}
$$

es el **span** de los vectores $x$ y $y$.



Para probar que es subespacio:

1. **Contiene el vector cero**

Tomando $\alpha=0$ y $\beta=0$

$$
0x + 0y = (0,0,0)
$$

Entonces $0 \in W$.

---



2. **Cerrado bajo suma**

Sean

$$
u_1 = \alpha_1 x + \beta_1 y
$$

$$
u_2 = \alpha_2 x + \beta_2 y
$$



Entonces

$$
u_1 + u_2
=
(\alpha_1+\alpha_2)x + (\beta_1+\beta_2)y
\in W
$$

---



3. **Cerrado bajo multiplicación por escalar**

Sea $\lambda \in \mathbb{R}$

$$
\lambda u
=
\lambda(\alpha x + \beta y)
=
(\lambda\alpha)x + (\lambda\beta)y
\in W
$$

---



- Por lo tanto

$$
W \leq \mathbb{R}^3
$$

es un **subespacio vectorial**.



Geométricamente:

* Si $x$ y $y$ **no son paralelos**, generan un **plano que pasa por el origen**, que es lo que muestra el gráfico.

En este caso no son proporcionales, por lo que

$$
\dim(W)=2
$$
