#  Introducción 

En muchas aplicaciones es necesario emplear subconjuntos de espacios vectoriales, que son a su vez espacios vectoriales. 

En el módulo 1 vimos que el conjunto de ternas ordenadas que forman un plano que pasa por el origen es un espacio vectorial, así que tanto $\mathbb{R}^{3}$ como este subconjunto son espacios vectoriales.

# 2.1 Definición y criterio de subespacio

Sea $H$ un subconjunto no vacío de un espacio vectorial $V$ y supóngase que $H$ en sí mismo es un espacio vectorial con las operaciones de suma y multiplicación por un escalar definidas en $V$.   

Entonces se dice que $H$ es un subespacio de $V$.


El siguiente teorema nos mostrará un resultado que simplificará notablemente el trabajo para determinar si un subconjunto de $V$ es o no un subespacio de $V$.

# Teorema 1: Criterio de subespacio  


Un subconjunto no vacío $H$ del espacio vectorial $V$ es un subespacio de $V$ si se
cumplen las dos reglas de cerradura. Esto es:  

1. Si $\mathbf{x} \in H$ e $\mathbf{y} \in H$, entonces $\mathbf{x}+\mathbf{y} \in H$. 
2. Si $\mathbf{x} \in H$, entonces $\alpha \mathbf{x} \in H$ para todo escalar $\alpha$. 


La demostración del teorema consiste en verificar que al darse las dos reglas de cerradura se verifican los demás axiomas de la definición de espacio vectorial.


Como los vectores de $H$ también están en $V$, las leyes de asociatividad, conmutatividad, de distribución y de identidad multiplicativa (axiomas ii, v, vii, viii, ix y x) se cumplen, es decir, ya tenemos 6 axiomas de 10 que se cumplen, más dos de la condición de teorema serían 8. Así solo nos faltan dos por verificar. 

Ahora, si, $x \in H$ entonces $0x \in H$ por la segunda condición del teorema anterior (condición 2) y $0x = 0$ (teorema 1, módulo 1).  

De igual manera, $(-1)x \in H$ y  $(-1)x = -x$ (teorema 1, módulo 1), verificándose el axioma iv.  


Con esto queda completa la demostración.

El teorema anterior establece que para probar si un conjunto $H$, $H \subset V$, es o no un
subespacio de $V$, basta con probar las dos reglas de cerradura.


# Consecuencias del Teorema 

Como consecuencia del teorema podemos expresar lo siguiente:  

i. Todo subespacio de un espacio vectorial $V$ contiene el $\mathbf{0}$.
ii. Todo espacio vectorial tiene dos subespacios triviales:  el mismo espacio $(V \subset V)$ y el conjunto que tiene como único elemento el módulo. ¿Por qué?    



Pues, porque el conjunto $\{0\}$ cumple las dos reglas de cerradura: 
- Si $x \in \{0\}$ e $y \in \{0\}$, entonces $x+y = 0+0 = 0 \in \{0\}$. 
- Si $x \in \{0\}$, entonces $\alpha x = \alpha 0 = 0 \in \{0\}$ para todo escalar $\alpha$.
por lo tanto, $\{0\}$ es un subespacio de $V$.


Los subespacios distintos a $V$ y a $\{ \mathbf{0} \}$ se llaman subespacios propios.

Para verificar la primera regla de cerradura, sean $(x_{1}, y_{1})$ y $(x_{2}, y_{2})$ dos vectores arbitrarios de $H$. 

Entonces, por definición de $H$, se tiene que: 
$y_{1} = mx_{1}$ y $y_{2} = mx_{2}$. 

 
Luego, la suma de estos vectores es:  
$$
(x_{1}, y_{1}) + (x_{2}, y_{2}) = (x_{1} + x_{2}, y_{1} + y_{2}) = (x_{1} + x_{2}, mx_{1} + mx_{2}) = (x_{1} + x_{2}, m(x_{1} + x_{2}))
$$  
Como $x_{1} + x_{2} \in \mathbb{R}$, se concluye que $(x_{1} + x_{2}, y_{1} + y_{2}) \in H$.


Por lo tanto, la primera regla de cerradura se cumple.  


Para verificar la segunda regla de cerradura, sea $(x, y) \in H$ y $\alpha \in \mathbb{R}$ un escalar arbitrario. 

Entonces, por definición de $H$, se tiene que $y = mx$.  
Luego, el producto por el escalar $\alpha$ es:      
$$
\alpha (x, y) = (\alpha x, \alpha y) = (\alpha x, \alpha mx) = (\alpha x, m(\alpha x))
$$  
Como $\alpha x \in \mathbb{R}$, se concluye que $(\alpha x, \alpha y) \in H$.  
Por lo tanto, la segunda regla de cerradura se cumple. 

 
Así, por el criterio de subespacio, $H$ es un subespacio de $\mathbb{R}^{2}$.

4. Sabemos que todo polinomio es una función continua, así que el conjunto de polinomios de grado menor o igual que $n$ en el intervalo cerrado $[a,b]$. $\mathbb{P}_{n}[a,b]$ es un subconjunto del conjunto de funciones continuas dentro del mismo intervalo.  

$$
\mathbb{P}_{n}[a,b] \subset C[a,b]
$$  

Como $\mathbb{P}_{n}[a,b]$ es un espacio vectorial para todo entero $n$, entonces es un subespacio de $C[a,b]$.  



````python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

x = np.linspace(-10, 10, 400)

def visualizar_polinomio(n, a, b):
    # Ajuste por si a > b
    if a >= b:
        print("Debe cumplirse que a < b")
        return

    # Dominio en [a, b]
    x = np.linspace(a, b, 400)

    # Coeficientes aleatorios enteros
    # c0 + c1 x + ... + cn x^n
    coeficientes = np.random.randint(-5, 6, size=n+1)

    # Evaluación del polinomio
    y = np.zeros_like(x, dtype=float)
    for k, ck in enumerate(coeficientes):
        y += ck * x**k

    # Gráfica
    plt.figure(figsize=(7, 5))
    plt.plot(x, y, label=f"Coeficientes: {coeficientes}")
    plt.axhline(0)
    plt.axvline(0)
    plt.xlim(a-1, b+1)
    plt.xticks(np.arange(a-1, b+1, step=1))
    plt.grid(True)

    plt.title(
        rf"Polinomio aleatorio en $\mathbb{{P}}_{n}[{a},{b}]$"
    )
    plt.xlabel("x")
    plt.ylabel("p(x)")
    plt.legend()
    plt.savefig(r"C:\Users\marco\Downloads\visualizacion_polinomio.png")
    plt.show()

interact(
    visualizar_polinomio,
    n=widgets.IntSlider(
        value=2,
        min=0,
        max=6,
        step=1,
        description='n'
    ),
    a=widgets.IntSlider(
        value=-5,
        min=-10,
        max=0,
        step=1,
        description='a'
    ),
    b=widgets.IntSlider(
        value=5,
        min=1,
        max=10,
        step=1,
        description='b'
    )
)
````

# 5. Sea $C'[ a,b ]$ el conjunto de funciones con primera derivada continua, definidas en $[a,b]$.   

Queremos demostrar que
$$
C'[a,b]={f:[a,b]\to\mathbb{R};|; f \text{ es derivable en } (a,b) \text{ y } f' \text{ es continua en } [a,b]}
$$
es un **subespacio vectorial** de
$$
C[a,b]={f:[a,b]\to\mathbb{R};|; f \text{ es continua en } [a,b]}.
$$

La idea es verificar las tres condiciones de subespacio.

---



# 1. $C'[a,b] \subseteq C[a,b]$

Si $f \in C'[a,b]$, entonces $f$ tiene derivada continua en $[a,b]$.

Sabemos por teoría básica del cálculo que:

**Si una función es derivable en un intervalo, entonces es continua en ese intervalo.**

Por lo tanto,

$$
f \text{ derivable } \Rightarrow f \text{ continua}.
$$

Luego,

$$
f \in C[a,b].
$$

Así,

$$
C'[a,b] \subseteq C[a,b].
$$

---



# 2. Contiene al vector cero

El vector cero de $C[a,b]$ es

$$
f(x)=0.
$$

Derivada:

$$
f'(x)=0.
$$

La función constante $0$ es continua en $[a,b]$.

Por lo tanto

$$
f \in C'[a,b].
$$

Así,

$$
0 \in C'[a,b].
$$

---



# 3. Cerrado bajo suma

Sean

$$
f,g \in C'[a,b].
$$

Entonces

* $f'$ es continua
* $g'$ es continua

Consideremos

$$
h=f+g.
$$

Sabemos que

$$
h'(x)=f'(x)+g'(x).
$$

Pero:

* suma de funciones continuas = continua

Luego

$$
h' \text{ es continua en } [a,b].
$$

Por tanto

$$
h \in C'[a,b].
$$

Así,

$$
f+g \in C'[a,b].
$$

---



# 4. Cerrado bajo multiplicación por escalar

Sea

$$
f \in C'[a,b], \quad \lambda \in \mathbb{R}.
$$

Definamos

$$
h=\lambda f.
$$

Entonces

$$
h'(x)=\lambda f'(x).
$$

Como $f'$ es continua y el producto por constante preserva continuidad,

$$
h' \text{ es continua}.
$$

Por lo tanto

$$
\lambda f \in C'[a,b].
$$

---



# 5. Conclusión

Se cumplen:

1. $0 \in C'[a,b]$
2. Cerrado bajo suma
3. Cerrado bajo producto por escalar
4. $C'[a,b] \subseteq C[a,b]$

Por lo tanto

$$
C'[a,b] \le C[a,b].
$$

Es decir,

$$
\boxed{C'[a,b] \text{ es un subespacio vectorial de } C[a,b].}
$$

---



# Interpretación matemática importante (para tus estudiantes)

Este resultado muestra que:

* El conjunto de funciones **suaves** (derivada continua)
* vive dentro del espacio de funciones continuas
* y **respeta la estructura lineal**.

En análisis funcional esto se usa para estudiar:

* ecuaciones diferenciales
* operadores diferenciales
* espacios normados como

$$
C^1[a,b].
$$

En el contexto actual, organizaciones como la **ACM** y el **IEEE** insisten en que las competencias computacionales son indispensables en la formación profesional, lo que refuerza la pertinencia de integrar lenguajes como **Python, LaTeX, Markdown e inglés** en ambientes como **Jupyter Notebooks** para la enseñanza universitaria de las matemáticas.

# Estado del arte con citas

El uso de la tecnología en la enseñanza de las matemáticas encuentra uno de sus pilares en **Papert (1980)**, quien afirmaba que *“los niños no aprenden sobre la computadora, aprenden con la computadora”* (p. 6). 

En su visión, la programación se convierte en un medio para la construcción activa de significados matemáticos, idea que se enlaza con la epistemología genética de **Piaget (1970)**, quien sostuvo que *“el conocimiento se construye a través de la acción”* (p. 15), y con la perspectiva de **Vygotsky (1978)**, al señalar que *“toda función en el desarrollo cultural del niño aparece dos veces: primero en el plano social y después en el individual”* (p. 57). 

Estos enfoques coinciden en otorgar a las herramientas culturales un papel esencial en la mediación del aprendizaje.



Posteriormente, **Wing (2006)** acuñó el término pensamiento computacional, definiéndolo como *“una habilidad fundamental para todos, no solo para los científicos de la computación”* (p. 33). Esta autora subrayó que debía enseñarse de manera transversal, en paralelo con la lectura, la escritura y la matemática. 

En la misma línea, el **NCTM (2000)** destacó que *“la tecnología es esencial en la enseñanza y el aprendizaje de las matemáticas; influye en la matemática que se enseña y mejora el aprendizaje de los estudiantes”* (p. 24).



Desde la didáctica de las matemáticas, **Duval (2006)** mostró que *“la comprensión matemática requiere la coordinación de al menos dos registros de representación diferentes”* (p. 107), lo que legitima la incorporación de registros digitales como LaTeX, Markdown y Python. Por su parte, **Godino (2017)** planteó que *“el significado de un objeto matemático está constituido por el sistema de prácticas en el que interviene”* (p. 7), con lo cual la práctica digital se convierte en parte constitutiva del aprendizaje matemático.



Finalmente, organismos internacionales como la **ACM y el IEEE (2013)** enfatizan que *“las habilidades computacionales son esenciales para el ejercicio profesional en cualquier disciplina”* (p. 12). 

De esta manera, se consolida un consenso: la enseñanza universitaria de las matemáticas debe integrar herramientas contemporáneas como **Jupyter Notebooks, Python, LaTeX, Markdown e inglés**, no para competir con la enseñanza tradicional, sino para **fortalecer la comprensión matemática, potenciar la comunicación científica y responder a las demandas del mundo laboral**.

# Subespacios Vectoriales y Estructura de los Datos

En ciencia de datos, los datos reales casi nunca ocupan todo el espacio $\mathbb{R}^n$.
Frecuentemente viven en **estructuras de menor dimensión**, llamadas **subespacios**.

Ejemplo en agricultura:

Variables medidas:

* lluvia
* humedad
* temperatura

Pero muchas veces existe relación entre ellas.

Por ejemplo

$$
temperatura \approx 0.5\cdot lluvia + 2
$$

Esto significa que los datos **no ocupan todo $\mathbb{R}^3$** sino una región más pequeña.

Comprender esto es clave en:

* reducción de dimensión
* PCA
* modelos predictivos
* redes neuronales

---



# 1. Problema motivador (contexto agropecuario)

Un sistema de sensores mide:

* nitrógeno
* fósforo
* potasio

Pero en cierto suelo se cumple aproximadamente

$$
N + P + K = 100
$$

Cada medición es un vector

$$
v = (N,P,K)
$$

Ejemplo:

$$
v = (40,30,30)
$$

Otro:

$$
w = (50,20,30)
$$

Pregunta:

¿Dónde viven todos estos vectores?

---



# 2. Exploración con Python

```python


````python
import numpy as np

v = np.array([40,30,30])
w = np.array([50,20,30])

print(v.sum())
print(w.sum())
````

```

Salida esperada

```
100
100
```

Interpretación:

Todos los vectores cumplen

$$
x_1 + x_2 + x_3 = 100
$$

Esto describe **un plano** en $\mathbb{R}^3$.

Pero ojo:

este plano **NO pasa por el origen**, por lo tanto **no es subespacio**.

Esto permite una discusión conceptual muy rica.

# 4. Definición de subespacio

Sea $V$ un espacio vectorial.

Un subconjunto

$$
W \subseteq V
$$

es subespacio si

1. $0 \in W$
2. Cerradura bajo suma
3. Cerradura bajo multiplicación escalar

---



# 5. Interpretación en ciencia de datos

Un subespacio representa:

* restricciones físicas
* dependencias entre variables
* redundancia de información

Ejemplo:

sensores altamente correlacionados.

---



# 6. Ejemplo fundamental

Considere

$$
W = {(x,y,z)\in \mathbb{R}^3 : x+y+z = 0}
$$

Este conjunto **sí es subespacio**.

Verificación conceptual:

Si

$$
x+y+z = 0
$$

y

$$
a+b+c = 0
$$

entonces

$$
(x+a)+(y+b)+(z+c) = 0
$$

---

# 7. Exploración computacional

```python
v = np.array([1,-1,0])
w = np.array([2,-2,0])

print(v.sum())
print(w.sum())
print((v+w).sum())
```

Discusión con estudiantes:

El conjunto es estable bajo suma.

---



# 8. Interpretación geométrica

Este conjunto es un **plano que pasa por el origen**.

En Machine Learning esto aparece cuando:

* las variables están centradas
* se eliminan medias
* se proyectan datos

---



# 9. Visualización

```python
import plotly.graph_objects as go
import numpy as np

x = np.linspace(-5,5,10)
y = np.linspace(-5,5,10)

X,Y = np.meshgrid(x,y)
Z = -X - Y

fig = go.Figure()

fig.add_surface(x=X,y=Y,z=Z,opacity=0.6)

fig.show()
```

Interpretación:

Este plano representa un **subespacio de datos posibles**.

---



# 10. Ejemplo real de datos agrícolas

Supongamos datos:

* fertilizante A
* fertilizante B
* fertilizante C

Pero por normativa

$$
A + B + C = dosis\ total
$$

Si centramos los datos restando la media:

$$
A+B+C = 0
$$

Los datos quedan en un **subespacio**.

Esto es exactamente lo que ocurre antes de aplicar **PCA**.

# 13. Conexión con reducción de dimensión

Si los datos viven en un plano dentro de $\mathbb{R}^{100}$:

la dimensión real es cercana a 2.

Esto motiva algoritmos como

* PCA
* SVD
* Autoencoders

---



# 14. Ejemplo simple de reducción

```python

X_reducido = pca.fit_transform(data)
```

Interpretación:

Encontramos el **subespacio que mejor explica los datos**.

# 15. Test práctico de subespacio

Para un conjunto

$$
W
$$

verificamos:

1.

$$
0 \in W
$$

2.

$$
u,v \in W \Rightarrow u+v \in W
$$

3.

$$
\alpha v \in W
$$

---



# 16. Ejemplo que NO es subespacio

Considere

$$
W = {(x,y)\in \mathbb{R}^2 : x+y = 1}
$$

Python:

```python
v = np.array([1,0])
w = np.array([0,1])

print(v+w)
```

Resultado

```
[1 1]
```

Pero

$$
1+1 \neq 1
$$

No hay cerradura.

---
