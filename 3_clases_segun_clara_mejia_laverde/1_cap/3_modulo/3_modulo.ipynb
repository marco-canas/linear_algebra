````python
# Veamos un ejemplo de subespacio generado por un conjunto de vectores. 
import numpy as np
import matplotlib.pyplot as plt

# Vector generador
v = np.array([1, -1])
# dominio de graficación
a,b = -5, 5
# Parámetro escalar t
t = np.linspace(a, b, 200)

# Subespacio generado: combinaciones escalares de v
x = t * v[0]
y = t * v[1]

# Crear figura
plt.figure(figsize=(6,6))

# Graficar la recta (subespacio generado)
plt.plot(x, y, label='Subespacio generado por (1, -1)', color = 'green')

# Graficar el vector generador
plt.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color = 'red', \
    label = 'vector generador (1,-1)')

# Ejes coordenados
plt.axhline(0)
plt.axvline(0)

# Configuración visual
plt.gca().set_aspect('equal', adjustable='box')
plt.xlim(a, b)
plt.ylim(a, b)
plt.xticks(np.arange(a,b,1))
plt.grid(True)
plt.legend()

# Título
plt.title('Subespacio generado por el vector (1, -1) en ℝ²')
plt.savefig(r"C:\Users\marco\Downloads\subespacio_generado_por_un_vector.png")
plt.show()
````

# Módulo 3: Definición y ejemplos de combinación lineal 

# Definición 1  


Sean $\mathbf{v}_{1}, \mathbf{v}_{2}, \ldots, \mathbf{v}_{n}$ vectores en un espacio vectorial $V$. Decimos que $\mathbf{x}$ es una combinación lineal de los vectores $\mathbf{v}_{1}, \mathbf{v}_{2}, \ldots, \mathbf{v}_{n}$ si existen escalares $a_{1}, a_{2}, \ldots, a_{n}$, tales que:



$$
\mathbf{x} = a_{1}\mathbf{v}_{1} + a_{2}\mathbf{v}_{2} + \ldots + a_{n}\mathbf{v}_{n}
$$
es decir, si existen escalares $a_{1}, a_{2}, \ldots, a_{n}$ tales que:
$$
\mathbf{x} = \sum_{i=1}^{n} a_{i}\mathbf{v}_{i}$$ 
$$

# Observaciones  

1. Toda combinación lineal de vectores de un espacio dado es otro vector del mismo espacio. Es decir, si combinas vectores de dos componentes, te da un vector de dos componente. 

2.  El módulo de un espacio vectorial es combinación lineal de cualquier conjunto de vectores del mismo espacio.  

$$
\mathbf{0} = 0\mathbf{v}_{1} + 0\mathbf{v}_{2} + \cdots + 0\mathbf{v}_{n} 
$$

3. Todo vector es combinación de sí mismo.  

$$
\mathbf{x} = 1\mathbf{x}
$$

1. La figura siguiente muestra un vector $\mathbf{v}$ en $\mathbb{R}^{2}$ o $\mathbb{R}^{3}$ como combinación lineal de los vectores $\mathbf{v}_{1}$ y $\mathbf{v}_{2}$.  

$$
\mathbf{v} = c_{1} \mathbf{v}_{1} + c_{2}\mathbf{v}_{2} 
$$

````python
# dame con matplotlib el grafico de dos vectores v1 y v1, 
# y el gráfico en el mismo plano de los múltiplos escalares c1v1 y c2v2,
# con escalares c1 y c2 positivos. Además inclute el trazo del vector 
# v = c1v1+c2v2 y el paralelogramo determinado por 
# c1v1 y c2v2  

import numpy as np
import matplotlib.pyplot as plt

# Definición de los vectores
v1 = np.array([2, 1])
v2 = np.array([1, 3])

# Escalares positivos
c1 = 2
c2 = 1.5

# Múltiplos escalares
c1v1 = c1 * v1
c2v2 = c2 * v2

# Vector resultante
v = c1v1 + c2v2

# Crear la figura
plt.figure(figsize=(4, 4))
plt.axhline(0)
plt.axvline(0)
plt.grid(True)

# Vectores originales
plt.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='blue')
plt.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='green')

# Múltiplos escalares
plt.quiver(0, 0, c1v1[0], c1v1[1], angles='xy', scale_units='xy', scale=1, color='cyan')
plt.quiver(0, 0, c2v2[0], c2v2[1], angles='xy', scale_units='xy', scale=1, color='lime')

# Vector suma
plt.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='red')

# Paralelogramo
x_par = [0, c1v1[0], v[0], c2v2[0], 0]
y_par = [0, c1v1[1], v[1], c2v2[1], 0]
plt.plot(x_par, y_par, 'k--')

# Etiquetas
plt.text(v1[0], v1[1]-0.5, r'$v_1$', color='blue')
plt.text(v2[0]-0.5, v2[1], r'$v_2$', color='green')
plt.text(c1v1[0]+0.4, c1v1[1], r'$c_1v_1$', color='cyan')
plt.text(c2v2[0]-0.7, c2v2[1], r'$c_2v_2$', color='lime')
plt.text(v[0], v[1], r'$v=c_1v_1+c_2v_2$', color='red')

# Ajustes finales
plt.axis('equal')
plt.xlim(0, 7)
plt.ylim(0, 7)
plt.title('Combinación lineal y paralelogramo generado')
plt.savefig(r"C:\Users\marco\Downloads\Combinación_lineal_y_paralelogramo_generado.png")
plt.show()
````

2.  Si  $\mathbf{a}$ y $\mathbf{b}$ son vectores libres, $Pr(a/b)$, la proyección de $a$ sobre $b$ es una combinación lineal de $b$ y $Pr(b/a)$ es una combinación lineal de $\mathbf{a}$.

````python
# traza dos vectores a y b y a las proyecciones de a sobre b y la proyección de b sobre a en un 
# mismo plano con matplotlib.  

import numpy as np
import matplotlib.pyplot as plt

# Definición de los vectores
a = np.array([4, 1])
b = np.array([2, 3])

# Producto punto
dot_ab = np.dot(a, b)

# Proyecciones
proj_a_on_b = (dot_ab / np.dot(b, b)) * b
proj_b_on_a = (dot_ab / np.dot(a, a)) * a

# Crear figura
plt.figure(figsize=(5, 5))
plt.axhline(0)
plt.axvline(0)
plt.grid(True)

# Vectores originales
plt.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color = 'red')
plt.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1, color = 'red')

# Proyecciones
plt.quiver(0, 0, proj_a_on_b[0], proj_a_on_b[1], angles='xy', scale_units='xy', scale=1, color = 'purple')
plt.quiver(0, 0, proj_b_on_a[0], proj_b_on_a[1], angles='xy', scale_units='xy', scale=1, color = 'purple')

# Segmentos auxiliares (perpendiculares)
plt.plot([a[0], proj_a_on_b[0]], [a[1], proj_a_on_b[1]], '--')
plt.plot([b[0], proj_b_on_a[0]], [b[1], proj_b_on_a[1]], '--')

# Etiquetas
plt.text(a[0], a[1], r'$a$', fontsize=12)
plt.text(b[0], b[1], r'$b$', fontsize=12)
plt.text(proj_a_on_b[0]-1.3, proj_a_on_b[1] - 0.2, r'$\mathrm{proj}_b(a)$', fontsize=12)
plt.text(proj_b_on_a[0], proj_b_on_a[1]-0.2, r'$\mathrm{proj}_a(b)$', fontsize=12)

# Ajustes visuales
plt.axis('equal')
plt.xlim(-1, 5)
plt.ylim(-1, 5)
plt.title('Proyección de vectores en el plano')
plt.savefig(r"C:\Users\marco\Downloads\Proyección_de_vectores_en_el_plano.png")
plt.show()
````

4. En $\mathbb{R}^{2}$, $\mathbf{v}_{1} = (2,1)$ y $\mathbf{v}_{2} = (-4, -2)$.. Veamos si $(6,3)$ es una combinación lineal de $\mathbf{v}_{1}$ y $\mathbf{v}_{2}$.  

Debemos entonces determinar si existen escalares $a_{1}$ y $a_{2}$ tales que 

$$v = a_{1}v_{1} + a_{2}v_{2} $$

$$(6,3) = a_{1}(2,1) + a_{2}(-4, -2)$$



$$
(6,3) = (2a_{1}, a_{1}) + (-4a_{2}, -2a_{2}) 
$$

$$
(6,3) = (2a_{1} - 4a_{2}, a_{1} - 2a_{2}) 
$$


Luego

$$
\begin{align*}
6 & = 2a_{1} - 4a_{2} \\
3 & = a_{1} - 2a_{2}  
\end{align*}
$$

$$
\begin{bmatrix}  2 & -4 \\ 1 & -2  \end{bmatrix}\begin{bmatrix}a_{1} \\ a_{2} \end{bmatrix} = \begin{bmatrix} 6 \\ 3 \end{bmatrix}
$$

La matriz aumentada es

$$
\begin{bmatrix} 2 & -4 & 6 \\ 1 & -2 & 3\end{bmatrix} \to \begin{bmatrix} 1 & -2 & 3 \\ 0 & 0 & 0\end{bmatrix} 
$$

El sistema tiene infinitas soluciones dadas por $a_{1} = 3 + 2a_{2}$, con $a_{2} \in \mathbb{R}$. 

 
Por ejemplo, si $a_{2} = 1$ y $a_{1} = 5$ y $(6,3) = 5(2,1) + (1)(-4,-2)$; pero si hacemos $a_{2} = 0$ y $a_{1} = 3$ $(6,3) = 3(2,1) + 0(4, 2)$, etc.  

De modo que $v$ es una combinación lineal de $v_{1}$ y $v_{2}$ de muchas maneras.


Podemos entonces observar que si un vector $v$ es una combinación lineal $v_{1}, v_{2}, \ldots, v_{k}$, entonces esa combinación lineal **puede no ser única**. 

````python
import numpy as np
import matplotlib.pyplot as plt

# =========================
# Definición de vectores
# =========================
v1 = np.array([2, 1])
v2 = np.array([-4, -2])
w  = np.array([6, 3])   # vector a verificar

# Escalar que muestra la combinación lineal
a = 3
v1_scaled = a * v1

# =========================
# Gráfico
# =========================
plt.figure()

origen = np.array([0, 0])

# vectores base
plt.quiver(*origen, *v1, angles='xy', scale_units='xy', scale=1)
plt.text(*v1, "v1=(2,1)")

plt.quiver(*origen, *v2, angles='xy', scale_units='xy', scale=1)
plt.text(*v2, "v2=(-4,-2)")

# vector resultado
plt.quiver(*origen, *w, angles='xy', scale_units='xy', scale=1)
plt.text(*w, "(6,3)")

# vector como múltiplo de v1
plt.quiver(*origen, *v1_scaled, angles='xy', scale_units='xy', scale=1)
plt.text(*(v1_scaled + np.array([0.2,0.2])), "3v1")

# recta generada por v1 y v2
t = np.linspace(-3, 3, 100)
recta = np.outer(t, v1)
plt.plot(recta[:,0], recta[:,1])

# ajustes visuales
plt.axhline(0)
plt.axvline(0)
plt.gca().set_aspect('equal', adjustable='box')
plt.title("Combinación lineal en R^2")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.savefig(r"C:\Users\marco\Downloads\combinacion_lineal_vectores_paralelos.png")
plt.show()
````

5.  En $\mathbb{P}_{3}$ sea $A = \{-1+x, 1 + x^{2}, 2x + x^{3}\}$. Veamos que $Q(x) = 3 + x^{2} + x^{3}$ es una combinación lineal de $A$.

Expresemos a $Q(x)$ como una combinación lineal de los polinomios de $A$.  

$$ 
3 + x^{2} + x^{3} = a_{1}(-1+x) + a_{2}(1+x^{2}) + a_{3}(2x + x^{3}) 
$$

$$ 
3 + x^{2} + x^{3} = -a_{1} + a_{1}x + a_{2} + a_{2}x^{2} + 2a_{3}x + a_{3}x^{3}
$$



$$
3 + x^{2} + x^{3} = (-a_{1} + a_{2}) + (a_{1} + 2a_{3})x + (a_{2})x^{2} + (a_{3})x^{3}
$$

Igualando los coeficientes de los términos del mismo grado en ambos lados de la igualdad tenemos:  

$$
\begin{align}
3 & = -a_{1} + a_{2} \\
0 & = a_{1} + 2a_{3}  \\  
1 & = a_{2} \\  
1 & a_{3}
\end{align}
$$

Reemplazando en la primera ecuación que $a_{2} = 1$, obtenemos que $a_{1} = -2$. 

Por tanto, hemos determinado escalares $a_{1}, a_{2}, a_{3}$ tales que:

$$ 
3 + x^{2} + x^{3} = -2(-1+x) + (1)(1+x^{2}) + (1)(2x + x^{3}) 
$$

````python
# Representación de esta combinación lineal de polinomios de P3  

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

# =========================
# Definición de polinomios
# =========================
def p1(x):   # -1 + x
    return -1 + x

def p2(x):   # 1 + x^2
    return 1 + x**2

def p3(x):   # 2x + x^3
    return 2*x + x**3

def Q(x):    # 3 + x^2 + x^3
    return 3 + x**2 + x**3

# dominio para graficar
x = np.linspace(-3, 3, 400)

# =========================
# Función interactiva
# =========================
def visualizar(a, b, c):
    suma = a*p1(x) + b*p2(x) + c*p3(x)

    plt.figure(figsize=(7,5))

    # múltiplos escalares
    plt.plot(x, a*p1(x), label=f"{a}(-1+x)")
    plt.plot(x, b*p2(x), label=f"{b}(1+x²)")
    plt.plot(x, c*p3(x), label=f"{c}(2x+x³)")

    # suma de múltiplos
    plt.plot(x, suma, linestyle='--', linewidth=2,
             label="Suma de los múltiplos")

    # polinomio Q
    plt.plot(x, Q(x), linewidth=3,
             label="Q(x)=3+x²+x³")

    plt.axhline(0)
    plt.axvline(0)
    plt.grid(True)
    plt.legend()
    plt.title("Combinaciones lineales en P₃")
    plt.xlabel("x")
    plt.ylabel("y")

    # verificación algebraica mostrada en pantalla
    print("Coeficientes elegidos:", a, b, c)
    if a == -3 and b == 1 and c == 1:
        print("✓ Con estos valores se obtiene exactamente Q(x).")

    plt.show()

# =========================
# Deslizadores enteros
# =========================
interact(
    visualizar,
    a=widgets.IntSlider(min=-3, max=3, step=1, value=0, description="a"),
    b=widgets.IntSlider(min=-3, max=3, step=1, value=0, description="b"),
    c=widgets.IntSlider(min=-3, max=3, step=1, value=0, description="c"),
);
````

# 3.3  Espacio generado por un conjunto de vectores 



# Definición 3  

Sean $v_{1}, v_{2}, \ldots, v_{n}$ $n$ vectores en un espacio vectorial $V$.   

El espacio generado por $\{v_{1}, v_{2}, \ldots, v_{n}\}$ es el conjunto de las combinaciones lineales de $v_{1}, v_{2}, \ldots, v_{n}$. Esto es 

$$
gen(\{v_{1}, v_{2}, \ldots, v_{n}\}) = \{v \mid v = a_{1}v_{1} + a_{2}v_{2}+ \cdots + a_{n}v_{n}, \quad a_{1}, a_{2}, \ldots, a_{n} \in \mathbb{R} \}
$$

Del conjunto $\{\mathbf{v}_{1}, v_{2}, \ldots, v_{n}\}$ se dice que es el generador. 

## Teorema 1 

Si $v_{1}, v_{2}, \ldots, v_{n}$ son $n$ vectores en un espacio vectorial $V$. El $gen \{v_{1}, v_{2}, \ldots, v_{n}\}$ es un subespacio vectorial de $V$.  

La demostración de este teorema se deja como ejercicio. 

# Ejemplos

9. En $\mathbb{R}$, sea $A = \{1\}$. 

$$
gen(A) = \{x \mid x = a(1), a \in \mathbb{R}\} = \mathbb{R}.
$$ 

$\{1\}$ es generador para $\mathbb{R}$. 

10. En $\mathbb{R}^{3}$, sea $A = \{(1, -2, 0), (3, 1, -1) \}$.  

    $gen(A) = \{ (x,y,z) \mid (x,y,z) = a_{1}(-1, 2, 0) + a_{2}(3, 1, -1), a_{1}, a_{2} \in \mathbb{R}\}$ 

Determinemos una relación entre $x, y$, $z$.  

$$
(x, y, z) = (a_{1} + 3_{2}, -2a_{1} + a_{2}, -a_{2} ),
$$



$$
\begin{align}
x & = a_{1} + 3a_{2} \\ 
y & = -2a_{1} + a_{2} \\
z & = -a_{2}
\end{align}
$$

$$
\begin{bmatrix}1&3&|&x\\
-2&1&|&y\\  
0&-1&|&z\end{bmatrix}
$$
Apliquemos el algoritmo de Gauss Jordan para obtener  

$$
\begin{bmatrix}
1&3&|&x\\
0&1&|&\frac{y+2x}{7}\\  
0&0&|&-z - \frac{y+2x}{7}
\end{bmatrix}
$$

Luego los elementos de $gen(A)$ deben satisfacer

$$
-z - \frac{y+2x}{7} = 0
$$

o sea 

$$
2x + y + 7z = 0
$$

$gen(A) = \{(x,y,z): 2x + y + 7z = 0 \}$ (plano que pasa por el origen).

# Teorema 2  

Sean $\mathbf{v}_{1}, v_{2}, \ldots, v_{n}, v_{n+1}$, $n+1$ vectores que están en un espacio vectorial $V$.    

Si $v_{1}, v_{2}, \ldots, v_{n}$ genera a $V$, entonces $v_{1}, v_{2}, \ldots, v_{n}, v_{n+1}$ también genera a $V$.    

Esto es, la adición de uno o más vectores a un conjunto generador de $V$ da por resultado otro conjunto generador de $V$.

## Ejemplo 

El espacio vectorial $\mathbb{R}^{2}$ puede ser generado por el conjunto $A = \left\{ \begin{pmatrix} 1 \\ 0 \end{pmatrix}, \begin{pmatrix} 0 \\ 1 \end{pmatrix}  \right\}$ 
También puede ser generado por un conjunto mayor $A' = \left\{ \begin{pmatrix} 1 \\ 0 \end{pmatrix}, \begin{pmatrix} 0 \\ 1 \end{pmatrix}, \begin{pmatrix} 1 \\ 1 \end{pmatrix}  \right\} $

````python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# Vector base fijo
v = np.array([2, 1])

# Puntos fijos que representan el subespacio (la recta)
t_linea = np.linspace(-10, 10, 50)
recta = np.outer(t_linea, v)

def actualizar_escalar(t):
    plt.figure(figsize=(4, 4))
    
    # 1. Dibujar la recta (subespacio generado) como puntos de referencia
    plt.scatter(recta[:, 0], recta[:, 1], color='gray', alpha=0.3, s=10, label='Subespacio (Span)')
    
    # 2. Dibujar el vector base original v (fijo)
    plt.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, 
               color='blue', label='Vector base v', zorder=3)
    
    # 3. Dibujar el vector múltiplo escalar t*v
    vt = t * v
    plt.quiver(0, 0, vt[0], vt[1], angles='xy', scale_units='xy', scale=1, 
               color='red', width=0.01, label=f'Vector t*v (t={t:.1f})', zorder=2)
    
    # Configuraciones de la gráfica
    plt.axhline(0, color='black', linewidth=1)
    plt.axvline(0, color='black', linewidth=1)
    plt.grid(True, linestyle='--', alpha=0.5)
    
    # Límites constantes para que la escala no salte al mover el slider
    plt.xlim(-10, 10)
    plt.ylim(-10, 10)
    
    plt.legend(loc='upper left')
    plt.title(f"Múltiplo escalar de un vector: $v' = {t} \cdot [2, 1]$")
    plt.show()

# Crear el deslizador para el escalar t
interact(actualizar_escalar, 
         t=FloatSlider(value=1.0, min=-5.0, max=5.0, step=0.1, description='Escalar t:'))
````

````python
import numpy as np
import plotly.graph_objects as go

# Definir vectores en R3
v1 = np.array([2, 1, 0])
v2 = np.array([1, 3, 0])

# Valores de escalares
a_vals = np.linspace(-2, 2, 21)
b_vals = np.linspace(-2, 2, 21)

# Inicial
a0, b0 = 1, 1
w0 = a0 * v1 + b0 * v2

# Crear figura
fig = go.Figure()

# v1
fig.add_trace(go.Scatter3d(
    x=[0, v1[0]], y=[0, v1[1]], z=[0, v1[2]],
    mode='lines+markers',
    name='v1',
    line=dict(width=6)
))

# v2
fig.add_trace(go.Scatter3d(
    x=[0, v2[0]], y=[0, v2[1]], z=[0, v2[2]],
    mode='lines+markers',
    name='v2',
    line=dict(width=6)
))

# w inicial
fig.add_trace(go.Scatter3d(
    x=[0, w0[0]], y=[0, w0[1]], z=[0, w0[2]],
    mode='lines+markers',
    name='w = a v1 + b v2',
    line=dict(width=6, dash='dash')
))

# Crear frames
frames = []
for i, a in enumerate(a_vals):
    for j, b in enumerate(b_vals):
        w = a * v1 + b * v2
        frames.append(go.Frame(
            data=[
                go.Scatter3d(x=[0, v1[0]], y=[0, v1[1]], z=[0, v1[2]]),
                go.Scatter3d(x=[0, v2[0]], y=[0, v2[1]], z=[0, v2[2]]),
                go.Scatter3d(x=[0, w[0]], y=[0, w[1]], z=[0, w[2]])
            ],
            name=f"a={i},b={j}"
        ))

fig.frames = frames

# Sliders
steps_a = []
for i, a in enumerate(a_vals):
    steps_a.append(dict(
        method="animate",
        args=[[f"a={i},b=10"],
              dict(mode="immediate", frame=dict(duration=0, redraw=True))],
        label=f"{round(a,1)}"
    ))

steps_b = []
for j, b in enumerate(b_vals):
    steps_b.append(dict(
        method="animate",
        args=[[f"a=10,b={j}"],
              dict(mode="immediate", frame=dict(duration=0, redraw=True))],
        label=f"{round(b,1)}"
    ))

# Layout 3D (CLAVE para rotar)
fig.update_layout(
    title="Combinaciones lineales en R³ (puedes rotar con el mouse)",
    scene=dict(
        xaxis=dict(range=[-10, 10]),
        yaxis=dict(range=[-10, 10]),
        zaxis=dict(range=[-10, 10]),
        aspectmode='cube'  # proporciones iguales
    ),
    sliders=[
        dict(
            active=10,
            currentvalue={"prefix": "a: "},
            pad={"t": 50},
            steps=steps_a
        ),
        dict(
            active=10,
            currentvalue={"prefix": "b: "},
            pad={"t": 100},
            steps=steps_b
        )
    ]
)

fig.show()
````

# 9. Definiciones (Látex)

* **Combinación Lineal:** Sean $v_1, v_2, \dots, v_n$ vectores en un espacio vectorial $V$. Un vector $w \in V$ es una combinación lineal de estos si existen escalares $c_1, c_2, \dots, c_n$ tales que:
$$w = c_1v_1 + c_2v_2 + \dots + c_nv_n = \sum_{i=1}^{n} c_iv_i$$

* **Conjunto Generador:** Se dice que los vectores $\{v_1, v_2, \dots, v_n\}$ generan a $V$ si cada vector en $V$ puede expresarse como una combinación lineal de ellos.

* **Espacio Generado (Span):** Es el conjunto de todas las combinaciones lineales posibles de un conjunto de vectores $S$:
$$gen(S) = \{ \sum_{i=1}^{k} c_iv_i \mid v_i \in S, c_i \in \mathbb{R} \}$$

### 11. Teoremas
* **Teorema de Existencia de Soluciones:** Un vector $b$ está en el espacio generado por las columnas de una matriz $A$ si y solo si el sistema $Ax = b$ es consistente.
* **Teorema de Independencia Lineal:** Si el conjunto de vectores es linealmente independiente y el número de vectores es igual a la dimensión del espacio ($n=3$ en nuestro caso), entonces generan todo el espacio $\mathbb{R}^3$.
