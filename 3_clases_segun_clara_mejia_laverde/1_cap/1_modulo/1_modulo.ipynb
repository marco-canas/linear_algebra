## Las fuerzas que pueden actuar sobre un punto

Consideremos un punto material $P$ y las fuerzas que pueden actuar sobre él. 

Si $F_{1}$ y $F_{2}$ son dos fuerzas que actúan sobre $P$, se puede encontrar una fuerza $F$ resultante de $F_{1}$ y $F_{2}$ con la propiedad de ejercer sobre $P$ la misma acción que $F_{1}$ y $F_{2}$  actuando simultáneamente. 

Así que tenemos una fuerza $F$, que resulta de sumar $F_{1}$ y $F_{2}$, ($F = F_{1} + F_{2}$).

Veamos qué propiedades tiene esta operación.  


Representemos $F_{1}$ y $F_{2}$  por vectores cuya dirección y sentido corresponden a los de las fuerzas dadas y su longitud (en una escala determinada) mide la magnitud; entonces, la resultante queda determinada por la diagonal del paralelogramo, dos de cuyos lados son los segmentos correspondientes a $F_{1}$ y $F_{2}$
(ver figura siguiente).

```python
import numpy as np
import matplotlib.pyplot as plt

# Definición de las fuerzas como vectores
F1 = np.array([4, 1])   # Fuerza F1
F2 = np.array([2, 3])   # Fuerza F2

# Resultante (diagonal del paralelogramo)
R = F1 + F2

# Origen
O = np.array([0, 0])

# Crear la figura
plt.figure(figsize=(6, 6))

# Dibujar los vectores F1 y F2 desde el origen
plt.quiver(*O, *F1, angles='xy', scale_units='xy', scale=1, color='blue', label='$F_1$')
plt.quiver(*O, *F2, angles='xy', scale_units='xy', scale=1, color='green', label='$F_2$')

# Dibujar los lados del paralelogramo
plt.plot([F1[0], R[0]], [F1[1], R[1]], 'k--', linewidth=1)
plt.plot([F2[0], R[0]], [F2[1], R[1]], 'k--', linewidth=1)
plt.scatter(*O, color='black', s = 100)  # Punto de origen

# Dibujar la resultante
plt.quiver(*O, *R, angles='xy', scale_units='xy', scale=1, color='red', label='Resultante')

# Etiquetas
plt.text(F1[0]+0.3, F1[1], '$F_1$', fontsize=12)
plt.text(F2[0]-0.3, F2[1], '$F_2$', fontsize=12)
plt.text(R[0], R[1], '$F$', fontsize=12)

# Configuración del gráfico
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.grid(True)
plt.axis('equal')
plt.xlim(0, 7)
plt.ylim(0, 6)
plt.legend()
plt.title('Suma de fuerzas: método del paralelogramo')
plt.savefig(r"C:\Users\marco\Downloads\suma_fuerzas_paralelogramo.png")
plt.show()
```

Sea $T$ el conjunto de todas las fuerzas que actúan sobre $P$; entonces, podemos decir que:  


1. Si $F_{1}$ y $F_{2}$ pertenecen a $T$, entonces $F_{1} + F_{2} = F$ pertenece a $T$. 

A esta propiedad la llamaremos Clausuratividad o Cerradura de la adición en $T$:

2. Asociatividad de la adición en $T$ 
   $$
   (F_{1} + F_{2}) + F_{3} = F_{1} + (F_{2} + F_{3}) 
   $$

```python
import numpy as np
import matplotlib.pyplot as plt

# Definición de los vectores
F1 = np.array([3, 1])
F2 = np.array([1, 3])
F3 = np.array([2, -1])

# Sumas parciales
S12 = F1 + F2          # F1 + F2
S23 = F2 + F3          # F2 + F3

# Resultante total
F = F1 + F2 + F3

# Origen
O = np.array([0, 0])

plt.figure(figsize=(7, 7))

# --- Vectores base ---
plt.quiver(*O, *F1, angles='xy', scale_units='xy', scale=1,
           color='blue', label='$F_1$')

plt.quiver(*F1, *F2, angles='xy', scale_units='xy', scale=1,
           color='green', label='$F_2$')

plt.quiver(*(F1 + F2), *F3, angles='xy', scale_units='xy', scale=1,
           color='purple', label='$F_3$')

# --- Suma F1 + F2 desde el origen ---
plt.quiver(*O, *S12, angles='xy', scale_units='xy', scale=1,
           color='orange', linestyle='dashed', linewidth=2,
           label='$F_1 + F_2$')

# --- Suma F2 + F3 aplicada en el extremo de F1 ---
plt.quiver(*F1, *S23, angles='xy', scale_units='xy', scale=1,
           color='brown', linestyle='dashed', linewidth=2,
           label='$F_2 + F_3$ (desde $F_1$)')

# --- Resultante final ---
plt.quiver(*O, *F, angles='xy', scale_units='xy', scale=1,
           color='red', linewidth=3, label='Resultante $F$')

# Etiquetas
plt.text(F1[0], F1[1]-0.2, '$F_1$', fontsize=11)
plt.text((F1 + F2 / 2)[0]+0.2, (F1 + F2 / 2)[1], '$F_2$', fontsize=11)
plt.text(*(F1 + F2 + F3 / 2), '$F_3$', fontsize=11)
plt.text(*S12, '$F_1+F_2$', fontsize=11)
plt.text(*(F1 + S23), '$F_2+F_3$', fontsize=11)
plt.text(*F, '$F$', fontsize=12)

# Configuración del gráfico
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.grid(True)
plt.axis('equal')
plt.xlim(-1, 8)
plt.ylim(-1, 6)
plt.legend()
plt.title('Asociatividad de la suma de vectores\n$(F_1 + F_2) + F_3 = F_1 + (F_2 + F_3)$')

plt.savefig(r"C:\Users\marco\Downloads\asociatividad_suma_vectores.png")
plt.show()
```

3. Modulatividad de la adición en $T$: Por otra parte, existe una fuerza de magnitud cero, fuerza nula, que podemos designar $\mathbf{0}$ y tiene la propiedad de que:

$$
F+\mathbf{0} = 0 + F = F
$$

Es decir, la adición en $T$ tiene un elemento neutro, la fuerza nula.

4. Ahora, si llamamos la opuesta de $F$ a una fuerza $F'$ que tiene la misma magnitud y dirección que $F$ pero sentido contrario, encontramos que la resultante de  $F$ y  $F'$ es cero, es decir:  



Para cada $F$ en $T$ existe ,$F'$ tal que $F + F' = 0$. Esta $F'$ se llamará $-F$.  

A esta propiedad de la adición en $T$ la llamaremos existencia del inverso aditivo o invertivilidad de la adición en $T$.  

```python
import numpy as np
import matplotlib.pyplot as plt

# Definición de la fuerza
F = np.array([4, 2])     # Fuerza F
minus_F = -F             # Inverso aditivo de F

# Origen
O = np.array([0, 0])

plt.figure(figsize=(6, 6))

# Vector F
plt.quiver(*O, *F, angles='xy', scale_units='xy', scale=1,
           color='blue', linewidth=2, label='$F$')

# Vector -F
plt.quiver(*O, *minus_F, angles='xy', scale_units='xy', scale=1,
           color='red', linewidth=2, label='$-F$')

# Etiquetas
plt.text(*F, '$F$', fontsize=12)
plt.text(minus_F[0] , minus_F[1] - 0.2, '$-F$', fontsize=12)

# Configuración del plano
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.grid(True)
plt.axis('equal')
plt.xlim(-5, 5)
plt.ylim(-5, 5)
plt.legend()
plt.title('Fuerza $F$ y su inverso aditivo $-F$')

plt.savefig(r"C:\Users\marco\Downloads\fuerza_y_su_inverso_aditivo.png")
plt.show()
```

5. Por último, es claro que el orden en que se consideran las fuerzas no altera la
resultante, esto es:  

$$
F_{1} + F_{2} = F_{2} + F_{1}
$$

A esta propiedad de la adición en $T$ la llamaremos conmutatividad de la adición en $T$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Definición de las fuerzas
F1 = np.array([4, 1])
F2 = np.array([1, 3])

# Resultante común
R = F1 + F2

# Origen
O = np.array([0, 0])

plt.figure(figsize=(7, 7))

# --- Recorrido 1: F1 + F2 ---
plt.quiver(*O, *F1, angles='xy', scale_units='xy', scale=1,
           color='blue', linewidth=2, label='$F_1$')

plt.quiver(*F1, *F2, angles='xy', scale_units='xy', scale=1,
           color='green', linewidth=2, label='$F_2$')

# --- Recorrido 2: F2 + F1 ---
plt.quiver(*O, *F2, angles='xy', scale_units='xy', scale=1,
           color='green', linestyle='dashed', linewidth=2)

plt.quiver(*F2, *F1, angles='xy', scale_units='xy', scale=1,
           color='blue', linestyle='dashed', linewidth=2)

# --- Resultante ---
plt.quiver(*O, *R, angles='xy', scale_units='xy', scale=1,
           color='red', linewidth=3, label='Resultante')

# Etiquetas
plt.text(*F1, '$F_1$', fontsize=11)
plt.text(*F2, '$F_2$', fontsize=11)
plt.text(*R, '$R$', fontsize=12)

# Configuración del gráfico
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.grid(True)
plt.axis('equal')
plt.xlim(-1, 7)
plt.ylim(-1, 6)
plt.legend()
plt.title('Conmutatividad de la suma de fuerzas\n$F_1 + F_2 = F_2 + F_1$')

plt.savefig(r"C:\Users\marco\Downloads\conmutatividad_suma_fuerzas.png")
plt.show()
```

Veamos ahora otra operación: si $F$ es una fuerza y $\alpha$ un número real, llamamos $\alpha F$ la fuerza que tiene la misma dirección de $F$, su magnitud es $\alpha$ veces la de $F$ y su sentido es el mismo de $F$ si $\alpha$ es positivo, y opuesto a $F$ si $\alpha$ es negativo.  


Hemos entonces definido una operación cuyos «factores» son un número real y una fuerza y su resultado es otra fuerza que actua sobre el punto $P$ del plano dado. 

Esta operación se dice que es externa ya que hace intervenir un elemento (el número $\alpha$) que no pertene
ce al conjunto $T$.


Veamos algunas propiedades de esta operación

6. Si $\alpha$ es un número real y $F \in T$, entonces $\alpha F \in T$ .

7. $\alpha(F_{1} + F_{2}) = \alpha F_{1} + \alpha F_{2}$. 

8. $(\alpha + \beta)F  = \alpha F + \beta F$

9. $\alpha(\beta F) = (\alpha \beta)F$

10. $1F = F$

# El espacio vectorial $\mathbb{R}^{2}$

Consideremos el conjunto $\mathbb{R}^{2} = \{(x,y) : x,y \in \mathbb{R}\}$, junto con las operaciones:

1. Suma de vectores: $(x_{1},y_{1}) + (x_{2},y_{2}) = (x_{1}+x_{2},\, y_{1}+y_{2})$.
2. Multiplicación por escalar: $\alpha(x,y) = (\alpha x, \alpha y)$, donde $\alpha \in \mathbb{R}$.


Verificaremos que $\mathbb{R}^{2}$ satisface los diez axiomas que definen un espacio vectorial sobre $\mathbb{R}$.



## Axioma 1: Cerradura bajo la suma





Si $u=(x_{1},y_{1})$ y $v=(x_{2},y_{2})$ pertenecen a $\mathbb{R}^{2}$, entonces

$$
u+v=(x_{1}+x_{2},\, y_{1}+y_{2}) \in \mathbb{R}^{2}.
$$

Como la suma de números reales es un número real, el resultado pertenece a $\mathbb{R}^{2}$.



## Axioma 2: Conmutatividad de la suma


### Axioma 2: Conmutatividad de la suma

Para $u=(x_{1},y_{1})$ y $v=(x_{2},y_{2})$,

$$
u+v=(x_{1}+x_{2},\, y_{1}+y_{2})=(x_{2}+x_{1},\, y_{2}+y_{1})=v+u.
$$




## Axioma 3: Asociatividad de la suma


### Axioma 3: Asociatividad de la suma



Para $u,v,w \in \mathbb{R}^{2}$,

$$
(u+v)+w=u+(v+w).
$$

Esto se cumple porque la suma de números reales es asociativa.

$$



## Axioma 4: Existencia del vector cero


### Axioma 4: Existencia del vector cero

Existe el vector $0=(0,0)$ tal que para todo $u=(x,y)$,

$$
\mathbf{u}+\mathbf{0}=(x,y)+(0,0)=(x+0, y+0)=(x,y)=\mathbf{u}.
$$


---



## Axioma 5: Existencia del inverso aditivo


### Axioma 5: Inverso aditivo

Para cada $u=(x,y)$ existe $-u=(-x,-y)$ tal que

$$
u+(-u)=(x,y)+(-x,-y)=(0,0) = \mathbf{0}.
$$






## Axioma 6: Cerradura bajo multiplicación por escalares


### Axioma 6: Cerradura bajo multiplicación por escalares

Si $\alpha \in \mathbb{R}$ y $u=(x,y) \in \mathbb{R}^{2}$,

$$
\alpha u=(\alpha x, \alpha y) \in \mathbb{R}^{2}.
$$




## Axioma 7: Distributividad respecto a la suma de vectores


### Axioma 7: Distributividad respecto a la suma de vectores

Para $\alpha \in \mathbb{R}$ y $u = (x_1,y_{1}), v = (x_2,y_{2}) \in \mathbb{R}^{2}$,

$$
\alpha(u+v)=\alpha u+\alpha v.
$$

Esto lo mostramos así: 

$$
\begin{align}
\alpha(u+v) & = \alpha[(x_1,y_{1}), (x_{2}, y_{2}) ] \\  
            & = \alpha (x_{1} + x_{2}, y_{1} + y_{2}) \\ 
            & = (\alpha(x_{1} + x_{2}, \alpha(y_{1} + y_{2}))) \\ 
            & = (\alpha x_{1} + \alpha x_{2}, \alpha y_{1} + \alpha y_{2} ) \\  
            & = (\alpha x_{1}, \alpha y_{1}) + ((\alpha x_{2}, \alpha y_{2})) \\ 
            & = \alpha (x_{1}, y_{1}) + \alpha (x_{1}, y_{2})  \\  
            & = \alpha u + \alpha v
\end{align}
$$




## Axioma 8: Distributividad respecto a la suma de escalares


## Axioma 8: Distributividad respecto a la suma de escalares

Para $\alpha, \beta \in \mathbb{R}$ y $u = (x,y) \in \mathbb{R}^{2}$,

$$
(\alpha+\beta)u=\alpha u+\beta u.
$$




## Axioma 9: Compatibilidad del producto escalar


### Axioma 9: Compatibilidad del producto escalar

Para $\alpha,\beta \in \mathbb{R}$ y $u \in \mathbb{R}^{2}$,

$$
\alpha(\beta u)=(\alpha\beta)u.
$$






## Axioma 10: Identidad multiplicativa


### Axioma 10: Identidad multiplicativa

Para todo $u = (x,y) \in \mathbb{R}^{2}$,

$$
1u=u.
$$

En efecto, 

$$
1u = 1(x,y) = (1x, 1y) = (x, y) = u.
$$  


Por lo tanto, $\mathbb{R}^{2}$ con estas operaciones satisface todos los axiomas de espacio vectorial sobre $\mathbb{R}$.

# 1.2 Definición de espacio vectorial  



Los ejemplos anteriores nos sirven para obtener un conjunto de propiedades comunes a ellos. 

Todos los resultados que podamos deducir a partir de dichas propiedades serán válidos en los ejemplos realizados y en cualquier otro que goce de las mismas propiedades.


A estas propiedades comunes las llamamos axiomas en la siguiente definición.


Un espacio vectorial $V$ es un conjunto de objetos llamados vectores con dos operaciones, suma y producto por un escalar, que satisface los siguientes axiomas:    

Sea $V$ un conjunto no vacío y $\mathbb{K}$ un cuerpo (por ejemplo, $\mathbb{R}$ o $\mathbb{C}$).   

Decimos que $V$ es un **espacio vectorial sobre $\mathbb{K}$** si en $V$ están definidas dos operaciones:

* una suma de vectores $+: V \times V \to V$,
* una multiplicación por escalares $\cdot: \mathbb{K} \times V \to V$,

tales que para todos $u, v, w \in V$ y para todos $\alpha,\beta \in \mathbb{K}$ se satisfacen los siguientes axiomas:

1. (Cerradura bajo la suma)
   $u+v \in V$.

2. (Conmutatividad de la suma)
   $u+v = v+u$.

3. (Asociatividad de la suma)
   $(u+v)+w = u+(v+w)$.

4. (Elemento neutro aditivo)
   Existe un vector $0 \in V$ tal que
   $u+0=u$ para todo $u \in V$.

5. (Inverso aditivo)
   Para cada $u \in V$ existe $-u \in V$ tal que
   $u+(-u)=0$.

6. (Cerradura bajo multiplicación por escalares)
   $\alpha u \in V$.

7. (Distributividad respecto a la suma de vectores)
   $$
   \alpha(u+v)=\alpha u+\alpha v.
   $$

8. (Distributividad respecto a la suma de escalares)
   $$
   (\alpha + \beta)u = \alpha u + \beta u.
   $$

9. (Compatibilidad del producto por escalares)
   $$
   \alpha(\beta u)=(\alpha\beta)u.
   $$

10. (Elemento identidad del producto escalar)
    $$
    1u=u.
    $$

Si todas estas propiedades se cumplen, el par $(V,+,\cdot)$ se denomina espacio vectorial sobre el cuerpo $\mathbb{K}$.


# Nota:    

cuando los escalares son números reales, decimos que se trata de un espacio vectorial real, y en caso de que sean números complejos hablamos de un espacio vectorial complejo.


En el axioma 8. debe notarse que el signo + al lado izquierdo de la igualdad se refiere a la suma en el conjunto de escalares, y el signo + al lado derecho denota la suma en el conjunto de vectores $V$.

1. $\mathbb{R}^{n}$ con las operaciones de suma y producto por un escalar definidas usualmente, tiene estructura de espacio vectorial.  

## Demostración de que $\mathbb{R}^{n}$ es un espacio vectorial

Sea $\mathbb{R}^{n}={(x_{1},x_{2},\ldots,x_{n}) : x_{i}\in\mathbb{R}}$, junto con las operaciones usuales:

* suma de vectores
  $$
  (x_{1},\ldots,x_{n})+(y_{1},\ldots,y_{n})=(x_{1}+y_{1},\ldots,x_{n}+y_{n}),
  $$

* multiplicación por escalar
  $$
  \alpha(x_{1},\ldots,x_{n})=(\alpha x_{1},\ldots,\alpha x_{n}),\quad \alpha\in\mathbb{R}.
  $$

Demostraremos que $\mathbb{R}^{n}$ satisface los axiomas de espacio vectorial sobre $\mathbb{R}$.

## 1. Cerradura bajo la suma

Si $u=(x_{1},\ldots,x_{n})$ y $v=(y_{1},\ldots,y_{n})$ pertenecen a $\mathbb{R}^{n}$, entonces

$$
u+v=(x_{1}+y_{1},\ldots,x_{n}+y_{n}).
$$

Como la suma de números reales es un número real, se tiene $u+v\in\mathbb{R}^{n}$.



\subsection*{2. Conmutatividad de la suma}

$$
u+v=(x_{1}+y_{1},\ldots,x_{n}+y_{n})=(y_{1}+x_{1},\ldots,y_{n}+x_{n})=v+u,
$$

porque la suma en $\mathbb{R}$ es conmutativa.

\subsection*{3. Asociatividad de la suma}

Para $u,v,w\in\mathbb{R}^{n}$,

$$
(u+v)+w=u+(v+w),
$$

ya que la suma de números reales es asociativa componente a componente.

\subsection*{4. Existencia del vector cero}

El vector

$$
0=(0,0,\ldots,0)
$$

satisface

$$
u+0=(x_{1},\ldots,x_{n})+(0,\ldots,0)=(x_{1},\ldots,x_{n})=u.
$$



\subsection*{5. Existencia del inverso aditivo}

Para $u=(x_{1},\ldots,x_{n})$ definimos

$$
-u=(-x_{1},\ldots,-x_{n}),
$$

y entonces

$$
u+(-u)=(0,\ldots,0).
$$



\subsection*{6. Cerradura bajo multiplicación por escalares}

Si $\alpha\in\mathbb{R}$ y $u\in\mathbb{R}^{n}$,

$$
\alpha u=(\alpha x_{1},\ldots,\alpha x_{n}),
$$

y cada componente pertenece a $\mathbb{R}$, por lo que $\alpha u\in\mathbb{R}^{n}$.



\subsection*{7. Distributividad respecto a la suma de vectores}

Para $\alpha\in\mathbb{R}$ y $u,v\in\mathbb{R}^{n}$,

$$
\alpha(u+v)=\alpha(x_{1}+y_{1},\ldots,x_{n}+y_{n})
=(\alpha x_{1}+\alpha y_{1},\ldots,\alpha x_{n}+\alpha y_{n})
=\alpha u+\alpha v.
$$

\subsection*{8. Distributividad respecto a la suma de escalares}

Para $\alpha,\beta\in\mathbb{R}$ y $u\in\mathbb{R}^{n}$,

$$
(\alpha+\beta)u=((\alpha+\beta)x_{1},\ldots,(\alpha+\beta)x_{n})
=(\alpha x_{1}+\beta x_{1},\ldots,\alpha x_{n}+\beta x_{n})
=\alpha u+\beta u.
$$



\subsection*{9. Compatibilidad del producto por escalares}

Para $\alpha,\beta\in\mathbb{R}$ y $u\in\mathbb{R}^{n}$,

$$
\alpha(\beta u)
=\alpha(\beta x_{1},\ldots,\beta x_{n})
=((\alpha\beta)x_{1},\ldots,(\alpha\beta)x_{n})
=(\alpha\beta)u.
$$



\subsection*{10. Identidad multiplicativa}

Para todo $u\in\mathbb{R}^{n}$,

$$
1u=(1x_{1},\ldots,1x_{n})=(x_{1},\ldots,x_{n})=u.
$$

\subsection*{Conclusión}

Como $\mathbb{R}^{n}$ con las operaciones usuales de suma y multiplicación por escalares satisface los diez axiomas anteriores, concluimos que

$$
\mathbb{R}^{n}
$$

es un espacio vectorial sobre el cuerpo $\mathbb{R}$.

# El espacio vectorial $\mathbb{P}_{n}$

Sea   

$$
V=\mathbb{P}_{n}=\left\{ p \mid p(x)=a_{n}x^{n}+a_{n-1}x^{n-1}+\cdots+a_{1}x+a_{0},\quad a_{i}\in\mathbb{R} \right\},
$$  

el conjunto de todos los polinomios reales de grado menor o igual que $n$.

En $V$ se definen las operaciones usuales:

- suma de polinomios: $(p+q)(x) = p(x) + q(x)$,
- multiplicación por escalar: $(\alpha p)(x) = \alpha p(x)$, con $\alpha \in \mathbb{R}$.

Demostraremos que $\mathbb{P}_{n}$ satisface los axiomas de espacio vectorial sobre $\mathbb{R}$.



## 1. Cerradura bajo la suma

Sean $p,q\in\mathbb{P}_{n}$, con

$$
p(x)=a_{n}x^{n}+\cdots+a_{0}, \quad q(x)=b_{n}x^{n}+\cdots+b_{0}.

Entonces

$$
(p+q)(x)=(a_{n}+b_{n})x^{n}+\cdots+(a_{0}+b_{0}),
$$

que es un polinomio de grado menor o igual que $n$. Por tanto, $p+q\in\mathbb{P}_{n}$.



\subsection*{2. Conmutatividad de la suma}

Para todo $p,q\in\mathbb{P}_{n}$,

$$
p(x)+q(x)=q(x)+p(x),
$$

porque la suma en $\mathbb{R}$ es conmutativa.



\subsection*{3. Asociatividad de la suma}

Para $p,q,r\in\mathbb{P}_{n}$,

$$
(p+q)+r=p+(q+r),
$$

ya que la suma de coeficientes reales es asociativa.



## 4. Existencia del polinomio cero

El polinomio

$$
0(x)=0
$$

pertenece a $\mathbb{P}_{n}$ y satisface

$$
p(x)+0(x)=p(x).
$$



### 5. Existencia del inverso aditivo

Para cada $p(x)=a_{n}x^{n}+\cdots+a_{0}$ definimos

$$
(-p)(x)=-a_{n}x^{n}-\cdots-a_{0},
$$

y entonces

$$
p(x)+(-p)(x)=0.
$$



\subsection*{6. Cerradura bajo multiplicación por escalares}

Sea $\alpha\in\mathbb{R}$ y $p\in\mathbb{P}_{n}$,

$$
(\alpha p)(x)=\alpha a_{n}x^{n}+\cdots+\alpha a_{0},
$$

que sigue siendo un polinomio de grado menor o igual que $n$.



\subsection*{7. Distributividad respecto a la suma de polinomios}

Para $\alpha\in\mathbb{R}$ y $p,q\in\mathbb{P}_{n}$,

$$
\alpha(p+q)=\alpha p+\alpha q.
$$

Esto se verifica al multiplicar cada coeficiente por $\alpha$.



\subsection*{8. Distributividad respecto a la suma de escalares}

Para $\alpha,\beta\in\mathbb{R}$ y $p\in\mathbb{P}_{n}$,

$$
(\alpha+\beta)p=\alpha p+\beta p.
$$



\subsection*{9. Compatibilidad del producto por escalares}

Para $\alpha,\beta\in\mathbb{R}$ y $p\in\mathbb{P}_{n}$,

$$
\alpha(\beta p)=(\alpha\beta)p.
$$



\subsection*{10. Identidad multiplicativa}

Para todo $p\in\mathbb{P}_{n}$,

$$
1p=p.
$$



### Conclusión

Como el conjunto $\mathbb{P}_{n}$ con las operaciones usuales de suma de polinomios y multiplicación por escalares satisface los diez axiomas anteriores, se concluye que

$$
\mathbb{P}_{n}
$$

es un espacio vectorial sobre el cuerpo $\mathbb{R}$.



# 3. Si $V = M_{mn}$ (conjunto de matrices $m$ por $n$) con la suma y la multiplicación escalar usuales, es fácil verificar que $M_{mn}$ es un espacio vectorial

$$
V=M_{m n}=\left\{ A=(a_{ij}) \mid a_{ij}\in\mathbb{R},\; 1\le i\le m,\; 1\le j\le n \right\},

el conjunto de todas las matrices reales de tamaño $m\times n$.

En este conjunto se definen las operaciones usuales:

- suma de matrices: si $A=(a_{ij})$ y $B=(b_{ij})$, entonces
$$
A+B=(a_{ij}+b_{ij}),

- multiplicación por escalar: si $\alpha\in\mathbb{R}$ y $A=(a_{ij})$, entonces

Demostraremos que $M_{m n}$ satisface los axiomas de espacio vectorial sobre $\mathbb{R}$.

## 1. Cerradura bajo la suma

Si $A=(a_{ij})$ y $B=(b_{ij})$ pertenecen a $M_{m n}$, entonces

$$
A+B=(a_{ij}+b_{ij}),
$$

y como la suma de números reales es un número real, cada entrada pertenece a $\mathbb{R}$; por tanto $A+B\in M_{m n}$.



## 2. Conmutatividad de la suma

Para $A,B\in M_{m n}$,

$$
A+B=(a_{ij}+b_{ij})=(b_{ij}+a_{ij})=B+A,
$$

ya que la suma en $\mathbb{R}$ es conmutativa.

\subsection*{3. Asociatividad de la suma}

Para $A,B,C\in M_{m n}$,

$$
(A+B)+C=A+(B+C),
$$

porque la suma de números reales es asociativa en cada componente.

\subsection*{4. Existencia del elemento neutro aditivo}

La matriz cero

$$
0=(0_{ij}),
$$

cuyas entradas son todas iguales a $0$, pertenece a $M_{m n}$ y satisface

$$
A+0=A
$$

para toda matriz $A\in M_{m n}$.



\subsection*{5. Existencia del inverso aditivo}

Para cada $A=(a_{ij})$ definimos

$$
-A=(-a_{ij}),
$$

y entonces

$$
A+(-A)=0.
$$



### 6. Cerradura bajo multiplicación por escalares

Si $\alpha\in\mathbb{R}$ y $A=(a_{ij})\in M_{m n}$,

$$
\alpha A=(\alpha a_{ij}),
$$

y cada entrada sigue siendo un número real; por tanto $\alpha A\in M_{m n}$.



\subsection*{7. Distributividad respecto a la suma de matrices}

Para $\alpha\in\mathbb{R}$ y $A,B\in M_{m n}$,

$$
\alpha(A+B)=\alpha A+\alpha B,
$$

ya que

$$
\alpha(a_{ij}+b_{ij})=\alpha a_{ij}+\alpha b_{ij}.
$$



\subsection*{8. Distributividad respecto a la suma de escalares}

Para $\alpha,\beta\in\mathbb{R}$ y $A\in M_{m n}$,

$$
(\alpha+\beta)A=\alpha A+\beta A.
$$



\subsection*{9. Compatibilidad del producto por escalares}

Para $\alpha,\beta\in\mathbb{R}$ y $A\in M_{m n}$,

$$
\alpha(\beta A)=(\alpha\beta)A,
$$

pues

$$
\alpha(\beta a_{ij})=(\alpha\beta)a_{ij}.
$$



\subsection*{10. Identidad multiplicativa}

Para toda matriz $A\in M_{m n}$,

$$
1A=A,
$$

ya que

$$
1\cdot a_{ij}=a_{ij}.
$$

Como el conjunto $M_{m n}$ con las operaciones usuales de suma de matrices y multiplicación por escalares satisface los diez axiomas anteriores, se concluye que

M_{m n}

es un espacio vectorial sobre el cuerpo $\mathbb{R}$.


# 4. Sea $S_{3}$ el conjunto de matrices invertibles de $3 \times 3$. Se define la suma $A \oplus B$ por $A \oplus B = AB$.  

## El conjunto $S_{3}$ con la operación $A \oplus B = AB$



Sea 
$$
S_{3}=\{A\in M_{3\times 3}(\mathbb{R}) : A \text{ es invertible}\},
$$

y definamos una operación interna

$$
A \oplus B = AB.
$$



Mostraremos que $(S_{3},\oplus, \cdot)$ no puede tener estructura de espacio vectorial.

## 1. La operación es cerrada

Si $A, B \in S_{3}$, entonces ambas matrices son invertibles y su producto $AB$ también es invertible. Por tanto,

$$
A\oplus B \in S_{3}.
$$



## 2. La operación es asociativa

Para $A,B,C\in S_{3}$,

$$
(A\oplus B)\oplus C = (AB)C = A(BC)=A \oplus (B \oplus C),
$$

ya que la multiplicación de matrices es asociativa.



### \subsection*{3. Existencia de elemento neutro}

La matriz identidad $I$ es invertible y satisface

$$
A\oplus I = AI = A, \quad I\oplus A = IA = A.
$$



### 4. Existencia de inverso

Para cada $A \in S_{3}$ existe $A^{-1} \in S_{3}$ tal que

$$
A\oplus A^{-1}=AA^{-1}=I.
$$



## 5. Falla de la conmutatividad

En un espacio vectorial la suma debe ser conmutativa. Sin embargo, en general la multiplicación de matrices no lo es.

Existen matrices invertibles $A,B$ tales que

$$
AB \ne BA.
$$

Por tanto,

$$
A\oplus B \ne B\oplus A,
$$

lo cual viola el axioma de conmutatividad de la suma en un espacio vectorial.

\subsection*{6. Incompatibilidad con la multiplicación por escalares}

Aun si se define la multiplicación por escalares usual

$$
\alpha A,
$$

no se satisface la distributividad requerida en un espacio vectorial:

$$
\alpha(A\oplus B)=\alpha(AB),
$$

mientras que

$$
(\alpha A)\oplus(\alpha B)=(\alpha A)(\alpha B)=\alpha^{2}AB.
$$

En general,

$$
\alpha AB \ne \alpha^{2}AB,
$$

por lo que falla el axioma

$$
\alpha(u+v)=\alpha u+\alpha v.
$$



## Conclusión

La operación $\oplus$ definida por la multiplicación de matrices no cumple los axiomas de suma de un espacio vectorial (en particular, no es conmutativa y no es compatible con la multiplicación por escalares). Por lo tanto,

$$
S_{3}
$$

con la operación $A \oplus B = AB$ no es un espacio vectorial.


# 5. Sea $V = C[a,b]$ conjunto de funciones continuas de valores reales definidas en el intervalo $[a,b]$. Se define  

$$
(f+g)(x) = f(x) + g(x), \quad \quad (\alpha f)(x) = \alpha f(x) 
$$

# El espacio vectorial $C[a,b]$

Sea 
$$
V=C[a,b]=\{f:[a,b]\to\mathbb{R}\mid f \text{ es continua en }[a,b]\},
$$
el conjunto de funciones continuas de valores reales definidas en el intervalo $[a,b]$.

Definimos las operaciones:

- suma de funciones
$$
(f+g)(x)=f(x)+g(x),
$$

- multiplicación por escalar
$$
(\alpha f)(x)=\alpha f(x), \quad \alpha\in\mathbb{R}.
$$

Demostraremos que $C[a,b]$ es un espacio vectorial sobre $\mathbb{R}$ verificando los diez axiomas.



# 1. Cerradura bajo la suma

Si $f,g\in C[a,b]$, entonces $f(x)$ y $g(x)$ son continuas en $[a,b]$.  
La suma de funciones continuas es continua, por lo tanto

$$
(f+g)(x)=f(x)+g(x)
$$

es continua en $[a,b]$, y así $f+g\in C[a,b]$.

\subsection*{2. Conmutatividad de la suma}

Para todo $f,g\in C[a,b]$,

$$
(f+g)(x)=f(x)+g(x)=g(x)+f(x)=(g+f)(x).
$$



### \subsection*{3. Asociatividad de la suma}

Para $f,g,h\in C[a,b]$,

$$
((f+g)+h)(x)=(f+g)(x)+h(x)=f(x)+g(x)+h(x),
$$

y

$$
(f+(g+h))(x)=f(x)+(g+h)(x)=f(x)+g(x)+h(x),
$$

luego

$$
(f+g)+h=f+(g+h).
$$



# 4. Existencia del elemento neutro aditivo

La función cero definida por

$$
0(x)=0 \quad \text{para todo } x\in[a,b]
$$

es continua y satisface

$$
(f+0)(x)=f(x)+0=f(x).
$$

Por tanto, $0\in C[a,b]$ es el neutro aditivo.



### \subsection*{5. Existencia del inverso aditivo}

Para $f\in C[a,b]$ definimos

$$
(-f)(x)=-f(x).
$$

Esta función es continua y cumple

$$
(f+(-f))(x)=f(x)-f(x)=0.
$$



# 6. Cerradura bajo multiplicación por escalares

Sea $\alpha\in\mathbb{R}$ y $f\in C[a,b]$.  
Como el producto de una función continua por un escalar es continuo,

$$
(\alpha f)(x)=\alpha f(x)
$$

es continua en $[a,b]$. Por tanto, $\alpha f\in C[a,b]$.



### \subsection*{7. Distributividad respecto a la suma de funciones}

Para $\alpha\in\mathbb{R}$ y $f,g\in C[a,b]$,

$$
\alpha(f+g)(x)=\alpha(f(x)+g(x))=\alpha f(x)+\alpha g(x)
=(\alpha f+\alpha g)(x).
$$



# 8. Distributividad respecto a la suma de escalares

Para $\alpha,\beta\in\mathbb{R}$ y $f\in C[a,b]$,

$$
((\alpha+\beta)f)(x)=(\alpha+\beta)f(x)
=\alpha f(x)+\beta f(x)
=(\alpha f+\beta f)(x).
$$



# Compatibilidad del producto por escalares  

Para $\alpha,\beta\in\mathbb{R}$ y $f\in C[a,b]$,

$$
(\alpha(\beta f))(x)=\alpha(\beta f(x))=(\alpha\beta)f(x)
=((\alpha\beta)f)(x).
$$



# 10. Identidad multiplicativa  

Para todo $f\in C[a,b]$,

$$
(1f)(x)=1\cdot f(x)=f(x).
$$



# Conclusión

Como el conjunto $C[a,b]$ con las operaciones de suma de funciones y multiplicación por escalares satisface los diez axiomas de espacio vectorial, concluimos que

$$
C[a,b]
$$

es un espacio vectorial sobre el cuerpo $\mathbb{R}$.


# 6. Sea $V$  el conjunto de  los números  reales $\mathbb{R}$ con las operaciones $a \oplus b = a - b$ (es la resta ordinaria) y la multiplicación por un escalar como la multiplicación ordinaria. Veamos si $V$ es un espacio vectorial.

## Un ejemplo que no es espacio vectorial

Sea 
$$
V=\mathbb{R}
$$

con las operaciones definidas por

$$
a \oplus b = a-b
$$

(resta ordinaria) y la multiplicación por escalar usual

$$
\alpha a = \alpha a, \quad \alpha \in \mathbb{R}.
$$

Analicemos si $V$ es un espacio vectorial verificando los axiomas.



## 1. Cerradura bajo la operación $\oplus$

Si $a,b\in\mathbb{R}$, entonces $a-b\in\mathbb{R}$.  
Por tanto, la operación es cerrada.

\subsection*{2. Conmutatividad}

En un espacio vectorial debe cumplirse

$$
a\oplus b=b\oplus a.
$$

Pero

$$
a\oplus b=a-b, \qquad b\oplus a=b-a.
$$

En general,

$$
a-b\neq b-a.
$$

Por ejemplo, si $a=3$ y $b=1$,

$$
a\oplus b=2, \qquad b\oplus a=-2.
$$

Luego la operación no es conmutativa.

## 3. Asociatividad

Debe cumplirse

$$
(a\oplus b)\oplus c=a\oplus(b\oplus c).
$$

Sin embargo,

$$
(a\oplus b)\oplus c=(a-b)-c=a-b-c,
$$

mientras que

$$
a\oplus(b\oplus c)=a-(b-c)=a-b+c.
$$

En general,

$$
a-b-c\neq a-b+c.
$$

Por ejemplo, si $a=5$, $b=2$, $c=1$,

$$
(a\oplus b)\oplus c=2, \qquad a\oplus(b\oplus c)=4.
$$

Por tanto, la operación no es asociativa.

## 4. Elemento neutro aditivo

Buscamos $0_V\in\mathbb{R}$ tal que

$$
a\oplus 0_V=a.
$$

Esto implica

$$
a-0_V=a \Rightarrow 0_V=0.
$$

Pero además debería cumplirse

$$
0_V\oplus a=a,
$$

lo cual daría

$$
0-a=a,
$$

lo que solo ocurre si $a=0$.  
Por tanto, no existe elemento neutro aditivo bilateral.



## 5. Incompatibilidad con la multiplicación por escalares

En un espacio vectorial debe cumplirse

$$
\alpha(a\oplus b)=\alpha a\oplus \alpha b.
$$

Calculemos ambos lados:

$$
\alpha(a\oplus b)=\alpha(a-b)=\alpha a-\alpha b,
$$

mientras que

$$
\alpha a\oplus \alpha b=\alpha a-\alpha b.
$$

Esta propiedad sí se cumple, pero no es suficiente para garantizar la estructura de espacio vectorial, ya que fallan axiomas fundamentales anteriores.



## Conclusión

La operación $a \oplus b = a-b$ no es conmutativa ni asociativa y no posee elemento neutro aditivo bilateral.  

Por lo tanto, el conjunto

$$
V=\mathbb{R}
$$

con las operaciones dadas no satisface los axiomas de un espacio vectorial. En consecuencia, no es un espacio vectorial sobre $\mathbb{R}$.


# 7. Sea $V = \{ (x,y,z) \mid ax+by + cz = 0, \quad a, b, c \in \mathbb{R} \text{ fijos} \}$. Esto es, $V$ es el conjunto de puntos en $\mathbb{R}^{3}$ que están en el plano con vector normal $(a, b, c)$ y que pasa por el origen.


Sea
$$
V=\{(x,y,z)\in\mathbb{R}^{3}\mid ax+by+cz=0,\quad a,b,c\in\mathbb{R}\text{ fijos}\},
$$

el conjunto de puntos del plano que pasa por el origen y tiene vector normal $(a,b,c)$.  
Las operaciones definidas en $V$ son las usuales heredadas de $\mathbb{R}^{3}$:



$$
(x_{1},y_{1},z_{1})+(x_{2},y_{2},z_{2})=(x_{1}+x_{2},y_{1}+y_{2},z_{1}+z_{2}),
$$

$$
\alpha(x,y,z)=(\alpha x,\alpha y,\alpha z),\quad \alpha\in\mathbb{R}.
$$

Demostraremos que $V$ es un espacio vectorial verificando los axiomas.



### 1. Cerradura bajo la suma

Sean $u=(x_{1},y_{1},z_{1})$ y $v=(x_{2},y_{2},z_{2})$ en $V$. Entonces

$$
ax_{1}+by_{1}+cz_{1}=0, \qquad ax_{2}+by_{2}+cz_{2}=0.
$$



Para $u+v$ se tiene

$$
a(x_{1}+x_{2})+b(y_{1}+y_{2})+c(z_{1}+z_{2})
=(ax_{1}+by_{1}+cz_{1})+(ax_{2}+by_{2}+cz_{2})=0.
$$

Luego $u+v\in V$.



## 2. Conmutatividad de la suma

Para todo $u,v\in V$,

$$
u+v=v+u,
$$

ya que la suma en $\mathbb{R}^{3}$ es conmutativa.



### 3. Asociatividad de la suma

Para $u,v,w\in V$,

$$
(u+v)+w=u+(v+w),
$$

pues la suma de vectores en $\mathbb{R}^{3}$ es asociativa.



### 4. Existencia del vector cero

El vector

$$
0=(0,0,0)
$$

satisface

$$
a\cdot 0+b\cdot 0+c\cdot 0=0,
$$

por lo que $0\in V$ y además

$$
u+0=u \quad \text{para todo } u\in V.
$$



### 5. Existencia del inverso aditivo

Si $u=(x,y,z)\in V$, entonces

$$
ax+by+cz=0.
$$

Consideremos $-u=(-x,-y,-z)$. Se cumple

$$
a(-x)+b(-y)+c(-z)=-(ax+by+cz)=0,
$$

por lo que $-u\in V$ y

$$
u+(-u)=0.
$$



### 6. Cerradura bajo multiplicación por escalares

Sea $\alpha\in\mathbb{R}$ y $u=(x,y,z)\in V$. Entonces

$$
ax+by+cz=0.
$$

Para $\alpha u$,

$$
a(\alpha x)+b(\alpha y)+c(\alpha z)
=\alpha(ax+by+cz)=\alpha\cdot 0=0,
$$

por lo que $\alpha u\in V$.



### 7. Distributividad respecto a la suma de vectores

Para $\alpha\in\mathbb{R}$ y $u,v\in V$,

$$
\alpha(u+v)=\alpha u+\alpha v,
$$

porque esta propiedad es válida en $\mathbb{R}^{3}$.



### 8. Distributividad respecto a la suma de escalares

Para $\alpha,\beta\in\mathbb{R}$ y $u\in V$,

$$
(\alpha+\beta)u=\alpha u+\beta u,
$$

propiedad heredada de $\mathbb{R}^{3}$.



### 9. Compatibilidad del producto por escalares

Para $\alpha,\beta\in\mathbb{R}$ y $u\in V$,

$$
\alpha(\beta u)=(\alpha\beta)u.
$$



### 10. Identidad multiplicativa

Para todo $u\in V$,

$$
1u=u.
$$



### Conclusión

El conjunto $V$ es no vacío, está cerrado bajo la suma y la multiplicación por escalares, y las operaciones satisfacen los axiomas heredados de $\mathbb{R}^{3}$. Por tanto,

$$
V = \{(x,y,z)\in\mathbb{R}^{3}\mid ax+by+cz = 0\}
$$

es un espacio vectorial sobre $\mathbb{R}$.  
Geométricamente, corresponde a un plano que pasa por el origen en $\mathbb{R}^{3}$, es decir, un subespacio vectorial de $\mathbb{R}^{3}$.


El siguiente ejemplo nos muestra que no necesariamente la suma vectorial ha de estar relacionada con la suma ordinaria, y que el vector cero, 0, no tiene que relacionarse con el número real 0.

8. Sea $V = \{x \mid x \in \mathbb{R} \text{ y } x > 0  \}$. La suma y la multiplicación por escalar en este conjunto se definen así:   

$$
x \oplus y = xy, \quad \quad \alpha \odot x = x^{\alpha}. 
$$

## El conjunto $V=\{x\in\mathbb{R}\mid x>0\}$ como espacio vectorial

Para todo $x,y\in V$ y $\alpha\in\mathbb{R}$. Demostraremos que $(V,\oplus,\odot)$ es un espacio vectorial sobre $\mathbb{R}$ verificando los axiomas.

## 1. Cerradura bajo la suma

Si $x,y>0$, entonces $xy>0$, por lo que

$$
x\oplus y\in V.
$$



## 2. Conmutatividad

Para todo $x,y\in V$,

$$
x\oplus y=xy=yx=y\oplus x.
$$



## 3. Asociatividad

Para $x,y,z\in V$,

$$
(x\oplus y)\oplus z=(xy)z=x(yz)=x\oplus(y\oplus z).
$$



## 4. Elemento neutro aditivo

Buscamos $e\in V$ tal que

$$
x\oplus e=x.
$$

Esto implica

$$
xe=x \Rightarrow e=1.
$$

Luego el neutro aditivo es

$$
e=1\in V.
$$



# 5. Inverso aditivo

Para cada $x\in V$ buscamos $y\in V$ tal que

$$
x\oplus y=1.
$$

Esto implica

$$
xy=1 \Rightarrow y=\frac{1}{x}>0.
$$

Por tanto, cada elemento tiene inverso aditivo.



## 6. Cerradura bajo multiplicación por escalares

Si $\alpha\in\mathbb{R}$ y $x>0$, entonces

$$
x^{\alpha}>0,
$$

por lo que $\alpha\odot x\in V$.



# 7. Distributividad respecto a la suma de vectores

Para $\alpha\in\mathbb{R}$ y $x,y\in V$,

$$
\alpha\odot(x\oplus y)
=(xy)^{\alpha}
=x^{\alpha}y^{\alpha}
=(\alpha\odot x)\oplus(\alpha\odot y).
$$



## 8. Distributividad respecto a la suma de escalares

Para $\alpha,\beta\in\mathbb{R}$ y $x\in V$,

$$
(\alpha+\beta)\odot x
=x^{\alpha+\beta}
=x^{\alpha}x^{\beta}
=(\alpha\odot x)\oplus(\beta\odot x).
$$



## 9. Compatibilidad del producto por escalares

Para $\alpha,\beta\in\mathbb{R}$ y $x\in V$,

$$
\alpha\odot(\beta\odot x)
=\alpha\odot(x^{\beta})
=(x^{\beta})^{\alpha}
=x^{\alpha\beta}
=(\alpha\beta)\odot x.
$$



# 10. Identidad multiplicativa

Para todo $x\in V$,

$$
1\odot x=x^{1}=x.
$$



## Conclusión

El conjunto $V$ es no vacío y las operaciones $\oplus$ y $\odot$ satisfacen los diez axiomas de espacio vectorial. Por lo tanto,

$$
V=\{x\in\mathbb{R}\mid x>0\}
$$

es un espacio vectorial sobre $\mathbb{R}$ con las operaciones

$$
x\oplus y=xy, \qquad \alpha\odot x=x^{\alpha}.
$$

# Línea argumentativa (Marco teórico)



El desarrollo del **pensamiento computacional** constituye hoy un eje fundamental en la educación matemática. Como lo plantea **Wing (2006)**, esta forma de razonamiento debe considerarse una competencia transversal y básica, al nivel de la lectura, la escritura y las matemáticas. 

En consecuencia, fomentar su aprendizaje desde los niveles iniciales hasta la educación superior resulta indispensable para formar profesionales capaces de enfrentar los retos de la sociedad contemporánea.



En el campo de la didáctica de las matemáticas, **Papert (1980, 1996)** había anticipado esta necesidad al concebir las computadoras no como simples herramientas de cálculo, sino como **“objetos para pensar”**, capaces de favorecer la construcción activa de significados matemáticos a través de la programación. 

Esta visión se alinea con el constructivismo de **Piaget**, que resalta el valor de la experimentación en el aprendizaje, y con la perspectiva socio-constructivista de **Vygotsky**, que otorga un papel central a las herramientas culturales en la mediación del conocimiento.



En este marco, el uso de entornos como **Jupyter Notebooks** y lenguajes como **Python, LaTeX, Markdown e inglés** no debe interpretarse como una rivalidad con la enseñanza tradicional de la matemática, sino como una **ampliación de los registros de representación** que describió **Duval (1993, 2006)**. 

Según su teoría semiótica, el aprendizaje significativo se potencia cuando el estudiante logra traducir y articular distintos sistemas de representación: gráfico, algebraico, numérico, verbal y, en nuestro tiempo, digital.



De igual manera, el **Enfoque Ontosemiótico de Godino (2002, 2017)** muestra que los significados matemáticos emergen de prácticas diversas, lo que justifica la incorporación de estos lenguajes como recursos que enriquecen las configuraciones de práctica matemática, en sintonía con las exigencias de la investigación y la práctica profesional.



En el ámbito internacional, organizaciones como el **NCTM (2000, 2014)** han subrayado que la tecnología es esencial para la enseñanza y el aprendizaje de la matemática, dado que facilita la experimentación, la validación de conjeturas y la comunicación de ideas. 

Asimismo, instancias como la **ACM** y el **IEEE** han señalado la importancia de desarrollar habilidades computacionales como requisito fundamental en la formación universitaria, tanto en el sector público como privado.



Por tanto, enseñar matemáticas integrando Python, LaTeX, Markdown e inglés en ambientes como Jupyter Notebooks contribuye a la formación de profesionales con **competencias científicas, comunicativas y tecnológicas**, capaces de comprender la matemática de manera interactiva y experimental, y de responder a las demandas del mundo laboral. 

Esta integración, lejos de ser inadecuada, constituye un **puente entre la enseñanza académica y las necesidades de la sociedad**, fortaleciendo la pertinencia de los programas universitarios.

El interés por integrar la tecnología en la enseñanza de las matemáticas tiene raíces en las propuestas de **Papert (1980, 1996)**, quien con su enfoque constructionista concibió la computadora como un “objeto para pensar”, capaz de mediar la construcción activa de significados matemáticos mediante la programación. 

Esta perspectiva se enlaza con los planteamientos de **Piaget (1970)** y **Vygotsky (1978)**, quienes reconocieron la importancia de la experimentación y de las herramientas culturales en la formación del pensamiento. 

Bajo esta mirada, el uso de entornos digitales no constituye una amenaza para la enseñanza tradicional de las matemáticas, sino un recurso que amplía las posibilidades de aprendizaje.



A comienzos del siglo XXI, la noción de **pensamiento computacional** introducida por **Wing (2006)** renovó el debate al proponer que esta competencia debía enseñarse de manera transversal desde los niveles iniciales hasta la educación superior, situándola al mismo nivel que la lectura, la escritura y las matemáticas. 

Este enfoque se alinea con los planteamientos del **NCTM (2000, 2014)**, que destacó la tecnología como un componente esencial para la experimentación, la representación y la comunicación en matemáticas. 

En paralelo, aportes teóricos como la **teoría de registros de representación semiótica de Duval (1993, 2006)** y el **Enfoque Ontosemiótico de Godino (2007, 2017)** muestran que el aprendizaje matemático se enriquece al transitar entre múltiples sistemas de representación, incluyendo los digitales.



En el contexto actual, organizaciones como la **ACM** y el **IEEE** insisten en que las competencias computacionales son indispensables en la formación profesional, lo que refuerza la pertinencia de integrar lenguajes como **Python, LaTeX, Markdown e inglés** en ambientes como **Jupyter Notebooks** para la enseñanza universitaria de las matemáticas. 

Así, el estado del arte evidencia un consenso creciente: lejos de ser un obstáculo, el pensamiento computacional y el uso de lenguajes digitales constituyen una vía para fortalecer la comprensión matemática, formar profesionales más competentes y responder a las demandas del mundo laboral y científico contemporáneo.

# Estado del arte con citas

El uso de la tecnología en la enseñanza de las matemáticas encuentra uno de sus pilares en **Papert (1980)**, quien afirmaba que *“los niños no aprenden sobre la computadora, aprenden con la computadora”* (p. 6). 

En su visión, la programación se convierte en un medio para la construcción activa de significados matemáticos, idea que se enlaza con la epistemología genética de **Piaget (1970)**, quien sostuvo que *“el conocimiento se construye a través de la acción”* (p. 15), y con la perspectiva de **Vygotsky (1978)**, al señalar que *“toda función en el desarrollo cultural del niño aparece dos veces: primero en el plano social y después en el individual”* (p. 57). 

Estos enfoques coinciden en otorgar a las herramientas culturales un papel esencial en la mediación del aprendizaje.



Posteriormente, **Wing (2006)** acuñó el término pensamiento computacional, definiéndolo como *“una habilidad fundamental para todos, no solo para los científicos de la computación”* (p. 33). Esta autora subrayó que debía enseñarse de manera transversal, en paralelo con la lectura, la escritura y la matemática. 

En la misma línea, el **NCTM (2000)** destacó que *“la tecnología es esencial en la enseñanza y el aprendizaje de las matemáticas; influye en la matemática que se enseña y mejora el aprendizaje de los estudiantes”* (p. 24).



Desde la didáctica de las matemáticas, **Duval (2006)** mostró que *“la comprensión matemática requiere la coordinación de al menos dos registros de representación diferentes”* (p. 107), lo que legitima la incorporación de registros digitales como LaTeX, Markdown y Python. 

Por su parte, **Godino (2017)** planteó que *“el significado de un objeto matemático está constituido por el sistema de prácticas en el que interviene”* (p. 7), con lo cual la práctica digital se convierte en parte constitutiva del aprendizaje matemático.



Finalmente, organismos internacionales como la **ACM y el IEEE (2013)** enfatizan que *“las habilidades computacionales son esenciales para el ejercicio profesional en cualquier disciplina”* (p. 12). 

De esta manera, se consolida un consenso: la enseñanza universitaria de las matemáticas debe integrar herramientas contemporáneas como **Jupyter Notebooks, Python, LaTeX, Markdown e inglés**, no para competir con la enseñanza tradicional, sino para **fortalecer la comprensión matemática, potenciar la comunicación científica y responder a las demandas del mundo laboral**.
