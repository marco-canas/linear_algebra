```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# Dominio
x = np.linspace(-5, 5, 400)

# Función general
def f(x, n):
    return x**n

# Función de actualización
def plot_function(n):
    plt.figure(figsize=(7, 5))
    
    if n == 0:
        y = np.ones_like(x)
        label = r"$f(x)=1$"
    else:
        y = f(x, n)
        label = rf"$f(x)=x^{n}$"
    plt.axis([-4,4,-30, 30])
    plt.plot(x, y, linewidth=2, color = 'red')
    plt.axhline(0)
    plt.axvline(0)
    plt.grid(True)
    plt.legend([label])
    plt.title(f"Representación de la función para n = {n}")
    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.show()

# Deslizador
slider_n = widgets.IntSlider(
    value=1,
    min=0,
    max=3,
    step=1,
    description='n',
    continuous_update=False
)

# Interactividad
widgets.interact(plot_function, n=slider_n)
```

#  Introducción



Veíamos que si un conjunto $S$, linealmente dependiente, genera un espacio vectorial, podemos eliminar en $S$ aquellos vectores que son combinación lineal del resto, hasta cuando en $S$ sólo quedan los vectores que son LI.   



Este nuevo conjunto es el generador de tamaño mínimo para el espacio vectorial.   



Dicho conjunto de vectores es una base para el espacio y el número de elementos del conjunto, su dimensión.  



Llegamos acá al punto central del capítulo de espacios vectoriales; las nociones anteriores nos permitirán definir estos dos conceptos a partir de los cuales podemos construir espacios vectoriales.

# 5.1 Definición y ejemplos de bases   



# Definición 1  

Un conjunto finito de vectores $\{\mathbf{v}_{1}, \mathbf{v}_{2}, \ldots, \mathbf{v}_{n}\}$ es una *base* para un espacio vectorial $V$ si:  

1. $\{\mathbf{v}_{1}, \mathbf{v}_{2}, \ldots, \mathbf{v}_{n}\}$ es linealmente independiente. 
2. $\{\mathbf{v}_{1}, \mathbf{v}_{2}, \ldots, \mathbf{v}_{n}\}$ genera a $V$

# Ejemplos

1. Los vectores $\mathbf{e}_{1} = (1,0)$ y $\mathbf{e}_{1} = (0,1)$ forman una base para $\mathbb{R}^{2}$.  

```python
import matplotlib.pyplot as plt
import numpy as np

# Vectores de la base canónica
e1 = np.array([1, 0])
e2 = np.array([0, 1])

# Figura
plt.figure(figsize=(6, 6))

# Dibujo de vectores
plt.quiver(0, 0, e1[0], e1[1], angles='xy', scale_units='xy', \
    scale=1, color = 'red')
plt.quiver(0, 0, e2[0], e2[1], angles='xy', scale_units='xy', \
    scale=1, color = 'red')

# Etiquetas
plt.text(1.05, 0, r"$e_1=(1,0)$", fontsize=12)
plt.text(0, 1.05, r"$e_2=(0,1)$", fontsize=12)

# Ejes
plt.axhline(0)
plt.axvline(0)
plt.grid(True)

# Configuración del plano
plt.xlim(-0.5, 1.5)
plt.ylim(-0.5, 1.5)
plt.gca().set_aspect('equal', adjustable='box')

# Título
plt.title(r"Base canónica de $\mathbb{R}^2$")

plt.xlabel("x")
plt.ylabel("y")
plt.savefig(r"C:\Users\marco\Downloads\base_canonica_r2.png")
plt.show()
```

```python
# mostremos que 1,x,x2 es una base para P2

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# Dominio
x = np.linspace(-5, 5, 400)

# =========================
# Widgets
# =========================
a0_box = widgets.FloatText(value=1, description='a₀:')
a1_box = widgets.FloatText(value=0, description='a₁:')
a2_box = widgets.FloatText(value=1, description='a₂:')

alpha_slider = widgets.IntSlider(min=-5, max=5, step=1, value=0, description="α")
beta_slider  = widgets.IntSlider(min=-5, max=5, step=1, value=0, description="β")
gamma_slider = widgets.IntSlider(min=-5, max=5, step=1, value=0, description="γ")

# =========================
# Función de visualización
# =========================
def visualizar(a0, a1, a2, alpha, beta, gamma):
    
    # Polinomio del usuario
    P = a0 + a1*x + a2*x**2
    
    # Combinación lineal de {1, x, x²}
    combinacion = alpha*1 + beta*x + gamma*x**2
    
    plt.figure(figsize=(7,5))
    
    plt.plot(x, P, linewidth=3, label="Polinomio del usuario P(x)")
    plt.plot(x, combinacion, linestyle='--', linewidth=3,
             label="α·1 + β·x + γ·x²")
    
    plt.axhline(0)
    plt.axvline(0)
    plt.grid(True)
    plt.legend()
    plt.title("Todo polinomio de grado ≤ 2 es combinación lineal de {1, x, x²}")
    plt.xlabel("x")
    plt.ylabel("y")
    
    print("Polinomio ingresado:",
          f"P(x) = {a0} + {a1}x + {a2}x²")
    print("Combinación lineal:",
          f"{alpha}·1 + {beta}·x + {gamma}·x²")
    
    if alpha == a0 and beta == a1 and gamma == a2:
        print("✓ Coinciden exactamente.")
    
    plt.show()

# =========================
# Conectar widgets con la gráfica
# =========================
ui = widgets.VBox([
    widgets.HBox([a0_box, a1_box, a2_box]),
    widgets.HBox([alpha_slider, beta_slider, gamma_slider])
])

out = widgets.interactive_output(
    visualizar,
    {
        'a0': a0_box,
        'a1': a1_box,
        'a2': a2_box,
        'alpha': alpha_slider,
        'beta': beta_slider,
        'gamma': gamma_slider
    }
)

display(ui, out)
```

# Teorema 1  

Si $\{ \mathbf{v}_{1}, \mathbf{v}_{2}, \ldots, \mathbf{v}_{n} \}$ es una base de $V$ y si, $\mathbf{v} \in V$ entonces existe un único conjunto de escalares $c_{1}, c_{2}, \ldots, c_{n}$ tales que 

$$
\mathbf{v} = c_{1}\mathbf{v}_{1} + c_{2}\mathbf{v}_{2} + \cdots + c_{n}\mathbf{v}_{n}
$$


¿Contienen todas las bases de un espacio vectorial el mismo número de vectores?

# Teorema 2

Si $\{u_{1}, u_{2}, \ldots, u_{m}\}$ y $\{v_{1}, v_{2}, \ldots, v_{m}\}$ son bases para un espacio vectorial $V$, entonces $m = n$

# Dimensión. Definición, ejemplos y propiedades. 



# Definición 2  


La dimensión de un espacio vectorial no nulo es el número de vectores en una base para $V$.  

Como el conjunto $\{ \mathbf{0} \}$ es linealmente dependiente, se dice que el espacio vectorial $\{ \mathbf{0} \}$ tiene dimensión cero.

Cuando la base del espacio vectorial $V$ es un conjunto finito de vectores, se dice que $V$ es de dimensión finita. 

Sin embargo, existen muchos espacios vectoriales para los cuales no hay una base con un número finito de vectores; estos espacios son de dimensión infinita, por ejemplo el espacio vectorial $\mathbb{P}$ de todos los polinomios y el espacio $C[a, b]$ de funciones continuas en $[a, b]$.  

La dimensión de $V$ se denota por $dim\ V$. 

# Ejemplos

6. $dim\ \mathbb{P}_{n} = n+1$. 

7. $dim\ M_{mn} = mn.$

Como todo polinomio es una función continua, entonces $\mathbb{P}[0,1] \subset C[0,1]$ (funciones continuas en el intervalo $[0,1]$). $$

---



## 3) Expresar variables en función de un parámetro

De la segunda ecuación:

$$
y = -3z.
$$



Sustituimos en la primera:

$$
x - (-3z) - z = 0
$$
$$
x + 3z - z = 0
$$
$$
x + 2z = 0 \Rightarrow x = -2z.
$$



Sea $z = t$ (parámetro libre). Entonces:

$$
x = -2t,\quad y = -3t,\quad z = t.
$$

---



## 4) Escribir el vector solución

$$
(x,y,z) = (-2t,-3t,t).
$$

Factorizamos (t):

$$
(x,y,z) = t(-2,-3,1).
$$

---



## 5) Determinar una base del espacio solución

El espacio solución es:

$$
\{ t(-2,-3,1) \mid t \in \mathbb{R} \}.
$$

Es una recta que pasa por el origen en $\mathbb{R}^3$.
Por tanto, una base es:

$$
\boxed{\{(-2, -3, 1)\}}.
$$

---



## 6) Interpretación geométrica

* El sistema homogéneo define la intersección de dos planos que pasan por el origen.
* Su intersección es una recta.
* El vector $(-2,-3,1)$ da la dirección de esa recta.
* Todo vector solución es un múltiplo escalar de ese vector.


```python
import numpy as np
import plotly.graph_objects as go

# =========================
# Mallado para los planos
# =========================
x = np.linspace(-5, 5, 40)
y = np.linspace(-5, 5, 40)
X, Y = np.meshgrid(x, y)

# Plano 1: x - y - z = 0  →  z = x - y
Z1 = X - Y

# Plano 2: 2x - y + z = 0 → z = y - 2x
Z2 = Y - 2*X

# =========================
# Recta intersección
# (x,y,z) = t(-2,-3,1)
# =========================
t = np.linspace(-4, 4, 50)
xr = -2*t
yr = -3*t
zr = t

# =========================
# Crear figura
# =========================
fig = go.Figure()

# Plano 1
fig.add_trace(go.Surface(
    x=X, y=Y, z=Z1,
    opacity=0.5,
    name="Plano x - y - z = 0"
))

# Plano 2
fig.add_trace(go.Surface(
    x=X, y=Y, z=Z2,
    opacity=0.5,
    name="Plano 2x - y + z = 0"
))

# Recta intersección resaltada
fig.add_trace(go.Scatter3d(
    x=xr, y=yr, z=zr,
    mode='lines',
    line=dict(width=8),
    name="Recta intersección"
))

# =========================
# Ejes coordenados
# =========================
L = 6

# eje X
fig.add_trace(go.Scatter3d(
    x=[-L, L], y=[0, 0], z=[0, 0],
    mode='lines',
    line=dict(width=4),
    name="Eje X"
))

# eje Y
fig.add_trace(go.Scatter3d(
    x=[0, 0], y=[-L, L], z=[0, 0],
    mode='lines',
    line=dict(width=4),
    name="Eje Y"
))

# eje Z
fig.add_trace(go.Scatter3d(
    x=[0, 0], y=[0, 0], z=[-L, L],
    mode='lines',
    line=dict(width=4),
    name="Eje Z"
))

# =========================
# Ajustes visuales
# =========================
fig.update_layout(
    title="Intersección de los planos x - y - z = 0 y 2x - y + z = 0",
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z'
    ),
    legend=dict(
        x=0,
        y=0,
        xanchor='left',
        yanchor='bottom',
        bgcolor='rgba(255,255,255,0.7)'
    )
)

fig.show()
fig.write_image(r"C:\Users\marco\Downloads\planos_recta_interseccion.png")
```

Un conjunto de vectores en un espacio vectorial $V$ es una base para $V$ si genera a $V$ y es LI.   

Ahora, si se sabe que la dimensión del espacio es $n$, basta verificar una de las dos condiciones.

# Teorema 5  

Sea $V$ un espacio vectorial de dimensión $n$. Y sea $S = \{v_{1}, v_{2}, \ldots, v_{n}\}$ un conjunto de $n$ vectores es $V$.  

1. Si $S$ es LI, entonces es una base para $V$.
2. Si $S$ genera a $V$, entonces es una base para $V$.

Para determinar si un subconjunto $S$ de $\mathbb{R}^{n}$ es una base para,  $\mathbb{R}^{n}$ primero contamos el número de elementos de $S$.   

Si $S$ tiene $n$ elementos, cualquiera de las partes del teorema 5 nos permite determinar si $S$ es base.   



Si $S$ no tiene $n$ elementos, no es base para  $\mathbb{R}^{n}$.   

De la misma forma se procede en cualquier espacio o subespacio vectorial cuya dimensión sea conocida.

# Teorema 6  

Si $S$ es un conjunto LI de vectores en un espacio vectorial $V$ de dimensión finita,
entonces existe una base $B$ para $V$, que contiene a $S$.


El teorema nos dice que un conjunto LI de vectores en un espacio vectorial $V$ se
puede extender a una base para $V$.  

# 5.4  El problema de la base



Este problema puede presentarse en una de estas dos formas:  

1. Construir una base para $V$, tomando los vectores de $V$. Si puede escoger un conjunto de vectores que genere a $V$, puede eliminar los vectores LD, si los hay, y obtener una base para $V$.  



2. Dado un conjunto $S$ de vectores de $V$, construir una base para $V$ añadiendo, o bien eliminando, algunos vectores de $S$, o ambas cosas.  



Si $S$ genera a $V$, se procede como en el caso anterior.   

Si no, se aumenta $S$ hasta obtener un generador y luego se procede como en el caso anterior.

# Ejemplo  

Sea $S = \{(1, -1, 1), (0, 1, -1)\}$. Obtenga una base de $\mathbb{R}^{3}$ que contenga a $S$.  

## Solución


Como $\mathbb{R}^{3}$ tiene dimensión tres, la base para $\mathbb{R}^{3}$ debe contener tres vectores.

El conjunto $S$ es LI, ya que los vectores de $S$ no son uno un múltiplo escalar del otro, luego lo que hace falta es añadir otro vector $\mathbf{v}$ tal que $\mathbf{v} \notin gen S$.

# Solución 1 utilizando el concepto de determinante   

Sea $S = \{ (1,-1,1), (0,1,-1) \}$. Queremos obtener una base de $\mathbb{R}^{3}$ que contenga a $S$.

---



## 1) Interpretación del problema

Buscamos un tercer vector que, junto con los dos vectores de $S$, forme un conjunto linealmente independiente y genere todo $\mathbb{R}^{3}$.

Sea

$$
v_{1}=(1,-1,1), \quad v_{2}=(0,1,-1).
$$

Debemos encontrar $v_{3}$ tal que $\{ v_{1}, v_{2}, v_{3} \}$ sea base de $\mathbb{R}^{3}$.

---



## 2) Condición de independencia lineal

Tomemos un vector general $v_{3}=(a,b,c)$ y verifiquemos cuándo

$$
\{ v_{1}, v_{2}, v_{3} \}
$$

es linealmente independiente. Para ello usamos el determinante de la matriz que los tiene como filas (o columnas):

$$
A=
\begin{pmatrix}
1 & -1 & 1 \\ 
0 & 1 & -1 \\ 
a & b & c
\end{pmatrix}.
$$



El conjunto será base de $\mathbb{R}^{3}$ si y solo si

$$
\det(A)\neq 0.
$$

---



## 3) Cálculo del determinante

$$
\det(A)=
\begin{vmatrix}
1 & -1 & 1 \\ 
0 & 1 & -1 \\ 
a & b & c
\end{vmatrix}.
$$



Desarrollando por la primera fila:

$$
\det(A)=
1
\begin{vmatrix}
1 & -1 \\ 
b & c
\end{vmatrix}
-(-1)
\begin{vmatrix}
0 & -1 \\ 
a & c
\end{vmatrix}
+1
\begin{vmatrix}
0 & 1 \\ 
a & b
\end{vmatrix}.
$$




$$
1(1\cdot c-(-1)b)
+1(0\cdot c-(-1)a)
+1(0\cdot b-1\cdot a).
$$

$$
= (c+b)+(a)+(-a)=b+c.
$$



Entonces la independencia lineal se cumple cuando

$$
b + c \neq 0.
$$

---



## 4) Elección de un tercer vector

Podemos escoger cualquier vector que cumpla $b+c\neq 0$.
Por ejemplo, tomemos

$$
v_{3}=(0,0,1).
$$



Aquí $b+c=0+1=1\neq 0$, por lo tanto el conjunto es linealmente independiente.

---



## 5) Base buscada

La base de $\mathbb{R}^{3}$ que contiene a $S$ es

$$
\mathcal{B}= \{ (1, -1, 1), (0, 1, -1), (0, 0, 1) \}.
$$

---



## 6) Verificación conceptual

* $S$ tiene dos vectores linealmente independientes.
* Se añade un tercer vector que no pertenece al plano generado por ellos.
* El nuevo conjunto tiene tres vectores linealmente independientes en $\mathbb{R}^{3}$.

Por tanto,

$$
\mathcal{B}
$$

es una base de $\mathbb{R}^{3}$ que contiene al conjunto $S$.


# Segundo método de solución, utilizando el concepto de producto vectorial y la definición del plano como conjunto de vectores perpendiculares a un mismo vector  



Sea $S=\{ (1,-1,1),(0,1,-1) \}$. Queremos obtener una base de $\mathbb{R}^{3}$ que contenga a $S$, determinando primero el plano que generan estos vectores y luego tomando como tercer vector un normal a dicho plano.

---



## 1) Vectores generadores del plano

Sean

$$
v_{1} = (1, -1, 1), \quad v_{2} = (0, 1, -1).
$$

El plano generado por ellos es el conjunto de todas las combinaciones lineales:

$$
\Pi = \{ \alpha v_{1} + \beta v_{2} \mid \alpha, \beta \in \mathbb{R} \}.
$$

---



## 2) Determinar un vector normal al plano

Un vector normal al plano se obtiene con el producto cruz:

$$
n=v_{1}\times v_{2}.
$$



Calculamos

$$
n=
\begin{vmatrix}
\mathbf{i} & \mathbf{j} & \mathbf{k} \\ 
1 & -1 & 1 \\
0 & 1 & -1
\end{vmatrix}.
$$



Desarrollando:

$$
n=
\mathbf{i}
\begin{vmatrix}
-1 & 1 \\ 
1 & -1
\end{vmatrix}
-\mathbf{j}
\begin{vmatrix}
1 & 1 \\ 
0 & -1
\end{vmatrix}
+\mathbf{k}
\begin{vmatrix}
1 & -1 \\ 
0 & 1
\end{vmatrix}.
$$



$$
\mathbf{i}\big((-1)(-1)-1\cdot 1\big)
-\mathbf{j}\big(1(-1)-1\cdot 0\big)
+\mathbf{k}\big(1\cdot 1-(-1)\cdot 0\big).
$$




$$
\mathbf{i}(1-1)-\mathbf{j}(-1)+\mathbf{k}(1).
$$

$$
=(0,1,1).
$$



Por tanto, un vector normal al plano es

$$
n=(0,1,1).
$$

---



## 3) Ecuación del plano generado

El plano que pasa por el origen y tiene vector normal $n$ cumple

$$
0x+1y+1z=0.
$$

Es decir,

$$
y+z=0.
$$

Este es el plano generado por $v_{1}$ y $v_{2}$.

---



## 4) Construcción de una base de $\mathbb{R}^{3}$

Los vectores $v_{1}$ y $v_{2}$ están en el plano.
El vector normal $n$ es perpendicular al plano, por lo que no pertenece a él.

Así, el conjunto

$$
\{ v_{1},v_{2}, n \}
$$

es linealmente independiente.

Como contiene tres vectores linealmente independientes en $\mathbb{R}^{3}$, forma una base.

---



## 5) Base pedida

Una base de $\mathbb{R}^{3}$ que contiene a $S$ es

$$
\mathcal{B} = \{ (1,-1,1), (0,1,-1), (0,1,1) \}.
$$

---



## 6) Interpretación geométrica

* $v_{1}$ y $v_{2}$ generan un plano que pasa por el origen.
* $n$ es perpendicular a ese plano.
* El plano más su normal $\mathbf{n}$ generan todo el espacio.

Por tanto,

$$
\mathbb{R}^{3}=\text{span} \{\mathbf{v}_{1}, \mathbf{v}_{2}, \mathbf{n} \}.
$$

# 7. Definición de base

Un conjunto

$$
\{ v_1, v_2, \dots, v_k \} 
$$

es una **base** de un espacio vectorial si



1. Genera el espacio

$$
span\{v_1, \dots, v_k \} = V
$$

2. Los vectores son linealmente independientes.

---



# 8. Interpretación intuitiva

Una base es el **sistema de coordenadas más eficiente**.

Ejemplo en $\mathbb{R}^2$:

Base clásica

$$
(1,0), (0,1)
$$



Pero también es base

$$
(1,1), (1,-1)
$$

---



# 9. Experimento con Python



```python
import numpy as np

v1 = np.array([1,1])
v2 = np.array([1,-1])

A = np.column_stack((v1,v2))

np.linalg.matrix_rank(A)
```

Resultado esperado

2.

Por lo tanto forman una base.

---



# 10. Interpretación geométrica

Dos vectores independientes en $\mathbb{R}^2$ generan todo el plano.

Tres vectores independientes en $\mathbb{R}^3$ generan el espacio.

$$

$$
