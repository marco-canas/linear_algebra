````python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# Dominio
x = np.linspace(-5, 5, 400)

# Función general
def f(x, n):
    return x**n

# Función de actualización
def plot_function(n):
    plt.figure(figsize=(7, 5))
    
    if n == 0:
        y = np.ones_like(x)
        label = r"$f(x)=1$"
    else:
        y = f(x, n)
        label = rf"$f(x)=x^{n}$"
    plt.axis([-4,4,-30, 30])
    plt.plot(x, y, linewidth=2, color = 'red')
    plt.axhline(0)
    plt.axvline(0)
    plt.grid(True)
    plt.legend([label])
    plt.title(f"Representación de la función para n = {n}")
    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.show()

# Deslizador
slider_n = widgets.IntSlider(
    value=1,
    min=0,
    max=3,
    step=1,
    description='n',
    continuous_update=False
)

# Interactividad
widgets.interact(plot_function, n=slider_n)
````

#  Introducción



Veíamos que si un conjunto $S$, linealmente dependiente, genera un espacio vectorial, podemos eliminar en $S$ aquellos vectores que son combinación lineal del resto, hasta cuando en $S$ sólo quedan los vectores que son LI.   



Este nuevo conjunto es el generador de tamaño mínimo para el espacio vectorial.   



Dicho conjunto de vectores es una base para el espacio y el número de elementos del conjunto, su dimensión.  



Llegamos acá al punto central del capítulo de espacios vectoriales; las nociones anteriores nos permitirán definir estos dos conceptos a partir de los cuales podemos construir espacios vectoriales.

# 5.1 Definición y ejemplos de bases   



# Definición 1  

Un conjunto finito de vectores $\{\mathbf{v}_{1}, \mathbf{v}_{2}, \ldots, \mathbf{v}_{n}\}$ es una *base* para un espacio vectorial $V$ si:  

1. $\{\mathbf{v}_{1}, \mathbf{v}_{2}, \ldots, \mathbf{v}_{n}\}$ es linealmente independiente. 
2. $\{\mathbf{v}_{1}, \mathbf{v}_{2}, \ldots, \mathbf{v}_{n}\}$ genera a $V$

# Ejemplos

1. Los vectores $e_{1} = (1,0)$ y $e_{1} = (0,1)$ forman una base para $\mathbb{R}^{2}$.

````python
# mostremos que 1,x,x2 es una base para P2

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# Dominio
x = np.linspace(-5, 5, 400)

# =========================
# Widgets
# =========================
a0_box = widgets.FloatText(value=1, description='a₀:')
a1_box = widgets.FloatText(value=0, description='a₁:')
a2_box = widgets.FloatText(value=1, description='a₂:')

alpha_slider = widgets.IntSlider(min=-5, max=5, step=1, value=0, description="α")
beta_slider  = widgets.IntSlider(min=-5, max=5, step=1, value=0, description="β")
gamma_slider = widgets.IntSlider(min=-5, max=5, step=1, value=0, description="γ")

# =========================
# Función de visualización
# =========================
def visualizar(a0, a1, a2, alpha, beta, gamma):
    
    # Polinomio del usuario
    P = a0 + a1*x + a2*x**2
    
    # Combinación lineal de {1, x, x²}
    combinacion = alpha*1 + beta*x + gamma*x**2
    
    plt.figure(figsize=(7,5))
    
    plt.plot(x, P, linewidth=3, label="Polinomio del usuario P(x)")
    plt.plot(x, combinacion, linestyle='--', linewidth=3,
             label="α·1 + β·x + γ·x²")
    
    plt.axhline(0)
    plt.axvline(0)
    plt.grid(True)
    plt.legend()
    plt.title("Todo polinomio de grado ≤ 2 es combinación lineal de {1, x, x²}")
    plt.xlabel("x")
    plt.ylabel("y")
    
    print("Polinomio ingresado:",
          f"P(x) = {a0} + {a1}x + {a2}x²")
    print("Combinación lineal:",
          f"{alpha}·1 + {beta}·x + {gamma}·x²")
    
    if alpha == a0 and beta == a1 and gamma == a2:
        print("✓ Coinciden exactamente.")
    
    plt.show()

# =========================
# Conectar widgets con la gráfica
# =========================
ui = widgets.VBox([
    widgets.HBox([a0_box, a1_box, a2_box]),
    widgets.HBox([alpha_slider, beta_slider, gamma_slider])
])

out = widgets.interactive_output(
    visualizar,
    {
        'a0': a0_box,
        'a1': a1_box,
        'a2': a2_box,
        'alpha': alpha_slider,
        'beta': beta_slider,
        'gamma': gamma_slider
    }
)

display(ui, out)
````

$$
\mathbf{v} = c_{1}\mathbf{v}_{1} + c_{2}\mathbf{v}_{2} + \cdots + c_{n}\mathbf{v}_{n}
$$

# Dimensión. Definición, ejemplos y propiedades. 



# Definición 2  


La dimensión de un espacio vectorial no nulo es el número de vectores en una base para $V$.  

Como el conjunto $\{\mathbf{0}\}$ es linealmente dependiente, se dice que el espacio vectorial $\{0\}$ tiene dimensión cero.

Cuando la base del espacio vectorial $V$ es un conjunto finito de vectores, se dice
que $V$ es de dimensión finita. 

Sin embargo, existen muchos espacios vectoriales para los cuales no hay una base con un número finito de vectores; estos espacios son de dimensión infinita, por ejemplo el espacio vectorial $\mathbb{P}$ de todos los polinomios y el espacio $C[a, b]$ de funciones continuas en $[a, b]$.  

La dimensión de $V$ se denota por $dim\ V$. 

# Ejemplos

6. $dim\ \mathbb{P}_{n} = n+1$. 

7. $dim\ M_{mn} = mn.$

````python
import numpy as np
import plotly.graph_objects as go

# =========================
# Mallado para los planos
# =========================
x = np.linspace(-5, 5, 40)
y = np.linspace(-5, 5, 40)
X, Y = np.meshgrid(x, y)

# Plano 1: x - y - z = 0  →  z = x - y
Z1 = X - Y

# Plano 2: 2x - y + z = 0 → z = y - 2x
Z2 = Y - 2*X

# =========================
# Recta intersección
# (x,y,z) = t(-2,-3,1)
# =========================
t = np.linspace(-4, 4, 50)
xr = -2*t
yr = -3*t
zr = t

# =========================
# Crear figura
# =========================
fig = go.Figure()

# Plano 1
fig.add_trace(go.Surface(
    x=X, y=Y, z=Z1,
    opacity=0.5,
    name="Plano x - y - z = 0"
))

# Plano 2
fig.add_trace(go.Surface(
    x=X, y=Y, z=Z2,
    opacity=0.5,
    name="Plano 2x - y + z = 0"
))

# Recta intersección resaltada
fig.add_trace(go.Scatter3d(
    x=xr, y=yr, z=zr,
    mode='lines',
    line=dict(width=8),
    name="Recta intersección"
))

# =========================
# Ejes coordenados
# =========================
L = 6

# eje X
fig.add_trace(go.Scatter3d(
    x=[-L, L], y=[0, 0], z=[0, 0],
    mode='lines',
    line=dict(width=4),
    name="Eje X"
))

# eje Y
fig.add_trace(go.Scatter3d(
    x=[0, 0], y=[-L, L], z=[0, 0],
    mode='lines',
    line=dict(width=4),
    name="Eje Y"
))

# eje Z
fig.add_trace(go.Scatter3d(
    x=[0, 0], y=[0, 0], z=[-L, L],
    mode='lines',
    line=dict(width=4),
    name="Eje Z"
))

# =========================
# Ajustes visuales
# =========================
fig.update_layout(
    title="Intersección de los planos x - y - z = 0 y 2x - y + z = 0",
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z'
    ),
    legend=dict(
        x=0,
        y=0,
        xanchor='left',
        yanchor='bottom',
        bgcolor='rgba(255,255,255,0.7)'
    )
)

fig.show()
fig.write_image(r"C:\Users\marco\Downloads\planos_recta_interseccion.png")
````

Si $S$ no tiene $n$ elementos, no es base para  $\mathbb{R}^{n}$.   

De la misma forma se procede en cualquier espacio o subespacio vectorial cuya dimensión sea conocida.

# 5.4  El problema de la base



Este problema puede presentarse en una de estas dos formas:  

1. Construir una base para $V$, tomando los vectores de $V$. Si puede escoger un conjunto de vectores que genere a $V$, puede eliminar los vectores LD, si los hay, y obtener una base para $V$.  



2. Dado un conjunto $S$ de vectores de $V$, construir una base para $V$ añadiendo, o bien eliminando, algunos vectores de $S$, o ambas cosas.  



Si $S$ genera a $V$, se procede como en el caso anterior.   

Si no, se aumenta $S$ hasta obtener un generador y luego se procede como en el caso anterior.

# Ejemplo  

Sea $S = \{(1, -1, 1), (0, 1, -1)\}$. Obtenga una base de $\mathbb{R}^{3}$ que contenga a $S$.  

## Solución


Como $\mathbb{R}^{3}$ tiene dimensión tres, la base para $\mathbb{R}^{3}$ debe contener tres vectores.

El conjunto $S$ es LI, ya que los vectores de $S$ no son uno un múltiplo escalar del otro, luego lo que hace falta es añadir otro vector $\mathbf{v}$ tal que $\mathbf{v} \notin gen S$.

# Solución 1 utilizando el concepto de determinante   

Sea $S = \{ (1,-1,1), (0,1,-1) \}$. Queremos obtener una base de $\mathbb{R}^{3}$ que contenga a $S$.

---



## 1) Interpretación del problema

Buscamos un tercer vector que, junto con los dos vectores de $S$, forme un conjunto linealmente independiente y genere todo $\mathbb{R}^{3}$.

Sea

$$
v_{1}=(1,-1,1), \quad v_{2}=(0,1,-1).
$$

Debemos encontrar $v_{3}$ tal que $\{ v_{1}, v_{2}, v_{3} \}$ sea base de $\mathbb{R}^{3}$.

---



## 2) Condición de independencia lineal

Tomemos un vector general $v_{3}=(a,b,c)$ y verifiquemos cuándo

$$
\{ v_{1}, v_{2}, v_{3} \}
$$

es linealmente independiente. Para ello usamos el determinante de la matriz que los tiene como filas (o columnas):

$$
A=
\begin{pmatrix}
1 & -1 & 1 \\ 
0 & 1 & -1 \\ 
a & b & c
\end{pmatrix}.
$$



El conjunto será base de $\mathbb{R}^{3}$ si y solo si

$$
\det(A)\neq 0.
$$

---



## 3) Cálculo del determinante

$$
\det(A)=
\begin{vmatrix}
1 & -1 & 1 \
0 & 1 & -1 \
a & b & c
\end{vmatrix}.
$$

Desarrollando por la primera fila:

$$
\det(A)=
1
\begin{vmatrix}
1 & -1 \
b & c
\end{vmatrix}
-(-1)
\begin{vmatrix}
0 & -1 \
a & c
\end{vmatrix}
+1
\begin{vmatrix}
0 & 1 \
a & b
\end{vmatrix}.
$$

# $$

1(1\cdot c-(-1)b)
+1(0\cdot c-(-1)a)
+1(0\cdot b-1\cdot a).
$$

$$
= (c+b)+(a)+(-a)=b+c.
$$

Entonces la independencia lineal se cumple cuando

$$
b+c\neq 0.
$$

---

## 4) Elección de un tercer vector

Podemos escoger cualquier vector que cumpla $b+c\neq 0$.
Por ejemplo, tomemos

$$
v_{3}=(0,0,1).
$$

Aquí $b+c=0+1=1\neq 0$, por lo tanto el conjunto es linealmente independiente.

---

## 5) Base buscada

La base de $\mathbb{R}^{3}$ que contiene a $S$ es

$$
\mathcal{B}={(1,-1,1),(0,1,-1),(0,0,1)}.
$$

---

## 6) Verificación conceptual

* $S$ tiene dos vectores linealmente independientes.
* Se añade un tercer vector que no pertenece al plano generado por ellos.
* El nuevo conjunto tiene tres vectores linealmente independientes en $\mathbb{R}^{3}$.

Por tanto,

$$
\mathcal{B}
$$

es una base de $\mathbb{R}^{3}$ que contiene al conjunto $S$.


# Segundo método de solución, utilizando el concepto de producto vectorial y la definición del plano como conjunto de vectores perpendiculares a un mismo vector  



Sea $S=\{ (1,-1,1),(0,1,-1) \}$. Queremos obtener una base de $\mathbb{R}^{3}$ que contenga a $S$, determinando primero el plano que generan estos vectores y luego tomando como tercer vector un normal a dicho plano.

---



## 1) Vectores generadores del plano

Sean

$$
v_{1}=(1,-1,1), \quad v_{2}=(0,1,-1).
$$

El plano generado por ellos es el conjunto de todas las combinaciones lineales:

$$
\Pi={, \alpha v_{1}+\beta v_{2} \mid \alpha,\beta\in\mathbb{R},}.
$$

---



## 2) Determinar un vector normal al plano

Un vector normal al plano se obtiene con el producto cruz:

$$
n=v_{1}\times v_{2}.
$$



Calculamos

$$
n=
\begin{vmatrix}
\mathbf{i} & \mathbf{j} & \mathbf{k} \\ 
1 & -1 & 1 \\
0 & 1 & -1
\end{vmatrix}.
$$



Desarrollando:

$$
n=
\mathbf{i}
\begin{vmatrix}
-1 & 1 \\ 
1 & -1
\end{vmatrix}
-\mathbf{j}
\begin{vmatrix}
1 & 1 \\ 
0 & -1
\end{vmatrix}
+\mathbf{k}
\begin{vmatrix}
1 & -1 \\ 
0 & 1
\end{vmatrix}.
$$



$$
\mathbf{i}\big((-1)(-1)-1\cdot 1\big)
-\mathbf{j}\big(1(-1)-1\cdot 0\big)
+\mathbf{k}\big(1\cdot 1-(-1)\cdot 0\big).
$$

# $$

\mathbf{i}(1-1)-\mathbf{j}(-1)+\mathbf{k}(1).
$$

$$
=(0,1,1).
$$

Por tanto, un vector normal al plano es

$$
n=(0,1,1).
$$

---



## 3) Ecuación del plano generado

El plano que pasa por el origen y tiene vector normal $n$ cumple

$$
0x+1y+1z=0.
$$

Es decir,

$$
y+z=0.
$$

Este es el plano generado por $v_{1}$ y $v_{2}$.

---



## 4) Construcción de una base de $\mathbb{R}^{3}$

Los vectores $v_{1}$ y $v_{2}$ están en el plano.
El vector normal $n$ es perpendicular al plano, por lo que no pertenece a él.

Así, el conjunto

$$
\{ v_{1},v_{2}, n \}
$$

es linealmente independiente.

Como contiene tres vectores linealmente independientes en $\mathbb{R}^{3}$, forma una base.

---



## 5) Base pedida

Una base de $\mathbb{R}^{3}$ que contiene a $S$ es

$$
\mathcal{B} = \{ (1,-1,1), (0,1,-1), (0,1,1) \}.
$$

---



## 6) Interpretación geométrica

* $v_{1}$ y $v_{2}$ generan un plano que pasa por el origen.
* $n$ es perpendicular a ese plano.
* El plano más su normal $\mathbf{n}$ generan todo el espacio.

Por tanto,

$$
\mathbb{R}^{3}=\text{span} \{\mathbf{v}_{1}, \mathbf{v}_{2}, \mathbf{n} \}.
$$

# Estado del arte



El interés por integrar la tecnología en la enseñanza de las matemáticas tiene raíces en las propuestas de **Papert (1980, 1996)**, quien con su enfoque constructionista concibió la computadora como un “objeto para pensar”, capaz de mediar la construcción activa de significados matemáticos mediante la programación. 

Esta perspectiva se enlaza con los planteamientos de **Piaget (1970)** y **Vygotsky (1978)**, quienes reconocieron la importancia de la experimentación y de las herramientas culturales en la formación del pensamiento. Bajo esta mirada, el uso de entornos digitales no constituye una amenaza para la enseñanza tradicional de las matemáticas, sino un recurso que amplía las posibilidades de aprendizaje.



A comienzos del siglo XXI, la noción de **pensamiento computacional** introducida por **Wing (2006)** renovó el debate al proponer que esta competencia debía enseñarse de manera transversal desde los niveles iniciales hasta la educación superior, situándola al mismo nivel que la lectura, la escritura y las matemáticas. Este enfoque se alinea con los planteamientos del **NCTM (2000, 2014)**, que destacó la tecnología como un componente esencial para la experimentación, la representación y la comunicación en matemáticas. En paralelo, aportes teóricos como la **teoría de registros de representación semiótica de Duval (1993, 2006)** y el **Enfoque Ontosemiótico de Godino (2007, 2017)** muestran que el aprendizaje matemático se enriquece al transitar entre múltiples sistemas de representación, incluyendo los digitales.



En el contexto actual, organizaciones como la **ACM** y el **IEEE** insisten en que las competencias computacionales son indispensables en la formación profesional, lo que refuerza la pertinencia de integrar lenguajes como **Python, LaTeX, Markdown e inglés** en ambientes como **Jupyter Notebooks** para la enseñanza universitaria de las matemáticas. 

Así, el estado del arte evidencia un consenso creciente: lejos de ser un obstáculo, el pensamiento computacional y el uso de lenguajes digitales constituyen una vía para fortalecer la comprensión matemática, formar profesionales más competentes y responder a las demandas del mundo laboral y científico contemporáneo.

El uso de la tecnología en la enseñanza de las matemáticas encuentra uno de sus pilares en **Papert (1980)**, quien afirmaba que *“los niños no aprenden sobre la computadora, aprenden con la computadora”* (p. 6). En su visión, la programación se convierte en un medio para la construcción activa de significados matemáticos, idea que se enlaza con la epistemología genética de **Piaget (1970)**, quien sostuvo que *“el conocimiento se construye a través de la acción”* (p. 15), y con la perspectiva de **Vygotsky (1978)**, al señalar que *“toda función en el desarrollo cultural del niño aparece dos veces: primero en el plano social y después en el individual”* (p. 57). Finalmente, organismos internacionales como la **ACM y el IEEE (2013)** enfatizan que *“las habilidades computacionales son esenciales para el ejercicio profesional en cualquier disciplina”* (p. 12). De esta manera, se consolida un consenso: la enseñanza universitaria de las matemáticas debe integrar herramientas contemporáneas como **Jupyter Notebooks, Python, LaTeX, Markdown e inglés**, no para competir con la enseñanza tradicional, sino para **fortalecer la comprensión matemática, potenciar la comunicación científica y responder a las demandas del mundo laboral**.

# 1. Problema motivador (datos agrícolas)

Un sistema de monitoreo agrícola registra:

* radiación solar
* temperatura
* evapotranspiración

Después de analizar datos históricos se encuentra aproximadamente

$$
evapotranspiración \approx 0.6\cdot radiación + 0.4\cdot temperatura
$$

Entonces realmente **no hay tres variables independientes**.

La pregunta clave del módulo:

**¿Cuántas direcciones realmente describen el fenómeno?**

---

# 2. Exploración con Python

Simulemos este escenario.

```python
import numpy as np
import pandas as pd

np.random.seed(10)

radiacion = np.random.normal(size=300)
temperatura = np.random.normal(size=300)

evapo = 0.6*radiacion + 0.4*temperatura + np.random.normal(scale=0.05,size=300)

df = pd.DataFrame({
"radiacion":radiacion,
"temperatura":temperatura,
"evapotranspiracion":evapo
})

df.head()
```

---

# 3. Observación geométrica

Cada observación es un vector

$$
x = (radiación, temperatura, evapotranspiración)
$$

Pero los datos viven cerca del plano

$$
z = 0.6x + 0.4y
$$

Esto significa que la dimensión real del fenómeno es cercana a **2**.

---

```python
import plotly.express as px

fig = px.scatter_3d(df,
x="radiacion",

fig.show()
```

Discusión:

Los puntos se organizan cerca de un plano.

$$

Esto es fundamental en:

# 6. Recordatorio

Ya sabemos:

* subespacios

Ahora aparece una nueva pregunta:

**¿Cuál es el conjunto mínimo de vectores que genera todo el espacio?**

---

# 7. Definición de base

Un conjunto

$$
{v_1,v_2,\dots,v_k}
$$

es una **base** de un espacio vectorial si

1. Genera el espacio

$$
span{v_1,\dots,v_k} = V
$$

2. Los vectores son linealmente independientes.

---

# 8. Interpretación intuitiva

Una base es el **sistema de coordenadas más eficiente**.

Ejemplo en $\mathbb{R}^2$:

Base clásica

$$
(1,0), (0,1)
$$

Pero también es base

$$
(1,1), (1,-1)
$$

---

# 9. Experimento con Python

```python
import numpy as np

v1 = np.array([1,1])
v2 = np.array([1,-1])

A = np.column_stack((v1,v2))

np.linalg.matrix_rank(A)
```

Resultado esperado

2.

Por lo tanto forman una base.

---

# 10. Interpretación geométrica

Dos vectores independientes en $\mathbb{R}^2$ generan todo el plano.

Tres vectores independientes en $\mathbb{R}^3$ generan el espacio.

---

# 11. Dimensión

La **dimensión** de un espacio vectorial es el número de vectores en cualquier base.

Ejemplos

$$
dim(\mathbb{R}^2)=2
$$

$$
dim(\mathbb{R}^3)=3
$$

---

# 12. Interpretación en ciencia de datos

Dimensión significa:

**cuántas variables independientes existen realmente**.

---

# 13. Experimento computacional

Calculemos el rango del dataset.

```python
X = df.values

np.linalg.matrix_rank(X)
```

Probablemente obtendremos

2.

Esto revela la dimensión del subespacio donde viven los datos.

---

# 14. Idea profunda usada en IA

Los datos reales viven en **variedades o subespacios de baja dimensión** dentro de espacios gigantes.

Ejemplo:

Una imagen de $100 \times 100$ píxeles vive en

$$
\mathbb{R}^{10000}
$$

pero la dimensión efectiva es muchísimo menor.

---

# 15. Introducción a PCA

La técnica más famosa para descubrir la base de los datos es:

**Análisis de Componentes Principales.**

La idea:

Encontrar nuevas direcciones

$$
v_1, v_2, \dots
$$

que capturen la mayor variabilidad posible.

---

# 16. Ejemplo de PCA

```python
from sklearn.decomposition import PCA

pca = PCA()

pca.fit(df)
```

---

# 17. Varianza explicada

```python
pca.explained_variance_ratio_
```

Discusión:

Si las dos primeras componentes explican casi toda la varianza, la dimensión efectiva es 2.

---

# 18. Nueva base del dataset

```python
X_pca = pca.transform(df)
```

Ahora los datos están expresados en una **nueva base**.

Esta base está adaptada al fenómeno.

---

# 19. Compresión de datos

Podemos mantener solo dos componentes.

```python
pca = PCA(n_components=2)

X_reducido = pca.fit_transform(df)
```

Pasamos de 3 variables a 2 sin perder casi información.

Esto es compresión.

---
