# 5.1 Definición y ejemplos de bases   



# Definición 1  

Un conjunto finito de vectores $\{\mathbf{v}_{1}, \mathbf{v}_{2}, \ldots, \mathbf{v}_{n}\}$ es una *base* para un espacio vectorial $V$ si:  

1. $\{\mathbf{v}_{1}, \mathbf{v}_{2}, \ldots, \mathbf{v}_{n}\}$ es linealmente independiente. 
2. $\{\mathbf{v}_{1}, \mathbf{v}_{2}, \ldots, \mathbf{v}_{n}\}$ genera a $V$

2.  En $\mathbb{P}_{n}$ la  base estándar está formada por el conjunto $\{1, x , x^{2}, \ldots, x^{n} \}$. ¿Por qué?

En el espacio vectorial de los polinomios de grado a lo más $n$, denotado por $\mathbb{P}_{n}$, la **base estándar** está formada por el conjunto

$$
\{ 1, x, x^{2}, \ldots, x^{n} \}
$$

porque este conjunto **cumple las dos propiedades que definen una base de un espacio vectorial**:

1. **Genera todo el espacio $\mathbb{P}_{n}$**.
2. **Sus elementos son linealmente independientes**.




### 1. El conjunto genera $\mathbb{P}_{n}$

Un polinomio cualquiera de $\mathbb{P}_{n}$ tiene la forma

$$
p(x) = a_0 + a_1 x + a_2 x^2 + \cdots + a_n x^n
$$

donde $a_0, a_1, \ldots, a_n$ son números reales (o del campo sobre el cual esté definido el espacio).

Este polinomio puede escribirse como **combinación lineal** de los elementos del conjunto $\{1, x, x^2, \ldots, x^n\}$:

$$
p(x) = a_0(1) + a_1(x) + a_2(x^2) + \cdots + a_n(x^n)
$$

Por lo tanto, **todo polinomio de $\mathbb{P}_n$ puede construirse a partir de estos elementos**, lo que significa que **el conjunto genera el espacio**.

---



### 2. El conjunto es linealmente independiente

Consideremos una combinación lineal igual a cero:

$$
a_0(1) + a_1(x) + a_2(x^2) + \cdots + a_n(x^n) = 0
$$

Esto significa que el polinomio

$$
a_0 + a_1 x + a_2 x^2 + \cdots + a_n x^n
$$

es el **polinomio cero**, es decir, el polinomio que vale cero para todo $x$.

La única forma en que un polinomio sea identicamente cero es que **todos sus coeficientes sean cero**, es decir,

$$
a_0 = a_1 = a_2 = \cdots = a_n = 0
$$

Por lo tanto, los polinomios ${1, x, x^2, \ldots, x^n}$ son **linealmente independientes**.

---



### 3. Conclusión

El conjunto

$$
\{1, x, x^{2}, \ldots, x^{n}\}
$$

es una **base de $\mathbb{P}_n$** porque:

* **genera todo el espacio de polinomios de grado $\le n$**, y
* **sus elementos son linealmente independientes**.



Además, el número de elementos de esta base es $n+1$, por lo que

$$
\dim(\mathbb{P}_n) = n + 1
$$

y cada polinomio queda **representado de manera única** mediante sus coeficientes respecto a esta base.

---


````python
# mostremos que 1,x,x2 es una base para P2

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# Dominio
x = np.linspace(-5, 5, 400)

# =========================
# Widgets
# =========================
a0_box = widgets.FloatText(value=1, description='a₀:')
a1_box = widgets.FloatText(value=0, description='a₁:')
a2_box = widgets.FloatText(value=1, description='a₂:')

alpha_slider = widgets.IntSlider(min=-5, max=5, step=1, value=0, description="α")
beta_slider  = widgets.IntSlider(min=-5, max=5, step=1, value=0, description="β")
gamma_slider = widgets.IntSlider(min=-5, max=5, step=1, value=0, description="γ")

# =========================
# Función de visualización
# =========================
def visualizar(a0, a1, a2, alpha, beta, gamma):
    
    # Polinomio del usuario
    P = a0 + a1*x + a2*x**2
    
    # Combinación lineal de {1, x, x²}
    combinacion = alpha*1 + beta*x + gamma*x**2
    
    plt.figure(figsize=(7,5))
    
    plt.plot(x, P, linewidth=3, label="Polinomio del usuario P(x)")
    plt.plot(x, combinacion, linestyle='--', linewidth=3,
             label="α·1 + β·x + γ·x²")
    
    plt.axhline(0)
    plt.axvline(0)
    plt.grid(True)
    plt.legend()
    plt.title("Todo polinomio de grado ≤ 2 es combinación lineal de {1, x, x²}")
    plt.xlabel("x")
    plt.ylabel("y")
    
    print("Polinomio ingresado:",
          f"P(x) = {a0} + {a1}x + {a2}x²")
    print("Combinación lineal:",
          f"{alpha}·1 + {beta}·x + {gamma}·x²")
    
    if alpha == a0 and beta == a1 and gamma == a2:
        print("✓ Coinciden exactamente.")
    
    plt.show()

# =========================
# Conectar widgets con la gráfica
# =========================
ui = widgets.VBox([
    widgets.HBox([a0_box, a1_box, a2_box]),
    widgets.HBox([alpha_slider, beta_slider, gamma_slider])
])

out = widgets.interactive_output(
    visualizar,
    {
        'a0': a0_box,
        'a1': a1_box,
        'a2': a2_box,
        'alpha': alpha_slider,
        'beta': beta_slider,
        'gamma': gamma_slider
    }
)

display(ui, out)
````

3. En $M_{22}$ el conjunto de matrices $\left\{ \begin{pmatrix}1&0\\0&0 \end{pmatrix}, \begin{pmatrix}0&1\\0&0 \end{pmatrix}, \begin{pmatrix}0&0\\1&0 \end{pmatrix}, \begin{pmatrix}0&0\\0&1 \end{pmatrix}  \right\}$ es una base para el espacio vectorial $M_{22}$. Ésta es la base estándar o canónica.

En el espacio vectorial $M_{22}$ de todas las matrices $2\times 2$ con entradas reales, el conjunto

$$
\left\{
\begin{pmatrix}1&0\\ 0&0 \end{pmatrix},
\begin{pmatrix}0&1\\ 0&0 \end{pmatrix},
\begin{pmatrix}0&0\\ 1&0 \end{pmatrix},
\begin{pmatrix}0&0\\ 0&1 \end{pmatrix}
\right\}
$$

es una **base** porque cumple las dos condiciones necesarias: **genera todo el espacio $M_{22}$** y **sus elementos son linealmente independientes**.

---



## 1. El conjunto genera el espacio $M_{22}$

Una matriz cualquiera de $M_{22}$ tiene la forma

$$
A=
\begin{pmatrix}
a & b \\ 
c & d
\end{pmatrix}
$$

donde $a,b,c,d\in\mathbb{R}$.



Esta matriz puede escribirse como combinación lineal de las matrices del conjunto dado:

$$
\begin{pmatrix}
a & b \\ 
c & d
\end{pmatrix}
=
a
\begin{pmatrix}1&0\\ 0&0\end{pmatrix}
+
b
\begin{pmatrix}0&1\\ 0&0\end{pmatrix}
+
c
\begin{pmatrix}0&0\\ 1&0\end{pmatrix}
+
d
\begin{pmatrix}0&0\\ 0&1\end{pmatrix}
$$

Por lo tanto, **toda matriz $2\times2$ puede expresarse como combinación lineal de estas cuatro matrices**, lo que demuestra que **el conjunto genera $M_{22}$**.

---



## 2. El conjunto es linealmente independiente

Consideremos una combinación lineal igual a la matriz cero:

$$
\alpha
\begin{pmatrix}1&0\\ 0&0\end{pmatrix}
+
\beta
\begin{pmatrix}0&1\\ 0&0\end{pmatrix}
+
\gamma
\begin{pmatrix}0&0\\ 1&0\end{pmatrix}
+
\delta
\begin{pmatrix}0&0\\ 0&1\end{pmatrix}
=
\begin{pmatrix}0&0\\ 0&0\end{pmatrix}
$$



Al efectuar la suma se obtiene

$$
\begin{pmatrix}
\alpha & \beta \\ 
\gamma & \delta
\end{pmatrix}
=
\begin{pmatrix}
0 & 0 \\ 
0 & 0
\end{pmatrix}
$$



De aquí se deduce que

$$
\alpha=0,\quad \beta=0,\quad \gamma=0,\quad \delta=0
$$

lo que muestra que **la única combinación lineal que produce la matriz cero es la trivial**. Por tanto, las matrices son **linealmente independientes**.

---



## 3. Conclusión

El conjunto

$$
\left\{
\begin{pmatrix}1&0\\ 0&0 \end{pmatrix},
\begin{pmatrix}0&1\\ 0&0 \end{pmatrix},
\begin{pmatrix}0&0\\ 1&0 \end{pmatrix},
\begin{pmatrix}0&0\\ 0&1 \end{pmatrix}
\right\}
$$

es una **base de $M_{22}$** porque:

* **genera todo el espacio de matrices $2\times2$**,
* **sus elementos son linealmente independientes**.



Estas matrices se denominan **matrices elementales de posición** porque cada una tiene un $1$ en una posición distinta y ceros en las demás. Por esta razón constituyen la **base estándar o base canónica** de $M_{22}$.

Además, como esta base tiene **4 elementos**, se concluye que

$$
\dim(M_{22})=4
$$

lo cual coincide con el número de entradas independientes que tiene una matriz $2\times2$.

# Definición 2  


La dimensión de un espacio vectorial no nulo es el número de vectores en una base para $V$.  

Como el conjunto $\{ \mathbf{0} \}$ es linealmente dependiente, se dice que el espacio vectorial $\{ \mathbf{0} \}$ tiene dimensión cero.

Cuando la base del espacio vectorial $V$ es un conjunto finito de vectores, se dice que $V$ es de dimensión finita. 

Sin embargo, existen muchos espacios vectoriales para los cuales no hay una base con un número finito de vectores; estos espacios son de dimensión infinita, por ejemplo el espacio vectorial $\mathbb{P}$ de todos los polinomios y el espacio $C[a, b]$ de funciones continuas en $[a, b]$.  

La dimensión de $V$ se denota por $dim\ V$.

Este problema puede presentarse en una de estas dos formas:

# Ejemplo  

Sea $S = \{(1, -1, 1), (0, 1, -1)\}$. Obtenga una base de $\mathbb{R}^{3}$ que contenga a $S$.  

## Solución


Como $\mathbb{R}^{3}$ tiene dimensión tres, la base para $\mathbb{R}^{3}$ debe contener tres vectores.

El conjunto $S$ es LI, ya que los vectores de $S$ no son uno un múltiplo escalar del otro, luego lo que hace falta es añadir otro vector $\mathbf{v}$ tal que $\mathbf{v} \notin gen\  S$.

# Solución 1 utilizando el concepto de determinante   

Sea $S = \{ (1,-1,1), (0,1,-1) \}$. Queremos obtener una base de $\mathbb{R}^{3}$ que contenga a $S$.

---



## 1) Interpretación del problema

Buscamos un tercer vector que, junto con los dos vectores de $S$, forme un conjunto linealmente independiente y genere todo $\mathbb{R}^{3}$.

Sea

$$
v_{1}=(1,-1,1), \quad v_{2}=(0,1,-1).
$$

Debemos encontrar $v_{3}$ tal que $\{ v_{1}, v_{2}, v_{3} \}$ sea base de $\mathbb{R}^{3}$.

---



## 2) Condición de independencia lineal

Tomemos un vector general $v_{3}=(a,b,c)$ y verifiquemos cuándo

$$
\{ v_{1}, v_{2}, v_{3} \}
$$

es linealmente independiente. Para ello usamos el determinante de la matriz que los tiene como filas (o columnas):

$$
A=
\begin{pmatrix}
1 & -1 & 1 \\ 
0 & 1 & -1 \\ 
a & b & c
\end{pmatrix}.
$$



El conjunto será base de $\mathbb{R}^{3}$ si y solo si

$$
\det(A)\neq 0.
$$

---



## 3) Cálculo del determinante

$$
\det(A)=
\begin{vmatrix}
1 & -1 & 1 \\ 
0 & 1 & -1 \\ 
a & b & c
\end{vmatrix}.
$$



Desarrollando por la primera fila:

$$
\det(A)=
1
\begin{vmatrix}
1 & -1 \\ 
b & c
\end{vmatrix}
-(-1)
\begin{vmatrix}
0 & -1 \\ 
a & c
\end{vmatrix}
+1
\begin{vmatrix}
0 & 1 \\ 
a & b
\end{vmatrix}.
$$




$$
1(1\cdot c-(-1)b)
+1(0\cdot c-(-1)a)
+1(0\cdot b-1\cdot a).
$$

$$
= (c+b)+(a)+(-a)=b+c.
$$



Entonces la independencia lineal se cumple cuando

$$
b + c \neq 0.
$$

---



## 4) Elección de un tercer vector

Podemos escoger cualquier vector que cumpla $b+c\neq 0$.
Por ejemplo, tomemos

$$
v_{3}=(0,0,1).
$$



Aquí $b+c=0+1=1\neq 0$, por lo tanto el conjunto es linealmente independiente.

---



## 5) Base buscada

La base de $\mathbb{R}^{3}$ que contiene a $S$ es

$$
\mathcal{B}= \{ (1, -1, 1), (0, 1, -1), (0, 0, 1) \}.
$$

---


````python
import numpy as np
import plotly.graph_objects as go

# =========================
# Vectores de la base
# =========================
v1 = np.array([1, -1, 1])
v2 = np.array([0, 1, -1])
v3 = np.array([0, 0, 1])

# =========================
# Plano generado por v1 y v2
# =========================
s = np.linspace(-2, 2, 20)
t = np.linspace(-2, 2, 20)

S, T = np.meshgrid(s, t)

X = S*v1[0] + T*v2[0]
Y = S*v1[1] + T*v2[1]
Z = S*v1[2] + T*v2[2]

# =========================
# Figura
# =========================
fig = go.Figure()

# Plano
fig.add_trace(go.Surface(
    x=X,
    y=Y,
    z=Z,
    opacity=0.4,
    showscale=False,
    name="Plano generado por v1 y v2"
))

# Función para agregar flechas
def vector_arrow(v, name, color):
    
    fig.add_trace(go.Scatter3d(
        x=[0, v[0]],
        y=[0, v[1]],
        z=[0, v[2]],
        mode="lines+markers",
        line=dict(width=8, color=color),
        marker=dict(size=3),
        name=name
    ))
    
    fig.add_trace(go.Cone(
        x=[v[0]],
        y=[v[1]],
        z=[v[2]],
        u=[v[0]],
        v=[v[1]],
        w=[v[2]],
        sizemode="absolute",
        sizeref=0.3,
        showscale=False,
        colorscale=[[0,color],[1,color]]
    ))

# =========================
# Dibujar vectores
# =========================
vector_arrow(v1, "v1 = (1,-1,1)", "red")
vector_arrow(v2, "v2 = (0,1,-1)", "blue")
vector_arrow(v3, "v3 = (0,0,1)", "green")

# =========================
# Layout
# =========================
fig.update_layout(
    title="Base de R^3 y plano generado por dos vectores",
    scene=dict(
        xaxis_title="x",
        yaxis_title="y",
        zaxis_title="z",
        aspectmode="cube"
    )
)

fig.show()
fig.write_image(r"C:\Users\marco\Downloads\base_R3_y_plano.png")
````


## 6) Verificación conceptual

* $S$ tiene dos vectores linealmente independientes.
* Se añade un tercer vector que no pertenece al plano generado por ellos.
* El nuevo conjunto tiene tres vectores linealmente independientes en $\mathbb{R}^{3}$.

Por tanto,

$$
\mathcal{B}
$$

es una base de $\mathbb{R}^{3}$ que contiene al conjunto $S$.


# Segundo método de solución, utilizando el concepto de producto vectorial y la definición del plano como conjunto de vectores perpendiculares a un mismo vector  



Sea $S=\{ (1,-1,1),(0,1,-1) \}$. Queremos obtener una base de $\mathbb{R}^{3}$ que contenga a $S$, determinando primero el plano que generan estos vectores y luego tomando como tercer vector un normal a dicho plano.

---



## 1) Vectores generadores del plano

Sean

$$
v_{1} = (1, -1, 1), \quad v_{2} = (0, 1, -1).
$$

El plano generado por ellos es el conjunto de todas las combinaciones lineales:

$$
\Pi = \{ \alpha v_{1} + \beta v_{2} \mid \alpha, \beta \in \mathbb{R} \}.
$$

---



## 2) Determinar un vector normal al plano

Un vector normal al plano se obtiene con el producto cruz:

$$
n=v_{1}\times v_{2}.
$$



Calculamos

$$
n=
\begin{vmatrix}
\mathbf{i} & \mathbf{j} & \mathbf{k} \\ 
1 & -1 & 1 \\
0 & 1 & -1
\end{vmatrix}.
$$



Desarrollando:

$$
n=
\mathbf{i}
\begin{vmatrix}
-1 & 1 \\ 
1 & -1
\end{vmatrix}
-\mathbf{j}
\begin{vmatrix}
1 & 1 \\ 
0 & -1
\end{vmatrix}
+\mathbf{k}
\begin{vmatrix}
1 & -1 \\ 
0 & 1
\end{vmatrix}.
$$



$$
\mathbf{i}\big((-1)(-1)-1\cdot 1\big)
-\mathbf{j}\big(1(-1)-1\cdot 0\big)
+\mathbf{k}\big(1\cdot 1-(-1)\cdot 0\big).
$$




$$
\mathbf{i}(1-1)-\mathbf{j}(-1)+\mathbf{k}(1).
$$

$$
=(0,1,1).
$$



Por tanto, un vector normal al plano es

$$
n=(0,1,1).
$$

---



## 3) Ecuación del plano generado

El plano que pasa por el origen y tiene vector normal $n$ cumple

$$
0x+1y+1z=0.
$$

Es decir,

$$
y+z=0.
$$

Este es el plano generado por $v_{1}$ y $v_{2}$.

---



## 4) Construcción de una base de $\mathbb{R}^{3}$

Los vectores $v_{1}$ y $v_{2}$ están en el plano.
El vector normal $n$ es perpendicular al plano, por lo que no pertenece a él.

Así, el conjunto

$$
\{ \mathbf{v}_{1}, \mathbf{v}_{2}, \mathbf{n} \}
$$

es linealmente independiente.

Como contiene tres vectores linealmente independientes en $\mathbb{R}^{3}$, forma una base.

---



## 5) Base pedida

Una base de $\mathbb{R}^{3}$ que contiene a $S$ es

$$
\mathcal{B} = \{ (1,-1,1), (0,1,-1), (0,1,1) \}.
$$

---



## 6) Interpretación geométrica

* $v_{1}$ y $v_{2}$ generan un plano que pasa por el origen.
* $n$ es perpendicular a ese plano.
* El plano más su normal $\mathbf{n}$ generan todo el espacio.

Por tanto,

$$
\mathbb{R}^{3}=\text{span} \{\mathbf{v}_{1}, \mathbf{v}_{2}, \mathbf{n} \}.
$$

# 7. Definición de base

Un conjunto

$$
\{ v_1, v_2, \dots, v_k \} 
$$

es una **base** de un espacio vectorial si



1. Genera el espacio

$$
span\{v_1, \dots, v_k \} = V
$$

2. Los vectores son linealmente independientes.

---



# 8. Interpretación intuitiva

Una base es el **sistema de coordenadas más eficiente**.

Ejemplo en $\mathbb{R}^2$:

Base clásica

$$
(1,0), (0,1)
$$



Pero también es base

$$
(1,1), (1,-1)
$$

---



# 9. Experimento con Python



````python
import numpy as np

v1 = np.array([1,1])
v2 = np.array([1,-1])
v1, v2
````

````python

A = np.column_stack((v1,v2))
A 
````

````python

np.linalg.matrix_rank(A)
````

Resultado esperado

2.

Por lo tanto forman una base.

---



# 10. Interpretación geométrica

Dos vectores independientes en $\mathbb{R}^2$ generan todo el plano.

Tres vectores independientes en $\mathbb{R}^3$ generan el espacio.

---



# 11. Dimensión

La **dimensión** de un espacio vectorial es el número de vectores en cualquier base.

Ejemplos

$$
dim(\mathbb{R}^2)=2
$$

$$
dim(\mathbb{R}^3)=3
$$

---



# 12. Interpretación en ciencia de datos

Dimensión significa:

**cuántas variables independientes existen realmente**.

La técnica más famosa para descubrir la base de los datos es:

La idea:

$$

# 16. Ejemplo de PCA


````python
from sklearn.decomposition import PCA # Principal Component Analysis, es decir, análisis de componentes principales 
# o análisis de componentes principales que es lo mismo pero dicho de otra forma y consiste en 
# encontrar una base ortogonal que explique la mayor parte de la varianza de los datos, es decir, encontrar una 
# base que permita representar los datos con la menor cantidad de dimensiones posible sin perder mucha información.


pca = PCA()

pca.fit(df)
````


# 17. Varianza explicada

En términos matemáticos, buscamos la **dimensión del subespacio generado por las variables**, lo cual puede estudiarse con:

# 1. Generación artificial del dataset

Supongamos que en realidad la evapotranspiración depende principalmente de:

* radiación
* temperatura
* viento

y que las otras variables están correlacionadas.



````python
import numpy as np
import pandas as pd

np.random.seed(123)

n = 200

# Variables independientes básicas
radiacion = np.random.normal(500, 50, n)
temperatura = np.random.normal(28, 3, n)
viento = np.random.normal(10, 2, n)

# Variables correlacionadas
humedad = 80 - 0.2*temperatura + np.random.normal(0,2,n)
lluvia = np.maximum(0, 5 - 0.01*radiacion + np.random.normal(0,1,n))

# Modelo físico aproximado de evapotranspiración
evapotranspiracion = (
    0.03*radiacion +
    0.8*temperatura +
    0.5*viento +
    np.random.normal(0,2,n)
)

df = pd.DataFrame({
    "lluvia": lluvia,
    "humedad": humedad,
    "temperatura": temperatura,
    "radiacion": radiacion,
    "viento": viento,
    "evapotranspiracion": evapotranspiracion
})

df.head()
````


# 2. Matriz de datos

Sea

$$
X=
\begin{pmatrix}
x_{11} & x_{12} & \cdots & x_{16}\\ 
x_{21} & x_{22} & \cdots & x_{26}\\ 
\vdots & \vdots & & \vdots
\end{pmatrix}
$$

donde cada columna es una variable.

La **dimensión real del fenómeno** corresponde al

$$
\text{rango}(X)
$$

---



# 3. Cálculo del rango

```python


````python
from numpy.linalg import matrix_rank

X = df.values

rango = matrix_rank(X)

print("Dimensión del fenómeno:", rango)
````


Interpretación posible:

* rango = 6 → todas las variables independientes
* rango < 6 → hay **dependencia entre variables**

---

# 5. PCA para estimar la dimensión real

Aplicamos Análisis de Componentes Principales.

```python
from sklearn.decomposition import PCA

pca = PCA()

pca.fit(df)

varianza = pca.explained_variance_ratio_

print(varianza)
```

# 6. Gráfica de varianza explicada

```python
import matplotlib.pyplot as plt

plt.ylabel("Varianza acumulada")
plt.title("Dimensión efectiva del fenómeno")

plt.grid()
plt.show()
```

# 7. Interpretación científica

Supongamos que obtenemos algo como

| componente | varianza |
| ---------- | -------- |
| PC1        | 0.60     |
| PC2        | 0.25     |
| PC3        | 0.10     |
| PC4        | 0.03     |

Por tanto:

$$
\text{dimensión real del fenómeno} \approx 3
$$
