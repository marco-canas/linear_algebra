# 8.1 Conjuntos ortonormales  



## Definición 1 (Definición de subconjunto ortonormal)  



En $\mathbb{R}^{n}$ (con el producto interno usual), un **subconjunto ortonormal** es un conjunto de vectores que son mutuamente ortogonales y además tienen norma unitaria.



Formalmente, un conjunto de vectores
$$
{v_1, v_2, \dots, v_k} \subset \mathbb{R}^n
$$
se dice **ortonormal** si cumple:

1. **Ortogonalidad:**
   $$
   v_i \cdot v_j = 0 \quad \text{si } i \neq j
   $$

2. **Normalización (norma 1):**
   $$
   |v_i| = 1 \quad \text{para todo } i
   $$



De manera equivalente, ambas condiciones se resumen como:
$$
v_i \cdot v_j =
\begin{cases}
1 & \text{si } i = j,\\  
0 & \text{si } i \neq j.
\end{cases}
$$



Es decir, los vectores del subconjunto son perpendiculares entre sí y cada uno tiene longitud igual a 1.



Un ejemplo típico en $\mathbb{R}^n$ es la base canónica:
$$
e_1=(1,0,\dots,0),\quad e_2=(0,1,\dots,0),\ \dots,\ e_n=(0,0,\dots,1),
$$
que forma un subconjunto ortonormal de $\mathbb{R}^n$.


# La norma en $\mathbb{R}^{n}$.  



En $\mathbb{R}^n$, la **norma** (o longitud) de un vector se determina usando la norma euclidiana, definida a partir del producto interno usual.

Si
$$
v=(v_1,v_2,\dots,v_n)\in \mathbb{R}^n,
$$


entonces su norma es:

$$
|v|=\sqrt{v_1^2+v_2^2+\cdots+v_n^2}.
$$



Equivalentemente, usando producto punto:

$$
|v|=\sqrt{v\cdot v}.
$$



### Interpretación

La norma mide la longitud del vector en el espacio. Es la generalización del teorema de Pitágoras a $n$ dimensiones.

Nota:

# Recordemos las propiedades del producto escalar en $\mathbb{R}^{n}$ 



En $\mathbb{R}^n$, el **producto escalar** (o producto punto) entre dos vectores
$v=(v_1,\dots,v_n)$ y $w=(w_1,\dots,w_n)$ se define como

$$
v\cdot w = v_1w_1 + v_2w_2 + \cdots + v_nw_n.
$$



Este producto satisface las siguientes propiedades fundamentales:



### 1. Conmutatividad

$$
v\cdot w = w\cdot v.
$$



### 2. Linealidad en cada argumento

Para $u, v, w \in \mathbb{R}^n$ y $\lambda\in\mathbb{R}$:

* Respecto a la suma:
  $$
  (v+u)\cdot w = v\cdot w + u\cdot w,
  $$
  $$
  v\cdot (w+u) = v\cdot w + v\cdot u.
  $$



* Respecto al producto por escalar:
  $$
  (\lambda v)\cdot w = \lambda (v\cdot w),
  $$
  $$
  v\cdot (\lambda w) = \lambda (v\cdot w).
  $$



### 3. Positividad

$$
v\cdot v \ge 0 \quad \text{para todo } v\in\mathbb{R}^n.
$$


````python
v.dot(v)
````


### 4. Definido positivo

$$
v\cdot v = 0 \iff v=0.
$$


````python
np.array([0,0,0]).dot(np.array([0,0,0]))
````


### 5. Relación con la norma

El producto escalar determina la norma euclidiana:
$$
|v|=\sqrt{v\cdot v}.
$$



### 6. Ortogonalidad

Dos vectores son ortogonales si y solo si:
$$
v\cdot w = 0.
$$



Estas propiedades hacen del producto escalar la herramienta central para medir:
* longitudes, 
* ángulos y 
* proyecciones 
  
en $\mathbb{R}^n$.


# Ejemplo 1 (Ejemplo de un conjunto ortogonal en $\mathbb{R}^{3}$ y el procedimiento para normalizarlo)  

Consideremos el siguiente conjunto de tres vectores en $\mathbb{R}^3$:

$$
v_1=(1,1,0), \quad v_2=(1,-1,0), \quad v_3=(0,0,2).
$$



### 1. Verificación de que es un conjunto ortogonal

Calculamos los productos escalares:

$$
v_1\cdot v_2 = 1\cdot 1 + 1\cdot(-1) + 0\cdot 0 = 1-1+0=0,
$$

$$
v_1\cdot v_3 = 1\cdot 0 + 1\cdot 0 + 0\cdot 2 = 0,
$$

$$
v_2\cdot v_3 = 1\cdot 0 + (-1)\cdot 0 + 0\cdot 2 = 0.
$$

Como todos los productos cruzados son cero, el conjunto es **ortogonal**.

---


````python
import numpy as np
v1 = np.array([1,1,0])
v2 = np.array([1,-1,0])
v3 = np.array([0,0,2])

v1.dot(v2), v1.dot(v3), v2.dot(v3)
````


### 2. Cálculo de las normas

$$
|v_1|=\sqrt{1^2+1^2+0^2}=\sqrt{2},
$$

$$
|v_2|=\sqrt{1^2+(-1)^2+0^2}=\sqrt{2},
$$

$$
|v_3|=\sqrt{0^2+0^2+2^2}=2.
$$

---


````python
np.linalg.norm(v1), np.linalg.norm(v2), np.linalg.norm(v3)
````


### 3. Procedimiento de normalización

Para normalizar un vector se divide por su norma:

$$
u_i=\frac{v_i}{|v_i|}.
$$

Aplicando esto:

$$
u_1=\frac{1}{\sqrt{2}}(1,1,0)=\left(\frac{1}{\sqrt{2}},\frac{1}{\sqrt{2}},0\right),
$$

$$
u_2=\frac{1}{\sqrt{2}}(1,-1,0)=\left(\frac{1}{\sqrt{2}},-\frac{1}{\sqrt{2}},0\right),
$$

$$
u_3=\frac{1}{2}(0,0,2)=(0,0,1).
$$

---


````python
u1 = v1/np.linalg.norm(v1)
u2 = v2/np.linalg.norm(v2)
u3 = v3/np.linalg.norm(v3)
````

````python
u1, u2, u3
````


### 4. Conjunto ortonormal obtenido

El conjunto normalizado es:

$$
\left\{
\left(\frac{1}{\sqrt{2}},\frac{1}{\sqrt{2}},0\right),
\left(\frac{1}{\sqrt{2}},-\frac{1}{\sqrt{2}},0\right),
(0,0,1)
\right\},
$$

el cual es un **conjunto ortonormal** en $\mathbb{R}^3$ porque:

* los vectores siguen siendo ortogonales entre sí,
* y cada uno tiene norma igual a $1$.

# 8.2 Otra forma del problema de la base. Que se resuelve con El Proceso de Ortonormalización de Gram - Schmidt   



## Planteamiento del problema

Dada una base $S = \{v_{1}, v_{2}, \ldots, v_{k} \}$ para un subespacio de $\mathbb{R}^{n}$, hallar una base ortonormal $B = \{u_{1}, u_{2}, \ldots, u_{k} \}$ para este mismo subespacio.  



La solución a este problema nos la brinda el siguiente teorema, conocido como proceso de ortonormalización de Gram-Schmidt



El **proceso de ortonormalización de Gram–Schmidt** permite transformar una base cualquiera de un subespacio de $\mathbb{R}^n$ en una **base ortonormal** del mismo subespacio, conservando el espacio generado.

---



## Planteamiento general

Sea
$$
S=\{v_1,v_2,\dots,v_k\}
$$
una base de un subespacio $W\subseteq\mathbb{R}^n$ (vectores linealmente independientes).



El objetivo es construir un conjunto
$$
B=\{u_1,u_2,\dots,u_k\}
$$
tal que:

* genere el mismo subespacio $W$,
* los vectores sean ortogonales entre sí,
* y cada vector tenga norma 1 (ortonormalidad).

El procedimiento tiene dos etapas: **ortogonalización** y **normalización**.

---



# Paso 1: Construcción de un sistema ortogonal



Se construyen vectores $w_1, w_2, \dots, w_k$ ortogonales entre sí a partir de los $v_i$.



### Primer vector

Se toma directamente:
$$
w_1 = v_1.
$$



### Segundo vector

Se elimina de $v_2$ la componente en la dirección de $w_1$:

$$
w_2 = v_2 - \operatorname{proj}_{w_1}(v_2),
$$

donde la proyección es:

$$
\operatorname{proj}_{w_1}(v_2)=
\frac{v_2\cdot w_1}{w_1\cdot w_1}w_1.
$$

---


````python
import numpy as np
import matplotlib.pyplot as plt

# =========================
# Vectores en el plano
# =========================
v1 = np.array([4, 2])     # más largo
v2 = np.array([1, 3])

# =========================
# Proyección de v2 sobre v1
# =========================
proj_v2_v1 = (np.dot(v2, v1) / np.dot(v1, v1)) * v1

# Vector ortogonal de Gram-Schmidt
w2 = v2 - proj_v2_v1

# =========================
# Crear figura
# =========================
plt.figure(figsize=(7,7))

# Ejes coordenados
plt.axhline(0)
plt.axvline(0)

# =========================
# Dibujar vectores
# =========================
plt.quiver(0, 0, v1[0], v1[1],
           angles='xy', scale_units='xy', scale=1,
           color='blue', linewidth=3, label=r'$v_1$')

plt.quiver(0, 0, v2[0], v2[1],
           angles='xy', scale_units='xy', scale=1,
           color='green', linewidth=3, label=r'$v_2$')

plt.quiver(0, 0, proj_v2_v1[0], proj_v2_v1[1],
           angles='xy', scale_units='xy', scale=1,
           color='red', linewidth=3,
           label=r'$\mathrm{proy}_{v_1}(v_2)$')

# Vector ortogonal desde la proyección hasta la punta de v2
plt.quiver(proj_v2_v1[0], proj_v2_v1[1], w2[0], w2[1],
           angles='xy', scale_units='xy', scale=1,
           color='purple', linewidth=3,
           label=r'$w_2 = v_2 - \mathrm{proy}_{v_1}(v_2)$')

# =========================
# Trazo de la proyección
# =========================
plt.plot([v2[0], proj_v2_v1[0]],
         [v2[1], proj_v2_v1[1]],
         linestyle='dashed',
         color='black',
         linewidth=2,
         label='Trazo de la proyección')

# =========================
# Ajustes visuales
# =========================
plt.xlim(-1, 5)
plt.ylim(-1, 5)
plt.gca().set_aspect('equal', adjustable='box')

plt.title('Paso de Gram–Schmidt en el plano')
plt.xlabel('x')
plt.ylabel('y')

plt.legend()
plt.grid(True)
plt.savefig(r"C:\Users\marco\Downloads\paso_gram_schmidt_en_el_plano.png")
plt.show()
````


### Tercer vector

Se eliminan las componentes en las direcciones de $w_1$ y $w_2$:

$$
w_3 = v_3 - \frac{v_3\cdot w_1}{w_1\cdot w_1}w_1 - \frac{v_3\cdot w_2}{w_2\cdot w_2}w_2.
$$

---



### Caso general

Para $j=1,\dots,k$:

$$
w_j = v_j - \sum_{i=1}^{j-1} \frac{v_j\cdot w_i}{w_i\cdot w_i}w_i.
$$



Con esto se obtiene un conjunto ortogonal:

$$
{w_1\dots w_k}.
$$

---



# Paso 2: Normalización

Se convierten los vectores ortogonales en unitarios:

$$
u_j=\frac{w_j}{|w_j|}, \quad j=1,\dots,k.
$$



Entonces:

$$
B= \{u_1,\dots,u_k \}
$$

es una base ortonormal del subespacio $W$.

---



# Idea geométrica del método

En cada paso:

* Se toma un vector nuevo $v_j$.
* Se le quitan las partes que apuntan en las direcciones ya construidas.
* El resultado $w_j$ queda perpendicular a todos los anteriores.
* Finalmente se normaliza.

Es como ir construyendo ejes perpendiculares “adaptados” al subespacio generado por los vectores originales.

---



# Propiedades importantes

1. El subespacio generado no cambia:  

$$
\text{gen}(S)=\text{gen}(B).
$$
$$
\text{span}(S)=\text{span}(B).
$$



2. Los vectores obtenidos son ortogonales:
   $$
   w_i\cdot w_j=0 \quad (i\neq j).
   $$



3. La base final es ortonormal:
   $$
   u_i\cdot u_j=
   \begin{cases}
   1 & i=j,\\ 
   0 & i\neq j.
   \end{cases}
   $$



4. El proceso funciona para cualquier base linealmente independiente en $\mathbb{R}^n$.

---



# Esquema operativo (receta práctica)

Dada una base $S= \{v_1,\dots,v_k \}$:

1. $w_1=v_1$


2. Para $j=2,\dots,k$:
   $$
   w_j=v_j-\sum_{i=1}^{j-1}\frac{v_j\cdot w_i}{w_i\cdot w_i}w_i
   $$


3. Normalizar:
   $$
   u_j=\frac{w_j}{|w_j|}
   $$


4. La base ortonormal buscada es:
   $$
   B=\{ u_1,\dots,u_k \}.
   $$

Este es el procedimiento estándar de Gram–Schmidt para construir bases ortonormales en $\mathbb{R}^n$.


# 8.3 Proyecciones ortogonales 



## Definición 2 ( Proyección ortogonal de un vector $v$ sobre un subespacio $H$ de $\mathbb{R}^{n}$ )  

Sea $H$ un subespacio de $\mathbb{R}^n$ y sea $v\in\mathbb{R}^n$.  

La **proyección ortogonal de $v$ sobre $H$** es el vector de $H$ que está “más cerca” de $v$ y cuya diferencia con $v$ es perpendicular a todo el subespacio.



Formalmente, el vector $\operatorname{proj}_H(v)\in H$ es la proyección ortogonal de $v$ sobre $H$ si satisface:

1. $\operatorname{proj}_H(v)\in H$,
2. $v-\operatorname{proj}_H(v)\perp H$, es decir,
   $$
   (v-\operatorname{proj}_H(v))\cdot h=0 \quad \text{para todo } h\in H.
   $$



Equivalentemente, si llamamos $p = \operatorname{proj}_H(v)$, entonces existe una descomposición única:

$$
v = p + q,
$$



donde

$$
p\in H \quad \text{y} \quad q \in H^\perp.
$$

---



### Expresión mediante una base ortonormal

Si $\{u_1,u_2,\dots,u_k\}$ es una base ortonormal de $H$, entonces la proyección ortogonal de $v$ sobre $H$ está dada por:

$$
\operatorname{proj}_H(v)
=
(v\cdot u_1)u_1
+
(v\cdot u_2)u_2
+\cdots+
(v\cdot u_k)u_k.
$$

Esta fórmula muestra que la proyección se obtiene sumando las componentes de $v$ en las direcciones ortonormales que generan el subespacio $H$.


Veamos un ejemplo concreto en $\mathbb{R}^3$ donde el subespacio $H$ es un plano.

---



# Ejemplo en $\mathbb{R}^3$

Consideremos el plano (subespacio)

$$
H = \{(x,y,z)\in\mathbb{R}^3: x+y+z=0 \},
$$

que pasa por el origen y tiene como vector normal

$$
n=(1,1,1). 
$$

Porque, los coeficientes de las variables en la ecuación canónica, son las componentes del vector normal al plano. 


Tomemos el vector

$$
v=(2,1,3), 
$$

y hallemos la proyección de $v$ sobre el subespacio $H$

# Primera forma de solución de este problema  

A continuación presentamos **otra forma de obtener** la proyección ortogonal de
$v=(2,1,3)$ sobre el plano

$$
H={(x,y,z)\in\mathbb{R}^3:;x+y+z=0},
$$

siguiendo tres pasos:

1. hallar una base del plano,
2. ortonormalizarla (Gram–Schmidt),
3. calcular la proyección usando la base ortonormal.

---



# 1. Hallar una base del plano $H$



El plano está dado por la ecuación:

$$
x+y+z=0.
$$

Despejamos, por ejemplo, $x=-y-z$:

$$
\begin{pmatrix}x\\ y\\ z\end{pmatrix} = \begin{pmatrix}-y-z\\ y\\ z\end{pmatrix} = y\begin{pmatrix}-1\\ 1\\ 0 \end{pmatrix}+ z\begin{pmatrix}-1 \\ 0 \\ 1 \end{pmatrix}.
$$



Por tanto, una base de $H$ es:

$$
\mathcal{B} = \{ w_1 = (-1,1,0), w_2 = (-1,0,1) \}.
$$

---



# 2. Ortonormalizar la base (Gram–Schmidt)

## Paso 1

Tomamos

$$
u_1=\frac{w_1}{|w_1|}.
$$

Como

$$
|w_1|=\sqrt{(-1)^2+1^2+0^2}=\sqrt{2},
$$

entonces

$$
u_1=\frac{1}{\sqrt{2}}(-1,1,0).
$$

---



## Paso 2

Ortogonalizamos $w_2$ respecto a $u_1$:

$$
\operatorname{proj}_{u_1}(w_2)=(w_2\cdot u_1)u_1.
$$



Calculamos:

$$
w_2\cdot u_1=
(-1,0,1)\cdot\frac{1}{\sqrt{2}}(-1,1,0)
=\frac{1}{\sqrt{2}}.
$$



Entonces

$$
\operatorname{proj}_{u_1}(w_2)
=\frac{1}{\sqrt{2}}u_1
=\frac{1}{2}(-1,1,0).
$$



Ahora:

$$
z_2=w_2-\operatorname{proj}_{u_1}(w_2)
=(-1,0,1)-\left(-\tfrac12,\tfrac12,0\right)
=\left(-\tfrac12,-\tfrac12,1\right).
$$



Su norma es:

$$
|z_2|=\sqrt{\tfrac14+\tfrac14+1}=\sqrt{\tfrac32}.
$$



Normalizamos:

$$
u_2=\frac{z_2}{|z_2|}
=\frac{1}{\sqrt{3/2}}\left(-\tfrac12,-\tfrac12,1\right)
=\left(-\frac{1}{\sqrt6},-\frac{1}{\sqrt6},\frac{2}{\sqrt6}\right).
$$

---



## Base ortonormal del plano

$$
\{u_1, u_2\}
$$




con
$$
u_1=\frac{1}{\sqrt2}(-1,1,0),\qquad
u_2=\left(-\frac{1}{\sqrt6},-\frac{1}{\sqrt6},\frac{2}{\sqrt6}\right).
$$

---



# 3. Proyección de $v$ sobre $H$

Si $\{ u_1, u_2 \}$ es una base ortonormal de $H$, entonces

$$
\operatorname{proj}_H(v)
= (v\cdot u_1)u_1 + (v\cdot u_2)u_2.
$$

---



## Cálculo de productos escalares

### Con $u_1$

$$
v\cdot u_1
=(2,1,3)\cdot\frac{1}{\sqrt2}(-1,1,0)
=\frac{-2+1}{\sqrt2}
=-\frac{1}{\sqrt2}.
$$



Entonces

$$
(v\cdot u_1)u_1
=-\frac{1}{\sqrt2}\cdot\frac{1}{\sqrt2}(-1,1,0)
=\left(\tfrac12,-\tfrac12,0\right).
$$

---



### Con $u_2$

$$
v\cdot u_2
=(2,1,3)\cdot\left(-\frac{1}{\sqrt6},-\frac{1}{\sqrt6},\frac{2}{\sqrt6}\right)
=\frac{-2-1+6}{\sqrt6}
=\frac{3}{\sqrt6}.
$$



Por tanto,

$$
(v\cdot u_2)u_2
=\frac{3}{\sqrt6}
\left(-\frac{1}{\sqrt6},-\frac{1}{\sqrt6},\frac{2}{\sqrt6}\right)
=\left(-\tfrac12,-\tfrac12,1\right).
$$

---



## Suma final

$$
\operatorname{proj}_H(v)
=\left(\tfrac12,-\tfrac12,0\right)
+\left(-\tfrac12,-\tfrac12,1\right)
=(0,-1,1).
$$

---



# Resultado

$$
\boxed{\operatorname{proj}_H(v)=(0,-1,1)}
$$

que coincide con el obtenido usando el vector normal.



Además,

$$
x + y + z = 0, \quad 0+(-1)+1=0,
$$

por lo que efectivamente pertenece al plano $H$.



## Proyección ortogonal de $v$ sobre el plano $H$

La proyección sobre un plano que pasa por el origen puede calcularse restando la componente de $v$ en la dirección del vector normal:

$$
\operatorname{proj}_H(v)=v-\operatorname{proj}_n(v),
$$



donde

$$
\operatorname{proj}_n(v)=\frac{v\cdot n}{n\cdot n},n.
$$



Calculamos:

$$
v\cdot n = 2+1+3=6,\qquad n\cdot n=1+1+1=3.
$$



Entonces

$$
\operatorname{proj}_n(v)=\frac{6}{3}(1,1,1)=(2,2,2).
$$



Finalmente,

$$
\operatorname{proj}_H(v)=v-(2,2,2)=(2,1,3)-(2,2,2)=(0,-1,1).
$$



Y se verifica que pertenece al plano:

$$
0+(-1)+1=0.
$$

# Teorema 3  (Expresión de una vector $v$ como suma de sus proyecciones con respecto a los vectores de una base ortonormal $B$ para $\mathbb{R}^{n}$)  



**Teorema (descomposición respecto de una base ortonormal de (\mathbb{R}^n)).**

Sea
$$
B=\{u_1,u_2,\dots,u_n\}
$$
una base ortonormal de $\mathbb{R}^n$, y sea $v\in\mathbb{R}^n$.
Entonces $v$ puede expresarse de manera única como suma de sus proyecciones sobre los vectores de la base:

$$
v=\sum_{i=1}^{n}(v\cdot u_i)u_i.
$$



Además, cada término

$$
(v\cdot u_i)u_i
$$

es la proyección ortogonal de $v$ sobre la recta generada por $u_i$:

$$
\operatorname{proj}_{ u_i }(v) = (v\cdot u_i)u_i.
$$



En consecuencia,

$$
v=\operatorname{proj}_{ u_1}(v)
+\operatorname{proj}_{u_2}(v)
+\cdots+
\operatorname{proj}_{u_n}(v).
$$



Esta descomposición es única y los coeficientes

$$
v \cdot u_1, v \cdot u_2, \dots, v \cdot u_n
$$

son precisamente las coordenadas de $v$ respecto de la base ortonormal $B$.


# Ejemplo 3  

Veamos un ejemplo concreto del teorema en $\mathbb{R}^3$ y su visualización con Plotly.

---



# Ejemplo en $\mathbb{R}^3$

Tomemos la base ortonormal canónica:

$$
B = \{e_1, e_2, e_3 \},\qquad
e_1=(1,0,0), \ e_2=(0,1,0),\ e_3=(0,0,1).
$$



Sea el vector

$$
v=(2,1,3)\in\mathbb{R}^3.
$$



## Proyecciones de $v$ sobre cada vector de la base



Como la base es ortonormal:

$$
\operatorname{proj}_{e_i}(v)=(v\cdot e_i)e_i.
$$



Calculamos:

$$
v\cdot e_1=2,\quad
v\cdot e_2=1,\quad
v\cdot e_3=3.
$$



Por tanto,

$$
\operatorname{proj}_{ e_1}(v)=2e_1=(2,0,0),
$$
$$
\operatorname{proj}_{ e_2}(v)=1e_2=(0,1,0),
$$
$$
\operatorname{proj}_{ e_3}(v)=3e_3=(0,0,3).
$$


# 8.4 Complemento ortogonal  



# Definición $3$  

Sea $H$ un subespacio de $\mathbb{R}^n$ con el producto escalar usual.

El **complemento ortogonal** de $H$, denotado por $H^\perp$, se define como el conjunto de todos los vectores de $\mathbb{R}^n$ que son ortogonales a cada vector de $H$:

$$
H^\perp=\{v\in\mathbb{R}^n:\ v\cdot h=0 \text{ para todo } h\in H\}.
$$

Es decir, un vector pertenece a $H^\perp$ si su producto escalar con cualquier vector del subespacio $H$ es cero.

---



### Propiedades fundamentales



1. $H^\perp$ es un subespacio de $\mathbb{R}^n$.

2. Se cumple:
   $$
   H\cap H^\perp={0}.
   $$

3. Todo vector $v\in\mathbb{R}^n$ puede descomponerse de manera única como
   $$
   v=p+q,
   $$

donde
   $$
   p\in H \quad \text{y} \quad q\in H^\perp.
   $$



4. Las dimensiones satisfacen:
   $$
   \dim(H)+\dim(H^\perp)=n.
   $$

5. Se cumple:
   $$
   (H^\perp)^\perp=H.
   $$



Geométricamente, $H^\perp$ contiene todas las direcciones “perpendiculares” al subespacio $H$; por ejemplo, si $H$ es un plano en $\mathbb{R}^3$, entonces $H^\perp$ es la recta generada por el vector normal a ese plano.



## Ejemplo en $\mathbb{R}^2$



Sea el subespacio

$$
H = \text{la recta generada por }\quad \quad  h = \begin{pmatrix} 2 \\  1 \end{pmatrix}.
$$

* Entonces $H = \{\alpha h : \alpha \in \mathbb{R} \}$.


* El **complemento ortogonal** $H^\perp$ es la recta formada por todos los vectores ortogonales a $h$, es decir, todos los vectores $v$ que satisfacen

$$
v \cdot h = 0.
$$

* Un vector que genera $H^\perp$ es $w = \begin{pmatrix}-1 \ 2 \end{pmatrix}$, porque

$$
h \cdot w = 2(-1) + 1(2) = -2 + 2 = 0.
$$

* Entonces

$$
H^\perp = \{ \beta w : \beta \in \mathbb{R} \}.
$$

Geométricamente, $H$ y $H^\perp$ son **rectas perpendiculares** que pasan por el origen.

---



## Visualización con matplotlib

Vamos a graficar:

* La recta $H$ en azul,
* Su complemento ortogonal $H^\perp$ en rojo,
* Un vector $v$ cualquiera descompuesto en $h + w$, con $h \in H$ y $w \in H^\perp$.



````python
import numpy as np
import matplotlib.pyplot as plt

# =========================
# Definir los vectores generadores
# =========================
h = np.array([2, 1])   # Generador de H
w = np.array([-1, 2])  # Generador de H^perp

# Vector v cualquiera
v = np.array([3, 2])

# =========================
# Proyección de v sobre H
# =========================
proj_v_H = (np.dot(v, h) / np.dot(h, h)) * h
# Componente en H^perp
comp_v_Hperp = v - proj_v_H

# =========================
# Crear figura
# =========================
plt.figure(figsize=(7,7))
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)

# Limites de la gráfica
lim = 5
plt.xlim(-lim, lim)
plt.ylim(-lim, lim)

# =========================
# Graficar H y H^perp
# =========================
t = np.linspace(-lim, lim, 100)
plt.plot(t, (h[1]/h[0])*t, 'b', label='H: span{h}')          # Recta H
plt.plot(t, (w[1]/w[0])*t, 'r', label="H^perp: span{w}")     # Recta H^perp

# =========================
# Graficar vectores v, h, w
# =========================
plt.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='green', linewidth=3, label='v')
plt.quiver(0, 0, proj_v_H[0], proj_v_H[1], angles='xy', scale_units='xy', scale=1, color='blue', linewidth=3, label='proy_H(v)')
plt.quiver(proj_v_H[0], proj_v_H[1], comp_v_Hperp[0], comp_v_Hperp[1], angles='xy', scale_units='xy', scale=1, color='red', linewidth=3, label='componente en H^perp')

# =========================
# Segmento de la descomposición
# =========================
plt.plot([0, proj_v_H[0]], [0, proj_v_H[1]], 'b--', linewidth=2)
plt.plot([proj_v_H[0], v[0]], [proj_v_H[1], v[1]], 'r--', linewidth=2)

# =========================
# Ajustes finales
# =========================
plt.gca().set_aspect('equal', adjustable='box')
plt.grid(True)
plt.title("Complemento ortogonal H^perp de un subespacio H en R²")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.savefig(r"C:\Users\marco\Downloads\complemento_ortogonal.png")
plt.show()
````

# Ejemplo, en el espacio, del concepto de complemento ortogoonal  



## Ejemplo en $\mathbb{R}^{3}$

Sea el subespacio $H \subset \mathbb{R}^{3}$ definido como el **plano que pasa por el origen** generado por los vectores:

$$
h_1 = \begin{pmatrix}1 \\  1 \\  0\end{pmatrix}, \quad
h_2 = \begin{pmatrix}0 \\  1 \\  1\end{pmatrix}.
$$



Entonces:

$$
H = \text{gen} \{h_1, h_2 \}.
$$



El **complemento ortogonal** $H^\perp$ es el conjunto de todos los vectores ortogonales a cada vector de $H$:

$$
H^\perp = \{ v \in \mathbb{R}^3 : v \cdot h_1 = 0 \text{ y } v \cdot h_2 = 0 \}.
$$



Si resolvemos el sistema:

$$
v = \begin{pmatrix}x \\  y \\  z\end{pmatrix}, \quad
v \cdot h_1 = x + y = 0, \quad v \cdot h_2 = y + z = 0,
$$



obtenemos:

$$
y = -x, \quad z = -y = x \implies v = x \begin{pmatrix}1 \ -1 \ 1\end{pmatrix}.
$$

Por lo tanto, $H^\perp = \text{gen}{(1,-1,1)}$.

## Teorema de aproximación de la norma (Proyección ortogonal)

Sea $H$ un subespacio de $\mathbb{R}^{n}$ y $v \in \mathbb{R}^{n}$.

Sea $\operatorname{proj}_H(v)$ la **proyección ortogonal** de $v$ sobre $H$. Entonces:

$$
|\ v - \operatorname{proj}_H(v)\ | = \min_{h \in H} |v - h|,
$$

es decir, **$\operatorname{proj}_H(v)$ es el vector de $H$ más cercano a $v$** en términos de la norma euclidiana.



Equivalente geométricamente:

$$
v - \operatorname{proj}_H(v) \in H^\perp,
$$

y para cualquier $h \in H$ diferente de $\operatorname{proj}_H(v)$, se cumple:

$$
|v - h| > |v - \operatorname{proj}_H(v)|.
$$

---



## Ejemplo en $\mathbb{R}^{2}$

Sea:

$$
H = \text{la recta generada por } h = \begin{pmatrix}2 \ 1\end{pmatrix}, \quad
v = \begin{pmatrix}3 \ 2\end{pmatrix}.
$$



La proyección ortogonal de $v$ sobre $H$ se calcula como:

$$
\operatorname{proj}_H(v) = \frac{v \cdot h}{h \cdot h}  h
= \frac{3\cdot 2 + 2 \cdot 1}{2^2 + 1^2} \begin{pmatrix}2 \ 1\end{pmatrix}
= \frac{8}{5} \begin{pmatrix}2 \ 1\end{pmatrix}
= \begin{pmatrix}\frac{16}{5} \ \frac{8}{5}\end{pmatrix}.
$$



Geométricamente, $\operatorname{proj}_H(v)$ es el punto de la recta $H$ **más cercano** a $v$.

1. La **línea azul** es el subespacio $H$.
2. El **punto verde** es $v$.
3. El **punto rojo** es $\operatorname{proj}_H(v)$, el **punto de $H$ más cercano a $v$**.
4. La línea punteada negra conecta $v$ con su proyección, mostrando la **distancia mínima** $|v - \operatorname{proj}_H(v)|$.

---



## Desigualdad de Cauchy-Schwarz en $\mathbb{R}^{n}$

Sea $u, v \in \mathbb{R}^{n}$. Entonces se cumple:

$$
|u \cdot v| \leq |u| |v|,
$$

donde $u \cdot v$ es el producto escalar usual en $\mathbb{R}^{n}$ y $|u|$, $|v|$ son las normas euclidianas de $u$ y $v$ respectivamente.

---



### Caso de igualdad

Se tiene igualdad si y solo si los vectores $u$ y $v$ son **linealmente dependientes**, es decir, si existe un escalar $\lambda \in \mathbb{R}$ tal que

$$
u = \lambda v \quad \text{o} \quad v = \lambda u.
$$

---



Esta desigualdad es fundamental en geometría y álgebra lineal, y es la base para demostrar otras desigualdades como la **desigualdad triangular** y conceptos de **ángulo entre vectores** en $\mathbb{R}^{n}$.

---


# Representación geométrica de la Desigualdad de Cauchy Schwartz  

Aquí tienes una **representación geométrica en el plano $\mathbb{R}^2$** de la desigualdad de Cauchy-Schwarz. La idea visual es:

* Dos vectores $u$ y $v$ salen del origen.
* El producto escalar $u\cdot v = |u||v|\cos\theta$.
* La desigualdad
  $$
  |u\cdot v|\le |u||v|
  $$
  dice que el valor absoluto de la proyección de uno sobre el otro **nunca supera el producto de sus longitudes**.
* Geométricamente: la proyección de $u$ sobre la dirección de $v$ siempre es menor o igual que $|u|$.

### Script para visualizarlo en el plano (matplotlib)



````python
import numpy as np
import matplotlib.pyplot as plt

# Vectores en R^2
u = np.array([3, 2])
v = np.array([4, 1])

# Proyección de u sobre v
proj_u_on_v = (np.dot(u, v) / np.dot(v, v)) * v

# Ángulo entre vectores
theta = np.arccos(np.dot(u, v) / (np.linalg.norm(u)*np.linalg.norm(v)))

# Figura
plt.figure(figsize=(6,6))
plt.axhline(0)
plt.axvline(0)

# Vector u
plt.arrow(0,0,u[0],u[1], head_width=0.2, length_includes_head=True)
plt.text(u[0]+0.1,u[1],'u')

# Vector v
plt.arrow(0,0,v[0],v[1], head_width=0.2, length_includes_head=True)
plt.text(v[0]+0.1,v[1],'v')

# Proyección de u sobre v
plt.arrow(0,0,proj_u_on_v[0],proj_u_on_v[1],
          linestyle='dashed',
          head_width=0.2,
          length_includes_head=True)
plt.text(proj_u_on_v[0]+0.1,proj_u_on_v[1],'proj_v(u)')

# Segmento perpendicular
plt.plot([u[0], proj_u_on_v[0]],
         [u[1], proj_u_on_v[1]],
         linestyle='dashed')

plt.title("Representación geométrica de la desigualdad de Cauchy-Schwarz en R²")
plt.gca().set_aspect('equal', adjustable='box')
plt.grid()
plt.show()

# Verificación numérica
print("u·v =", np.dot(u,v))
print("|u||v| =", np.linalg.norm(u)*np.linalg.norm(v))
````


### Lectura geométrica del gráfico

* El vector punteado es la proyección de $u$ sobre la dirección de $v$.
* Su longitud es
  $$
  |u|\cos\theta.
  $$
* Al multiplicar por $|v|$ se obtiene $|u||v|\cos\theta = u\cdot v$.
* Como $ |\cos\theta|\le 1 $, necesariamente:
  $$
  |u\cdot v|\le |u||v|.
  $$



### Intuición clave para clase

* Si los vectores están alineados → igualdad.
* Si forman ángulo → el producto escalar disminuye.
* Si son ortogonales → producto escalar cero.

Es decir, Cauchy-Schwarz es una **restricción geométrica de cuánto puede “medir” la proyección de un vector sobre otro**.


# Aclaración sobre el razonamiento anterior  

Buena observación: el paso clave es entender **de dónde sale exactamente**
$$
|u||v|\cos\theta = u\cdot v.
$$

No es que “aparezca” por multiplicar sin más; lo que ocurre es que estamos usando **la fórmula geométrica del producto escalar**.

Vamos despacio.

---



## 1) Qué representa la proyección

La longitud de la proyección de (u) sobre la dirección de (v) es

$$
|u|\cos\theta,
$$

donde (\theta) es el ángulo entre (u) y (v).

Eso es pura geometría: en el triángulo formado por (u) y su proyección,

$$
\cos\theta=\frac{\text{cateto adyacente}}{\text{hipotenusa}}
=\frac{\text{longitud de la proyección}}{|u|}.
$$

Por eso la longitud proyectada es $|u|\cos\theta$.

---



## 2) Cómo aparece $|v|$

El producto escalar no mide solo la proyección “en sí”, sino la proyección **ponderada por la longitud del vector sobre el que proyectas**.



Geométricamente:

$$
u\cdot v
= (\text{longitud de la proyección de }u\text{ sobre la dirección de }v)\times |v|.
$$



Sustituimos la longitud de la proyección:

$$
u\cdot v = (|u|\cos\theta),|v|.
$$



Y reordenando:

$$
u\cdot v = |u||v|\cos\theta.
$$

Ese es el paso que te generó la duda: **no es una multiplicación arbitraria**, es la definición geométrica del producto escalar.

---



## 3) ¿Por qué esta expresión coincide con la definición algebraica?

Recordemos:

$$
u\cdot v = u_1v_1 + u_2v_2 + \cdots + u_nv_n.
$$



Se puede demostrar (con trigonometría o con la ley del coseno) que esta suma es exactamente igual a

$$
|u||v|\cos\theta.
$$

De hecho, esta identidad es una de las formas más importantes del producto escalar: conecta

* la **definición algebraica** (sumas de coordenadas),
* con la **interpretación geométrica** (ángulo entre vectores).

---



## 4) Ahora sí: el paso de la desigualdad

Sabemos que:

$$
u\cdot v = |u||v|\cos\theta.
$$



Tomamos valor absoluto:

$$
|u\cdot v|
= |u||v|\cos\theta|
= |u||v||\cos\theta|.
$$



Y como siempre

$$
|\cos\theta|\le 1,
$$



entonces

$$
|u\cdot v| \le |u||v|.
$$

Eso es exactamente la desigualdad de Cauchy-Schwarz.

---



## 5) Intuición corta

* $ |u|\cos\theta $ = cuánto “apunta” $u$ en la dirección de $v$.
* Multiplicar por $ |v| $ convierte esa proyección en el producto escalar.
* Como $ |\cos\theta|\le 1 $, el producto escalar nunca puede superar el producto de longitudes.

---



## 6) Idea clave para recordarlo

Piensa así:

$$
u\cdot v = (\text{proyección de }u\text{ sobre }v)\times |v|.
$$

Y la proyección nunca puede ser más larga que $ |u| $.



Por eso inevitablemente:

$$
|u\cdot v|\le |u||v|.
$$

La desigualdad es, en el fondo, una afirmación geométrica muy simple:
**ninguna proyección puede ser más larga que el vector original.**

El interés por integrar la tecnología en la enseñanza de las matemáticas tiene raíces en las propuestas de **Papert (1980, 1996)**, quien con su enfoque constructionista concibió la computadora como un “objeto para pensar”, capaz de mediar la construcción activa de significados matemáticos mediante la programación. Esta perspectiva se enlaza con los planteamientos de **Piaget (1970)** y **Vygotsky (1978)**, quienes reconocieron la importancia de la experimentación y de las herramientas culturales en la formación del pensamiento. Bajo esta mirada, el uso de entornos digitales no constituye una amenaza para la enseñanza tradicional de las matemáticas, sino un recurso que amplía las posibilidades de aprendizaje.



A comienzos del siglo XXI, la noción de **pensamiento computacional** introducida por **Wing (2006)** renovó el debate al proponer que esta competencia debía enseñarse de manera transversal desde los niveles iniciales hasta la educación superior, situándola al mismo nivel que la lectura, la escritura y las matemáticas. Este enfoque se alinea con los planteamientos del **NCTM (2000, 2014)**, que destacó la tecnología como un componente esencial para la experimentación, la representación y la comunicación en matemáticas. 

En paralelo, aportes teóricos como la **teoría de registros de representación semiótica de Duval (1993, 2006)** y el **Enfoque Ontosemiótico de Godino (2007, 2017)** muestran que el aprendizaje matemático se enriquece al transitar entre múltiples sistemas de representación, incluyendo los digitales.



En el contexto actual, organizaciones como la **ACM** y el **IEEE** insisten en que las competencias computacionales son indispensables en la formación profesional, lo que refuerza la pertinencia de integrar lenguajes como **Python, LaTeX, Markdown e inglés** en ambientes como **Jupyter Notebooks** para la enseñanza universitaria de las matemáticas. Así, el estado del arte evidencia un consenso creciente: lejos de ser un obstáculo, el pensamiento computacional y el uso de lenguajes digitales constituyen una vía para fortalecer la comprensión matemática, formar profesionales más competentes y responder a las demandas del mundo laboral y científico contemporáneo.

# Estado del arte con citas

El uso de la tecnología en la enseñanza de las matemáticas encuentra uno de sus pilares en **Papert (1980)**, quien afirmaba que *“los niños no aprenden sobre la computadora, aprenden con la computadora”* (p. 6). En su visión, la programación se convierte en un medio para la construcción activa de significados matemáticos, idea que se enlaza con la epistemología genética de **Piaget (1970)**, quien sostuvo que *“el conocimiento se construye a través de la acción”* (p. 15), y con la perspectiva de **Vygotsky (1978)**, al señalar que *“toda función en el desarrollo cultural del niño aparece dos veces: primero en el plano social y después en el individual”* (p. 57). Estos enfoques coinciden en otorgar a las herramientas culturales un papel esencial en la mediación del aprendizaje.



Posteriormente, **Wing (2006)** acuñó el término pensamiento computacional, definiéndolo como *“una habilidad fundamental para todos, no solo para los científicos de la computación”* (p. 33). Esta autora subrayó que debía enseñarse de manera transversal, en paralelo con la lectura, la escritura y la matemática. En la misma línea, el **NCTM (2000)** destacó que *“la tecnología es esencial en la enseñanza y el aprendizaje de las matemáticas; influye en la matemática que se enseña y mejora el aprendizaje de los estudiantes”* (p. 24).



Desde la didáctica de las matemáticas, **Duval (2006)** mostró que *“la comprensión matemática requiere la coordinación de al menos dos registros de representación diferentes”* (p. 107), lo que legitima la incorporación de registros digitales como LaTeX, Markdown y Python. Por su parte, **Godino (2017)** planteó que *“el significado de un objeto matemático está constituido por el sistema de prácticas en el que interviene”* (p. 7), con lo cual la práctica digital se convierte en parte constitutiva del aprendizaje matemático.



Finalmente, organismos internacionales como la **ACM y el IEEE (2013)** enfatizan que *“las habilidades computacionales son esenciales para el ejercicio profesional en cualquier disciplina”* (p. 12). De esta manera, se consolida un consenso: la enseñanza universitaria de las matemáticas debe integrar herramientas contemporáneas como **Jupyter Notebooks, Python, LaTeX, Markdown e inglés**, no para competir con la enseñanza tradicional, sino para **fortalecer la comprensión matemática, potenciar la comunicación científica y responder a las demandas del mundo laboral**.

# 3. Recordatorio: Base de un espacio vectorial

Una base de $\mathbb{R}^{n}$ es un conjunto

$$
B = {v_1, v_2, \dots, v_n}
$$

tal que

1. los vectores son linealmente independientes
2. generan todo el espacio

Todo vector $x$ puede escribirse como

$$
x = c_1 v_1 + c_2 v_2 + \dots + c_n v_n
$$

---

# 4. Ortogonalidad

Dos vectores son ortogonales si

$$
v \cdot w = 0
$$

Ejemplo en $\mathbb{R}^{2}$

$$
v = (1,0), \quad w = (0,1)
$$

---

# 5. Base Ortogonal

Una base es ortogonal si

$$
v_i \cdot v_j = 0 \quad i \neq j
$$

Ventaja fundamental:

Los coeficientes se calculan fácilmente.

---

# 6. Base Ortonormal

Una base es ortonormal si

$$
v_i \cdot v_j = 0
$$

y además

$$
|v_i| = 1
$$

Ejemplo clásico en $\mathbb{R}^{3}$

$$
e_1 =
\begin{bmatrix}
1 \
0 \
0
\end{bmatrix}
\quad
e_2 =
\begin{bmatrix}
0 \
1 \
0
\end{bmatrix}
\quad
e_3 =
\begin{bmatrix}
0 \
0 \
1
\end{bmatrix}
$$

---

# 7. Propiedad Fundamental

Si ${u_1,\dots,u_n}$ es base ortonormal entonces

$$
x =
(x\cdot u_1)u_1 +
(x\cdot u_2)u_2 +
\dots +
(x\cdot u_n)u_n
$$

Esto es **clave en Machine Learning**.

Los coeficientes son simplemente productos punto.

---

# 8. Interpretación en Ciencia de Datos

Un vector de datos

$$
x =
\begin{bmatrix}
lluvia \
temperatura \
humedad
\end{bmatrix}
$$

puede representarse en una nueva base

$$
u_1, u_2, u_3
$$

donde

* $u_1$ captura la mayor variación
* $u_2$ la segunda
* $u_3$ ruido

Esto es la idea de **PCA**.

---

# 9. Proyección de un vector sobre otro

Sea $u$ un vector.

La proyección de $x$ sobre $u$ es

$$
proj_u(x) =
\frac{x\cdot u}{u\cdot u}u
$$

Si $u$ es unitario

$$
proj_u(x) = (x\cdot u)u
$$

---

# 10. Interpretación geométrica

La proyección representa

**la mejor aproximación de $x$ en la dirección de $u$.**

En Machine Learning esto significa:

**reducir dimensionalidad**.

---

# 11. Ejemplo aplicado

Supongamos datos de cultivo

$$
x =
\begin{bmatrix}
120 \
30
\end{bmatrix}
$$

donde

* lluvia
* temperatura

Dirección de crecimiento del cultivo

$$
u =
\begin{bmatrix}
1 \
1
\end{bmatrix}
$$

La proyección mide

**cuánto del vector está alineado con esa tendencia.**

---

# 12. Implementación en Python

```python
import numpy as np

x = np.array([120,30])
u = np.array([1,1])

proj = (x @ u)/(u @ u) * u

proj
```

---

# 14. Proyección sobre un subespacio

Sea una base ortonormal

$$
{u_1,u_2,\dots,u_k}
$$

La proyección es

$$
proj(x) =
(x\cdot u_1)u_1 +
(x\cdot u_2)u_2 +
\dots +
(x\cdot u_k)u_k
$$

Esto es exactamente lo que hace **PCA**.

# 16. Conexión directa con PCA

PCA busca vectores

$$
u_1, u_2, \dots
$$

tales que

* sean ortonormales
* maximicen la varianza de los datos

Luego proyecta los datos:

$$
z = U^T x
$$

Esto reduce dimensión.

---

# 17. Experimento computacional

Uso de PCA en Python.

```python
from sklearn.decomposition import PCA

X = datos.values

pca = PCA()

pca.fit(X)

pca.components_
```

Cada fila es un **vector ortonormal**.

---

# 18. Visualización de reducción de dimensión

```python
pca = PCA(n_components=2)

Z = pca.fit_transform(X)

Z[:5]
```

Esto es exactamente

**proyección en un subespacio de dimensión 2.**
