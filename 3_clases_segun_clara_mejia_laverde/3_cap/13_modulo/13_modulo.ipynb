# Introducción  



En el ejemplo 3 del módulo 11 vimos que una matriz $A_{m \times n}$ induce una transformación lineal  de $\mathbb{R}^{n}$ en $\mathbb{R}^{m}$ dada por $Tx = Ax$.  Cuando $T$ tiene esta forma, el núcleo de $T$ es el núcleo de $A$, y la imagen de $T$,  formada por los vectores $y = Ax$ para algún $x \in \mathbb{R}^{n}$, será el espacio columna de $A$, ya que $y$ es una combinación lineal de las columnas de $A$. Estas coincidencias  entre el núcleo y la imagen de $T$ con el núcleo y el espacio columna de $A$ permiten estudiar la transformación a través de la matriz $A$.
En esta sección veremos que cualquier transformación lineal definida entre espacios vectoriales de dimensión finita se puede expresar en la forma $T(x) = Ax$ donde $A$ es la matriz que representa la transformación.

# 13.1 Tercer problema básico del álgebra lineal  



Dada una transformación $T: V \to W$ con $dim\ V = n$  y  $dim W = m$,  hallar una matriz $A$  de $m \times n$ que represente a $T$.


Resolveremos primero el problema para $T:\mathbb{R}^{n} \to \mathbb{R}^{m}$ y luego lo extenderemos a espacios $V$ y $W$ de dimensiones $n$ y $m$, respectivamente.

## Teorema 1  

### Teorema (Existencia de la matriz de una transformación lineal)

Sea
$$
T:\mathbb{R}^{n} \to \mathbb{R}^{m}
$$
una **transformación lineal**.

Entonces **existe una única matriz**
$$
A \in \mathbb{R}_{m \times n}
$$
tal que para todo vector
$$
x \in \mathbb{R}^{n}
$$
se cumple

$$
T(x) = Ax.
$$

Es decir, **toda transformación lineal entre espacios euclidianos finitos puede representarse como una multiplicación matricial**.

---



# Idea intuitiva del teorema

El punto clave es que:

* Todo vector de $\mathbb{R}^n$ puede escribirse como combinación lineal de la **base canónica**.

Si

$$
x =
\begin{bmatrix}
x_1 \\ 
x_2 \\ 
\vdots \\ 
x_n
\end{bmatrix}
=
x_1 e_1 + x_2 e_2 + \dots + x_n e_n
$$

donde

$$
e_1, e_2, \dots, e_n
$$

son los vectores de la base canónica.

Como $T$ es lineal,

$$
T(x)
=
x_1T(e_1)+x_2T(e_2)+\dots+x_nT(e_n)
$$

Por lo tanto:

**la transformación queda completamente determinada por lo que hace con los vectores de la base.**

---



# Cómo se construye la matriz $A$

La matriz se obtiene colocando como columnas las imágenes de la base canónica.

$$
A =
\begin{bmatrix}
| & | &  & | \\ 
T(e_1) & T(e_2) & \dots & T(e_n) \\ 
| & | &  & |
\end{bmatrix}
$$



# Procedimiento didáctico paso a paso

### Paso 1 — Escribir la base canónica

En $\mathbb{R}^n$

$$
e_1 =
\begin{bmatrix}
1\\ 0\\ \vdots\\ 0
\end{bmatrix},
\quad
e_2 =
\begin{bmatrix}
0\\ 1\\ \vdots\\ 0
\end{bmatrix},
\dots,
e_n
$$

---



### Paso 2 — Aplicar la transformación

Calcular

$$
T(e_1), T(e_2), \dots, T(e_n)
$$

Cada uno será un vector de $\mathbb{R}^m$.

---



### Paso 3 — Formar la matriz

Colocar esos vectores como columnas.

$$
A =
\begin{bmatrix}
T(e_1) & T(e_2) & \dots & T(e_n)
\end{bmatrix}
$$

---



# Ejemplo sencillo

Sea

$$
T:\mathbb{R}^2 \to \mathbb{R}^2
$$

definida por

$$
T\begin{pmatrix}x \\ y\end{pmatrix} = \begin{pmatrix}2x+y \\ x-y\end{pmatrix}
$$

---



### 1. Base canónica

$$
e_1=(1,0),\qquad e_2=(0,1)
$$

---



### 2. Aplicar $T$

$$
T(e_1)=T(1,0)=(2,1)
$$

$$
T(e_2)=T(0,1)=(1,-1)
$$

---



### 3. Construir la matriz

$$
A=
\begin{bmatrix}
2 & 1 \\ 
1 & -1
\end{bmatrix}
$$

---



### 4. Verificación

Si

$$
\mathbf{x}=
\begin{bmatrix}
x \\ 
y
\end{bmatrix}
$$

entonces

$$
Ax=
\begin{bmatrix}
2 & 1 \\ 
1 & -1
\end{bmatrix}
\begin{bmatrix}
x \\ 
y
\end{bmatrix}
=
\begin{bmatrix}
2x+y \\ 
x-y
\end{bmatrix}
=
T(x,y)
$$

---



# Interpretación geométrica

La matriz contiene **la acción de la transformación sobre los ejes del espacio**.

* Primera columna → a dónde va el eje $x$
* Segunda columna → a dónde va el eje $y$

Por eso la matriz **codifica completamente la transformación**.  



La transformación lineal está dada por la matriz

$$
A =
\begin{bmatrix}
2 & 1 \
1 & -1
\end{bmatrix}
$$

Esto significa que transforma cualquier vector del plano según

$$
T(x,y) = (2x + y,; x - y)
$$

Una buena forma de entenderla es observar qué ocurre con los **vectores base canónicos** del plano.

---

# Visualización con Python (Matplotlib)

```python


````python
import numpy as np
import matplotlib.pyplot as plt

# Matriz de transformación
A = np.array([[2, 1],
              [1, -1]])

# Vectores base
e1 = np.array([1, 0])
e2 = np.array([0, 1])

# Transformación
Te1 = A @ e1
Te2 = A @ e2

plt.figure(figsize=(6,6))

# Ejes originales
plt.quiver(0,0,e1[0],e1[1],angles='xy',scale_units='xy',scale=1, color='blue')
plt.quiver(0,0,e2[0],e2[1],angles='xy',scale_units='xy',scale=1, color='red')

# Ejes transformados
plt.quiver(0,0,Te1[0],Te1[1],angles='xy',scale_units='xy',scale=1, color='blue', alpha=0.5)
plt.quiver(0,0,Te2[0],Te2[1],angles='xy',scale_units='xy',scale=1, color='red', alpha=0.5)

# Límites
plt.xlim(-3,3)
plt.ylim(-3,3)

# Cuadrícula
plt.axhline(0)
plt.axvline(0)
plt.grid()

plt.gca().set_aspect('equal', adjustable='box')

plt.legend([
    "eje x",
    "eje y",
    "A(e1)",
    "A(e2)"
])

plt.title("Transformación lineal de los ejes")
plt.savefig(r"C:\Users\marco\Downloads\transformacion_lineal.png")
plt.show()
````

# Qué le ocurre a los ejes

Los ejes del plano son los vectores base:

$$
e_1 = (1,0), \qquad e_2 = (0,1)
$$



Aplicamos la transformación.

### Imagen del eje $x$

$$
T(1,0) =
\begin{bmatrix}
2 \
1
\end{bmatrix}
$$

Entonces el eje $x$ se transforma en un vector que apunta hacia

* derecha
* arriba

con dirección

$$
(2,1)
$$

Geométricamente:

* el eje horizontal **se inclina hacia arriba**.

---



### Imagen del eje $y$

$$
T(0,1) =
\begin{bmatrix}
1 \
-1
\end{bmatrix}
$$

Entonces el eje $y$ se transforma en

* derecha
* abajo

con dirección

$$
(1,-1)
$$

Geométricamente:

* el eje vertical **se inclina hacia abajo**.

---



# Interpretación geométrica

La transformación produce tres efectos simultáneos:

### 1. Cizallamiento (shear)

Los ejes dejan de ser perpendiculares.

Antes:

* $90^\circ$

Después:

* forman un paralelogramo.

---



### 2. Rotación parcial de direcciones

* el eje $x$ gira hacia el primer cuadrante
* el eje $y$ gira hacia el cuarto cuadrante

---



### 3. Escalamiento diferente

Longitudes:

$$
||A e_1|| = \sqrt{5}
$$

$$
||A e_2|| = \sqrt{2}
$$

La transformación **estira más en la dirección del eje (x)** que en la del eje $y$.

---



# Interpretación conceptual para clase

Una forma poderosa de explicarlo a estudiantes:

> Una matriz no transforma puntos aislados.
> **Transforma todo el sistema de coordenadas.**

Los nuevos ejes son

$$
A e_1 ,; A e_2
$$

y cualquier vector se reconstruye como combinación lineal de ellos.

---


# Veamos lo que le hace a un cuadrado, esta transformación lineal  

Para ver cómo la matriz transforma el **cuadrado unitario**, basta transformar sus cuatro vértices con la matriz

$$
A=
\begin{bmatrix}
2 & 1 \
1 & -1
\end{bmatrix}
$$

El cuadrado unitario tiene vértices

$$
(0,0),;(1,0),;(1,1),;(0,1)
$$

y su imagen será un **paralelogramo**.

---

# Código en Python con Matplotlib

```python


````python
import numpy as np
import matplotlib.pyplot as plt

# Matriz de transformación
A = np.array([[2,1],
              [1,-1]])

# Vértices del cuadrado unitario
square = np.array([
    [0,0],
    [1,0],
    [1,1],
    [0,1],
    [0,0]
])

# Transformar vértices
transformed = square @ A.T

plt.figure(figsize=(6,6))

# Cuadrado original
plt.plot(square[:,0], square[:,1], marker='o', label='Cuadrado unitario')

# Cuadrado transformado
plt.plot(transformed[:,0], transformed[:,1], marker='o', label='Imagen por A')

# Ejes
plt.axhline(0)
plt.axvline(0)
plt.grid()

plt.gca().set_aspect('equal', adjustable='box')

plt.legend()
plt.title("Transformación del cuadrado unitario")
plt.show()
````

# Interpretación visual para estudiantes

**Las matrices transforman cuadrados en paralelogramos.**

porque

$$

$$

Es decir, toda la geometría del plano se reconstruye a partir de las imágenes de los ejes.

# 2. Transformación de toda la malla del plano

Esto es **muy poderoso visualmente**.


````python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([[2,1],
              [1,-1]])

grid_range = np.arange(-5,6)

fig, ax = plt.subplots(1,2, figsize=(12,6))

# -------------------------
# MALLA ORIGINAL
# -------------------------

for x in grid_range:
    ax[0].plot([x]*len(grid_range), grid_range)

for y in grid_range:
    ax[0].plot(grid_range, [y]*len(grid_range))

ax[0].axhline(0)
ax[0].axvline(0)
ax[0].set_aspect('equal')
ax[0].grid()
ax[0].set_title("Malla original")

# -------------------------
# MALLA TRANSFORMADA
# -------------------------

for x in grid_range:
    pts = np.array([[x,y] for y in grid_range])
    t = pts @ A.T
    ax[1].plot(t[:,0], t[:,1])

for y in grid_range:
    pts = np.array([[x,y] for x in grid_range])
    t = pts @ A.T
    ax[1].plot(t[:,0], t[:,1])

ax[1].axhline(0)
ax[1].axvline(0)
ax[1].set_aspect('equal')
ax[1].grid()
ax[1].set_title("Malla transformada por A")

plt.suptitle("Efecto geométrico de la transformación lineal")

plt.show()
plt.show()
````

Interpretación:

* las líneas horizontales y verticales se vuelven **familias inclinadas**
* se ve claramente el **cizallamiento**.

# 3. Vectores base + cuadrado + transformación

Esta es la **mejor para explicar en clase**.


````python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([[2,1],
              [1,-1]])

e1 = np.array([1,0])
e2 = np.array([0,1])

Te1 = A @ e1
Te2 = A @ e2

square = np.array([
    [0,0],
    [1,0],
    [1,1],
    [0,1],
    [0,0]
])

transformed = square @ A.T

plt.figure(figsize=(7,7))

# cuadrado original
plt.plot(square[:,0], square[:,1])

# cuadrado transformado
plt.plot(transformed[:,0], transformed[:,1])

# vectores base
plt.quiver(0,0,e1[0],e1[1],angles='xy',scale_units='xy',scale=1)
plt.quiver(0,0,e2[0],e2[1],angles='xy',scale_units='xy',scale=1)

# vectores transformados
plt.quiver(0,0,Te1[0],Te1[1],angles='xy',scale_units='xy',scale=1)
plt.quiver(0,0,Te2[0],Te2[1],angles='xy',scale_units='xy',scale=1)

plt.axhline(0)
plt.axvline(0)
plt.grid()

plt.gca().set_aspect('equal')
plt.title("Cómo una matriz transforma el plano")

plt.show()
````

# Explicación conceptual para estudiantes

Una matriz $2\times2$ hace exactamente esto:

1. mueve los **vectores base**
2. el cuadrado unitario se convierte en un **paralelogramo**
3. toda la malla del plano se deforma con la misma regla

porque

$$
A(x,y)=xAe_1+yAe_2
$$

Es decir:

> La matriz redefine el sistema de coordenadas.

---



# Observación bonita para clase

El área cambia por

$$
|\det(A)|=3
$$

Entonces:

* el cuadrado de área $1$
* pasa a un paralelogramo de área $3$
* además se **invierte la orientación**.

Esto se puede mostrar fácilmente en la gráfica.

---



Una forma muy clara de mostrar la deformación continua es **interpolar entre la matriz identidad y la matriz $A$**.
Así los estudiantes ven cómo el cuadrado “viaja” lentamente hasta convertirse en el paralelogramo.

La familia de transformaciones es

$$
T_t = (1-t)I + tA, \qquad t\in[0,1]
$$

* $t=0$ → no hay transformación
* $t=1$ → transformación completa.

---



# Script con ipywidgets


````python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# matriz de transformación
A = np.array([[2,1],
              [1,-1]])

# identidad
I = np.eye(2)

# cuadrado unitario
square = np.array([
    [0,0],
    [1,0],
    [1,1],
    [0,1],
    [0,0]
])

def deformacion(t=0):

    T = (1-t)*I + t*A
    transformed = square @ T.T

    plt.figure(figsize=(6,6))

    # cuadrado original
    plt.plot(square[:,0], square[:,1])

    # cuadrado deformado
    plt.plot(transformed[:,0], transformed[:,1])

    # ejes
    plt.axhline(0)
    plt.axvline(0)

    plt.xlim(-2,4)
    plt.ylim(-3,3)

    plt.gca().set_aspect('equal')
    plt.grid()

    plt.title(f"Interpolación lineal de la transformación   t = {t:.2f}")
    plt.show()

interact(
    deformacion,
    t=FloatSlider(min=0,max=1,step=0.02,value=0)
)
````

# Qué verán los estudiantes

Cuando mueves el slider:

1. El cuadrado comienza rígido.
2. Los lados empiezan a **inclinarse**.
3. El área aumenta.
4. Finalmente aparece el **paralelogramo final**.

Esto hace visible que

$$
T(x,y)=(2x+y,;x-y)
$$

está deformando **todo el espacio**, no solo una figura.

---



# Interpretación matemática para explicar mientras corre

# Teorema (Matriz de una transformación lineal respecto a bases)

Sean

* $V$ y $W$ espacios vectoriales de dimensión finita.
* $T:V \to W$ una transformación lineal.
* $B_1 = \{ v_1, v_2, \dots, v_n \}$ una base de $V$.
* $B_2 = \{ w_1, w_2, \dots, w_m \}$ una base de $W$.

Entonces **existe una única matriz**

$$
A_T \in \mathbb{R}^{m\times n}
$$

tal que para todo $x \in V$

$$
[T(x)]_{B_2} = A_T [x]_{B_1}.
$$

---



# Interpretación

La matriz $A_T$ **traduce las coordenadas de un vector en la base $B_1$** a **las coordenadas de su imagen en la base $B_2$**.

Es decir,

$$
\text{coordenadas en } V
\quad \xrightarrow{A_T} \quad
\text{coordenadas en } W.
$$

---



# Cómo se construye la matriz

La matriz se obtiene calculando las imágenes de los vectores de la base del dominio.

1. Tomar cada vector de la base $B_1$

$$
v_1, v_2, \dots, v_n
$$



2. Aplicar la transformación

$$
T(v_1), T(v_2), \dots, T(v_n)
$$



3. Expresar cada resultado en la base $B_2$

$$
T(v_j) =
a_{1j} w_1 +
a_{2j} w_2 +
\dots +
a_{mj} w_m
$$



4. Esos coeficientes forman la columna $j$.

---



# Forma de la matriz

$$
A_T =
\begin{bmatrix}
| & | &  & | \\ 
[T(v_1)]_{B_2} &
[T(v_2)]_{B_2} &
\dots &
[T(v_n)]_{B_2} \\ 
| & | &  & |
\end{bmatrix}
$$

---



# Significado profundo

La matriz contiene **la acción de la transformación sobre la base del dominio**, pero **expresada en la base del codominio**.

---



# Caso particular importante

Si

$$
B_1 = B_2 = \text{base canónica}
$$

entonces

$$
T(x)=Ax
$$

que es el teorema que mostraste antes en $\mathbb{R}^n$.

---



# Versión corta para estudiantes

**Teorema.**
Sea $T:V\to W$ lineal y sean $B_1$ y $B_2$ bases de $V$ y $W$ respectivamente.
Existe una única matriz $A_T$ tal que

$$
[T(x)]_{B_2}=A_T[x]_{B_1}.
$$

Las columnas de $A_T$ son las coordenadas de

$$
T(v_1),\dots,T(v_n)
$$

en la base $B_2$.

---



# Intuición geométrica

Una transformación lineal no depende del sistema de coordenadas.

Pero **la matriz sí**.

Cambiar bases es como **cambiar la cámara desde donde miramos la transformación**.


# 2. Transformación fija

La transformación geométrica será siempre la misma.


````python
A_std = np.array([
    [2,1],
    [1,-1]
])
````

Esta es la matriz de $T$ en la **base canónica**.

---



# 3. Funciones auxiliares

## Matriz de cambio de base


````python
def matriz_base(v1, v2):
    return np.column_stack([v1, v2])
````


## Cálculo de la matriz de $T$ en nuevas bases

Teorema fundamental:

$$
A_{B_2,B_1}
=
P_{B_2}^{-1}
A
P_{B_1}
$$



````python
def matriz_transformacion(A, B1, B2):
    
    P1 = matriz_base(B1[:,0], B1[:,1])
    P2 = matriz_base(B2[:,0], B2[:,1])
    
    return np.linalg.inv(P2) @ A @ P1
````


# 4. Plano de puntos



````python
x = np.linspace(-5,5,20)
y = np.linspace(-5,5,20)

X,Y = np.meshgrid(x,y)

points = np.vstack([X.ravel(), Y.ravel()])
````


# 5. Visualización interactiva


````python
def visualizar(a,b,c,d,e,f,g,h):

    B1 = np.array([[a,b],[c,d]])
    B2 = np.array([[e,f],[g,h]])
    
    if abs(np.linalg.det(B1)) < 0.1 or abs(np.linalg.det(B2)) < 0.1:
        print("La base no es válida")
        return
    
    A_new = np.linalg.inv(B2) @ A_std @ B1
    
    transformed = A_std @ points
    
    plt.figure(figsize=(8,8))
    
    plt.scatter(points[0], points[1], alpha=0.15, label="Espacio original", color = 'blue') 
    plt.scatter(transformed[0], transformed[1], alpha=0.15, label="T(x)", color = 'red')
    
    plt.quiver(0,0,B1[0,0],B1[1,0], angles='xy', scale_units='xy', scale=1)
    plt.quiver(0,0,B1[0,1],B1[1,1], angles='xy', scale_units='xy', scale=1)
    
    plt.quiver(0,0,B2[0,0],B2[1,0], angles='xy', scale_units='xy', scale=1)
    plt.quiver(0,0,B2[0,1],B2[1,1], angles='xy', scale_units='xy', scale=1)
    
    plt.xlim(-6,6)
    plt.ylim(-6,6)
    
    plt.axhline(0)
    plt.axvline(0)
    
    plt.grid()
    
    plt.title("Cambio de bases en una transformación lineal")
    
    plt.legend()
    
    plt.show()
    
    print("Matriz de T en las nuevas bases:\n")
    print(A_new)
````

# 3. Definir transformación


````python
def T(a,b,c):

    p = a + b*x + c*x**2
    
    dp = b + 2*c*x
    
    p0 = a
    
    Tp = dp + p0
    
    return p, dp, Tp
````

# Teorema (Rango–Nulidad para transformaciones lineales)

Sea

$$
T:V \rightarrow W
$$

una transformación lineal entre espacios vectoriales de dimensión finita. Entonces se cumple que

$$
\dim(V) = \operatorname{nul}(T) + \operatorname{rg}(T).
$$

donde

* $\operatorname{nul}(T) = \dim(\ker T)$ es la **nulidad** de $T$,
* $\operatorname{rg}(T) = \dim(\operatorname{Im}T)$ es el **rango** de $T$.

---



# Relación con la matriz que representa a $T$

Sea $A$ la matriz que representa la transformación lineal $T$ respecto a ciertas bases del dominio y del codominio.

Entonces se cumple que

$$
\operatorname{rg}(T) = \operatorname{rg}(A)
$$

y

$$
\operatorname{nul}(T) = \operatorname{nul}(A).
$$

Por lo tanto

$$
\dim(V) = \operatorname{rg}(A) + \operatorname{nul}(A).
$$

---



# Interpretación

El teorema dice que el espacio del dominio se descompone en dos partes:

1. **Vectores que la transformación aplasta a cero**
   (el núcleo).

2. **Direcciones que realmente sobreviven en la imagen**
   (el rango).

---



# Forma intuitiva para estudiantes

Todo vector del dominio puede verse como

$$
\text{parte que desaparece} + \text{parte que se transforma}.
$$

La cantidad de direcciones que desaparecen es la **nulidad**
y la cantidad de direcciones que sobreviven es el **rango**.

---

$$

$$

$$

\begin{bmatrix}

\end{bmatrix}.
$$

El rango es

$$
\operatorname{rg}(A)=2
$$

porque las filas son independientes.

Entonces

$$
\operatorname{nul}(A)=3-2=1.
$$

Y se cumple

$$
3 = 2 + 1.
$$

# Interpretación geométrica

> la nulidad mide cuánta información se pierde.

# 1. Teorema: matriz de una transformación con cambio de base

Sea

$$
T: V \rightarrow W
$$

una transformación lineal.



Sean:

* $B_1$ una base de $V$
* $B_2$ una base de $W$

y sean

* $P_1$ la matriz de cambio de base de $B_1$ a la base canónica de $V$
* $P_2$ la matriz de cambio de base de $B_2$ a la base canónica de $W$

Si $A$ es la matriz de $T$ respecto a las **bases canónicas**, entonces la matriz de $T$ respecto a $B_1$ y $B_2$ es

$$
A_{B_2 \leftarrow B_1}
=
P_2^{-1}
A
P_1
$$

---

La fórmula dice:

1. $P_1$ convierte coordenadas desde la base $B_1$ a la base estándar.
2. $A$ aplica la transformación en coordenadas estándar.
3. $P_2^{-1}$ convierte el resultado a la base $B_2$.

Es decir

$$
[T(x)]_{B_2}
=
P_2^{-1}
A
P_1
[x]_{B_1}
$$

# 13.2 Matrices Similares y Cambio de Base  

Al construir una representación matricial $A$ para una transformación $T$ podemos analizar $T$ trabajando con $A$. Tendremos una ventaja en este hecho si $A$ es una matriz «sencilla», esto es, que simplifique la manipulación algebraica. Como bases diferentes dan por resultado matrices diferentes, la elección «correcta» de una base para obtener una matriz $A$ sencilla es importante.

Este es el fenómeno de **diagonalización** de una transformación lineal. Veamos un ejemplo claro para clase.

---

# Ejemplo de transformación (T:\mathbb{R}^2 \to \mathbb{R}^2)

Definimos

[
T(x,y)=(3x+y,;x+3y)
]

---

# 1. Matriz en la base canónica

La base canónica es

[
E={(1,0),(0,1)}
]

Calculamos:

[
T(1,0)=(3,1)
]

[
T(0,1)=(1,3)
]

Entonces la matriz es

[
A=
\begin{bmatrix}
3 & 1 \
1 & 3
\end{bmatrix}
]

Esta matriz **no es diagonal**.

---

# 2. Buscamos una base que diagonalice la transformación

Debemos encontrar **vectores propios**.

## Polinomio característico

[
\det(A-\lambda I)=0
]

[
\begin{vmatrix}
3-\lambda & 1 \
1 & 3-\lambda
\end{vmatrix}=0
]

[
(3-\lambda)^2-1=0
]

[
\lambda^2-6\lambda+8=0
]

---

# 3. Valores propios

[
\lambda_1=4
]

[
\lambda_2=2
]

---

# 4. Vector propio para (\lambda=4)

Resolvemos

[
(A-4I)v=0
]

[
\begin{bmatrix}
-1 & 1 \
1 & -1
\end{bmatrix}
]

De aquí

[
v_1=(1,1)
]

---

# 5. Vector propio para (\lambda=2)

[
(A-2I)=
\begin{bmatrix}
1 & 1 \
1 & 1
\end{bmatrix}
]

Entonces

[
v_2=(1,-1)
]

---

# 6. Base que diagonaliza

Tomamos

[
B_1={(1,1),(1,-1)}
]

Esta base **no es la canónica**.

---

# 7. Matriz de la transformación en esta base

Sabemos que

[
T(v_1)=4v_1
]

[
T(v_2)=2v_2
]

Por lo tanto la matriz es

$$
[T]_{B_1}=
\begin{bmatrix}
4 & 0 \\ 
0 & 2
\end{bmatrix}
$$

---



# 8. Interpretación geométrica

La transformación:

* **estira** en la dirección $(1,1)$
* **comprime** en la dirección $(1,-1)$

La base canónica mezcla estas direcciones, por eso la matriz no es diagonal.

---



# 9. Matriz de cambio de base

La matriz formada por los vectores propios es

$$
P=
\begin{bmatrix}
1 & 1 \\ 
1 & -1
\end{bmatrix}
$$



Entonces

$$
P^{-1}AP=
\begin{bmatrix}
4 & 0 \\ 
0 & 2
\end{bmatrix}
$$

---



# 10. Interpretación para estudiantes

La diagonalización significa:

> Encontrar un sistema de coordenadas donde la transformación actúe solo como **escalamiento en direcciones especiales**.

Esas direcciones son los **vectores propios**.

````python
import numpy as np
import matplotlib.pyplot as plt

# Matriz de la transformación
A = np.array([[3,1],
              [1,3]])

# vectores propios
v1 = np.array([1,1])
v2 = np.array([1,-1])

# normalizar para visualización
v1 = v1 / np.linalg.norm(v1)
v2 = v2 / np.linalg.norm(v2)

# malla del plano
x = np.linspace(-5,5,20)
y = np.linspace(-5,5,20)

X, Y = np.meshgrid(x,y)

points = np.vstack([X.ravel(), Y.ravel()])

# transformación
Tpoints = A @ points

# --- Gráfica ---

plt.figure(figsize=(8,8))

# plano original
plt.scatter(points[0], points[1], alpha=0.25, label="Plano original")

# plano transformado
plt.scatter(Tpoints[0], Tpoints[1], alpha=0.25, label="Plano transformado")

# vectores propios
plt.quiver(0,0,v1[0],v1[1], angles='xy', scale_units='xy', scale=1,
           label="Dirección propia 1", color='red')

plt.quiver(0,0,v2[0],v2[1], angles='xy', scale_units='xy', scale=1,
           label="Dirección propia 2", color='blue')

# transformación de vectores propios
Tv1 = A @ v1
Tv2 = A @ v2

plt.quiver(0,0,Tv1[0],Tv1[1], angles='xy', scale_units='xy', scale=1,
           label="T(v1)", color='green')

plt.quiver(0,0,Tv2[0],Tv2[1], angles='xy', scale_units='xy', scale=1,
           label="T(v2)", color='orange')

plt.axhline(0)
plt.axvline(0)

plt.xlim(-10,10)
plt.ylim(-10,10)

plt.grid()
plt.legend()
plt.title("Deformación del plano por una transformación lineal")

plt.show()
````

# Notebook interactivo



````python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

# matriz de la transformación
A = np.array([[3,1],
              [1,3]])

# vectores propios
v1 = np.array([1,1])
v2 = np.array([1,-1])

v1 = v1/np.linalg.norm(v1)
v2 = v2/np.linalg.norm(v2)

# malla del plano
x = np.linspace(-5,5,20)
y = np.linspace(-5,5,20)

X,Y = np.meshgrid(x,y)
points = np.vstack([X.ravel(),Y.ravel()])


def visualizar(t=0, mostrar_propios=True):

    plt.figure(figsize=(8,8))

    # interpolación entre identidad y A
    T = (1-t)*np.eye(2) + t*A

    transformed = T @ points

    # plano original
    plt.scatter(points[0], points[1], alpha=0.25, label="Plano original")

    # plano transformado
    plt.scatter(transformed[0], transformed[1], alpha=0.25,
                label="Plano deformado")

    if mostrar_propios:

        Tv1 = T @ v1
        Tv2 = T @ v2

        plt.quiver(0,0,v1[0],v1[1],
                   angles='xy',scale_units='xy',scale=1,
                   label="v1 dirección propia", color='red')

        plt.quiver(0,0,v2[0],v2[1],
                   angles='xy',scale_units='xy',scale=1,
                   label="v2 dirección propia", color='blue')

        plt.quiver(0,0,Tv1[0],Tv1[1],
                   angles='xy',scale_units='xy',scale=1,
                   label="T(v1)", color='green')

        plt.quiver(0,0,Tv2[0],Tv2[1],
                   angles='xy',scale_units='xy',scale=1,
                   label="T(v2)", color='orange')

    plt.axhline(0)
    plt.axvline(0)

    plt.xlim(-10,10)
    plt.ylim(-10,10)

    plt.grid()
    plt.legend()
    plt.title("Deformación continua del plano por una transformación lineal")

    plt.show()


interact(
    visualizar,
    t=widgets.FloatSlider(min=0,max=1,step=0.05,value=0),
    mostrar_propios=True
)
````

La pregunta que podemos resolver en este momento es la siguiente: ¿qué relación existe entre las diferentes representaciones matriciales de una transformación lineal dada?  Veámoslo para el caso en que la transformación lineal está definida de $V$ en $V$.

## Teorema 5  

Sea $T: V \to V$ una transformación lineal con matriz $A_{T}$ respecto a una base $B_{1}$ y matriz $B_{T}$ respecto a una base $B_{2}$. Si $P$ es la matriz de transición de la base $B_{2}$ a la base $B_{1}$, entonces: 

$$
B_{T} = P^{-1}A_{T}P
$$


## Definición 2  

Dos matrices $A$ y $B$ de $n \times n$ son similares o semejantes si existe una matriz invertible $P$ tal que 

$$
B = P^{-1}AP
$$

# Teorema 6 Propiedades de las matrices similares.  

Dos matrices cuadradas $A$ y $B$ de tamaño $n \times n$ se dicen **similares** si existe una matriz invertible $P$ tal que

$$
B = P^{-1}AP.
$$

Esto significa que representan **la misma transformación lineal en bases diferentes**.

A partir de esta definición se desprenden varias propiedades fundamentales.

---



# Propiedades de las matrices similares

## 1. Representan la misma transformación lineal en distintas bases

Si $A$ y $B$ son similares, entonces ambas matrices representan la **misma transformación lineal**, pero con respecto a bases distintas del espacio vectorial.

Geométricamente:

* la transformación es la misma
* solo cambia el sistema de coordenadas.

---



# 2. Tienen el mismo determinante

$$
\det(A)=\det(B)
$$

Demostración breve:

$$
\det(B)=\det(P^{-1}AP)
$$

$$
= \det(P^{-1})\det(A)\det(P)
$$

$$
= \det(A)
$$

---



# 3. Tienen la misma traza

$$
\text{tr}(A)=\text{tr}(B)
$$

Esto se debe a que la traza es invariante bajo transformaciones de similitud.

---



# 4. Tienen el mismo polinomio característico

$$
p_A(\lambda)=p_B(\lambda)
$$

Es decir

$$
\det(A-\lambda I)=\det(B-\lambda I)
$$

Por lo tanto comparten los mismos **valores propios**.

---



# 5. Tienen el mismo rango

$$
\operatorname{rg}(A)=\operatorname{rg}(B)
$$

Esto ocurre porque multiplicar por matrices invertibles no cambia la dimensión del espacio columna.

---



# 6. Tienen la misma nulidad

$$
\text{nul}(A)=\text{nul}(B)
$$

Como consecuencia directa del rango.

---



# Consecuencia importante

También se preservan:

* los **autovalores**
* la **multiplicidad algebraica**
* el **polinomio mínimo**
* la **invertibilidad**.

---



# Interpretación para estudiantes

Las matrices similares son:

> La misma transformación vista desde sistemas de coordenadas distintos.

Por eso **las propiedades esenciales no cambian**.

Solo cambia la forma de la matriz.

---



# Frase pedagógica útil en clase

Puedes decir:

> Cambiar la base no cambia la transformación, solo cambia la forma en que la escribimos.

---


Te dejo una **demostración visual muy clara para estudiantes** de que dos matrices similares representan la **misma transformación en diferentes bases**.

La idea matemática es:

$$
B = P^{-1}AP
$$

* $A$ es la transformación en la base canónica
* $B$ es la misma transformación en otra base
* $P$ cambia coordenadas.

---



# Script para Jupyter Notebook



````python
import numpy as np
import matplotlib.pyplot as plt

# Transformación original
A = np.array([[3,1],
              [1,3]])

# Nueva base (vectores propios)
P = np.array([[1,1],
              [1,-1]])

# Matriz en la nueva base
B = np.linalg.inv(P) @ A @ P

print("Matriz en base canónica A:\n",A)
print("\nMatriz en nueva base B:\n",B)

# -------------------------

# puntos del plano
x = np.linspace(-4,4,20)
y = np.linspace(-4,4,20)

X,Y = np.meshgrid(x,y)

points = np.vstack([X.ravel(),Y.ravel()])

# transformación usando A
TA = A @ points

# coordenadas en la nueva base
coords_new = np.linalg.inv(P) @ points

# transformación con B en esa base
TB_coords = B @ coords_new

# volver a base canónica
TB = P @ TB_coords

# -------------------------

plt.figure(figsize=(8,8))

plt.scatter(points[0],points[1],alpha=0.25,label="Plano original")

plt.scatter(TA[0],TA[1],alpha=0.6,label="Transformación con A")

plt.scatter(TB[0],TB[1],alpha=0.4,label="Transformación usando B y cambio de base")

plt.axhline(0)
plt.axvline(0)

plt.xlim(-10,10)
plt.ylim(-10,10)

plt.grid()
plt.legend()

plt.title("Matrices similares representan la misma transformación")

plt.show()
````

### Transformación con $A$

### Transformación usando $B$

# Notebook interactivo (Jupyter / Colab)


````python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

# matriz original
A = np.array([[3,1],
              [1,3]])

# matriz de cambio de base (vectores propios)
P = np.array([[1,1],
              [1,-1]])

Pinv = np.linalg.inv(P)

# matriz similar
B = Pinv @ A @ P


# plano
x = np.linspace(-4,4,20)
y = np.linspace(-4,4,20)
X,Y = np.meshgrid(x,y)
points = np.vstack([X.ravel(),Y.ravel()])


def visualizar(t=0.0, mostrar_base=True):

    plt.figure(figsize=(8,8))

    # interpolación para animar
    T = (1-t)*np.eye(2) + t*A

    # transformación directa
    TA = T @ points

    # cambio de base
    coords_new = Pinv @ points
    TB_coords = ((1-t)*np.eye(2) + t*B) @ coords_new
    TB = P @ TB_coords

    plt.scatter(points[0],points[1],alpha=0.25,label="Plano original")

    plt.scatter(TA[0],TA[1],alpha=0.7,label="Usando A")

    plt.scatter(TB[0],TB[1],alpha=0.4,label="Usando B + cambio de base")

    if mostrar_base:
        plt.quiver(0,0,P[0,0],P[1,0],angles='xy',scale_units='xy',scale=1,label="Base nueva v1", color='red')
        plt.quiver(0,0,P[0,1],P[1,1],angles='xy',scale_units='xy',scale=1,label="Base nueva v2", color='blue')

    plt.axhline(0)
    plt.axvline(0)

    plt.xlim(-10,10)
    plt.ylim(-10,10)

    plt.grid()
    plt.legend()

    plt.title("Matrices similares → misma transformación")

    plt.show()


interact(
    visualizar,
    t=widgets.FloatSlider(min=0,max=1,step=0.05,value=0),
    mostrar_base=True
)
````

# Notebook interactivo con dos sliders


````python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

# matriz de la transformación
A = np.array([[3,1],
              [1,3]])

# plano
x = np.linspace(-4,4,20)
y = np.linspace(-4,4,20)
X,Y = np.meshgrid(x,y)
points = np.vstack([X.ravel(),Y.ravel()])


def visualizar(t=1.0, angulo=0):

    plt.figure(figsize=(8,8))

    # matriz de rotación para construir una nueva base
    theta = np.deg2rad(angulo)
    P = np.array([
        [np.cos(theta), -np.sin(theta)],
        [np.sin(theta),  np.cos(theta)]
    ])

    Pinv = np.linalg.inv(P)

    # matriz en la nueva base
    B = Pinv @ A @ P

    # interpolación para animar
    T = (1-t)*np.eye(2) + t*A

    # transformación directa
    TA = T @ points

    # transformación usando cambio de base
    coords_new = Pinv @ points
    TB_coords = ((1-t)*np.eye(2) + t*B) @ coords_new
    TB = P @ TB_coords

    plt.scatter(points[0],points[1],alpha=0.25,label="Plano original")
    plt.scatter(TA[0],TA[1],alpha=0.7,label="Usando A")
    plt.scatter(TB[0],TB[1],alpha=0.4,label="Usando B + cambio de base")

    # base actual
    plt.quiver(0,0,P[0,0],P[1,0],angles='xy',scale_units='xy',scale=1,label="v1 base", color='red')
    plt.quiver(0,0,P[0,1],P[1,1],angles='xy',scale_units='xy',scale=1,label="v2 base", color='blue')

    plt.axhline(0)
    plt.axvline(0)

    plt.xlim(-10,10)
    plt.ylim(-10,10)

    plt.grid()
    plt.legend()

    plt.title(f"Matriz en la base actual:\n{np.round(B,2)}")

    plt.show()


interact(
    visualizar,
    t=widgets.FloatSlider(min=0,max=1,step=0.05,value=1),
    angulo=widgets.FloatSlider(min=0,max=180,step=5,value=0)
)
````

# Línea argumentativa (Marco teórico)



El desarrollo del **pensamiento computacional** constituye hoy un eje fundamental en la educación matemática. Como lo plantea **Wing (2006)**, esta forma de razonamiento debe considerarse una competencia transversal y básica, al nivel de la lectura, la escritura y las matemáticas. 

En consecuencia, fomentar su aprendizaje desde los niveles iniciales hasta la educación superior resulta indispensable para formar profesionales capaces de enfrentar los retos de la sociedad contemporánea.



En el campo de la didáctica de las matemáticas, **Papert (1980, 1996)** había anticipado esta necesidad al concebir las computadoras no como simples herramientas de cálculo, sino como **“objetos para pensar”**, capaces de favorecer la construcción activa de significados matemáticos a través de la programación. 

Esta visión se alinea con el constructivismo de **Piaget**, que resalta el valor de la experimentación en el aprendizaje, y con la perspectiva socio-constructivista de **Vygotsky**, que otorga un papel central a las herramientas culturales en la mediación del conocimiento.



En este marco, el uso de entornos como **Jupyter Notebooks** y lenguajes como **Python, LaTeX, Markdown e inglés** no debe interpretarse como una rivalidad con la enseñanza tradicional de la matemática, sino como una **ampliación de los registros de representación** que describió **Duval (1993, 2006)**. 

Según su teoría semiótica, el aprendizaje significativo se potencia cuando el estudiante logra traducir y articular distintos sistemas de representación: gráfico, algebraico, numérico, verbal y, en nuestro tiempo, digital.



De igual manera, el **Enfoque Ontosemiótico de Godino (2002, 2017)** muestra que los significados matemáticos emergen de prácticas diversas, lo que justifica la incorporación de estos lenguajes como recursos que enriquecen las configuraciones de práctica matemática, en sintonía con las exigencias de la investigación y la práctica profesional.



En el ámbito internacional, organizaciones como el **NCTM (2000, 2014)** han subrayado que la tecnología es esencial para la enseñanza y el aprendizaje de la matemática, dado que facilita la experimentación, la validación de conjeturas y la comunicación de ideas. 

Asimismo, instancias como la **ACM** y el **IEEE** han señalado la importancia de desarrollar habilidades computacionales como requisito fundamental en la formación universitaria, tanto en el sector público como privado.



Por tanto, enseñar matemáticas integrando Python, LaTeX, Markdown e inglés en ambientes como Jupyter Notebooks contribuye a la formación de profesionales con **competencias científicas, comunicativas y tecnológicas**, capaces de comprender la matemática de manera interactiva y experimental, y de responder a las demandas del mundo laboral. 

Esta integración, lejos de ser inadecuada, constituye un **puente entre la enseñanza académica y las necesidades de la sociedad**, fortaleciendo la pertinencia de los programas universitarios.

# Estado del arte



El interés por integrar la tecnología en la enseñanza de las matemáticas tiene raíces en las propuestas de **Papert (1980, 1996)**, quien con su enfoque constructionista concibió la computadora como un “objeto para pensar”, capaz de mediar la construcción activa de significados matemáticos mediante la programación. 

Esta perspectiva se enlaza con los planteamientos de **Piaget (1970)** y **Vygotsky (1978)**, quienes reconocieron la importancia de la experimentación y de las herramientas culturales en la formación del pensamiento. 

Bajo esta mirada, el uso de entornos digitales no constituye una amenaza para la enseñanza tradicional de las matemáticas, sino un recurso que amplía las posibilidades de aprendizaje.



A comienzos del siglo XXI, la noción de **pensamiento computacional** introducida por **Wing (2006)** renovó el debate al proponer que esta competencia debía enseñarse de manera transversal desde los niveles iniciales hasta la educación superior, situándola al mismo nivel que la lectura, la escritura y las matemáticas. 

Este enfoque se alinea con los planteamientos del **NCTM (2000, 2014)**, que destacó la tecnología como un componente esencial para la experimentación, la representación y la comunicación en matemáticas. 

En paralelo, aportes teóricos como la **teoría de registros de representación semiótica de Duval (1993, 2006)** y el **Enfoque Ontosemiótico de Godino (2007, 2017)** muestran que el aprendizaje matemático se enriquece al transitar entre múltiples sistemas de representación, incluyendo los digitales.



En el contexto actual, organizaciones como la **ACM** y el **IEEE** insisten en que las competencias computacionales son indispensables en la formación profesional, lo que refuerza la pertinencia de integrar lenguajes como **Python, LaTeX, Markdown e inglés** en ambientes como **Jupyter Notebooks** para la enseñanza universitaria de las matemáticas. 

Así, el estado del arte evidencia un consenso creciente: lejos de ser un obstáculo, el pensamiento computacional y el uso de lenguajes digitales constituyen una vía para fortalecer la comprensión matemática, formar profesionales más competentes y responder a las demandas del mundo laboral y científico contemporáneo.

# Estado del arte con citas



El uso de la tecnología en la enseñanza de las matemáticas encuentra uno de sus pilares en **Papert (1980)**, quien afirmaba que *“los niños no aprenden sobre la computadora, aprenden con la computadora”* (p. 6). 

En su visión, la programación se convierte en un medio para la construcción activa de significados matemáticos, idea que se enlaza con la epistemología genética de **Piaget (1970)**, quien sostuvo que *“el conocimiento se construye a través de la acción”* (p. 15), y con la perspectiva de **Vygotsky (1978)**, al señalar que *“toda función en el desarrollo cultural del niño aparece dos veces: primero en el plano social y después en el individual”* (p. 57). 

Estos enfoques coinciden en otorgar a las herramientas culturales un papel esencial en la mediación del aprendizaje.



Posteriormente, **Wing (2006)** acuñó el término pensamiento computacional, definiéndolo como *“una habilidad fundamental para todos, no solo para los científicos de la computación”* (p. 33). 

Esta autora subrayó que debía enseñarse de manera transversal, en paralelo con la lectura, la escritura y la matemática. 

En la misma línea, el **NCTM (2000)** destacó que *“la tecnología es esencial en la enseñanza y el aprendizaje de las matemáticas; influye en la matemática que se enseña y mejora el aprendizaje de los estudiantes”* (p. 24).



Desde la didáctica de las matemáticas, **Duval (2006)** mostró que *“la comprensión matemática requiere la coordinación de al menos dos registros de representación diferentes”* (p. 107), lo que legitima la incorporación de registros digitales como LaTeX, Markdown y Python. 

Por su parte, **Godino (2017)** planteó que *“el significado de un objeto matemático está constituido por el sistema de prácticas en el que interviene”* (p. 7), con lo cual la práctica digital se convierte en parte constitutiva del aprendizaje matemático.



Finalmente, organismos internacionales como la **ACM y el IEEE (2013)** enfatizan que *“las habilidades computacionales son esenciales para el ejercicio profesional en cualquier disciplina”* (p. 12). De esta manera, se consolida un consenso: la enseñanza universitaria de las matemáticas debe integrar herramientas contemporáneas como **Jupyter Notebooks, Python, LaTeX, Markdown e inglés**, no para competir con la enseñanza tradicional, sino para **fortalecer la comprensión matemática, potenciar la comunicación científica y responder a las demandas del mundo laboral**.
