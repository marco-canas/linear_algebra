# 12.1  Propiedades de las transformaciones lineales  



## Teorema 1  
Sea $T: V \to W$ una transformación lineal entre espacios vectoriales sobre $\mathbb{R}$. Entonces se cumplen las siguientes propiedades fundamentales:

---



### a) Preservación del vector cero

La transformación lineal envía el vector cero de (V) en el vector cero de (W):

$$
T(0_V)=0_W.
$$

---



### b) Compatibilidad con la resta (distributividad respecto a la suma y el opuesto)

Para todo $u,v \in V$,

$$
T(u-v)=T(u)-T(v).
$$

Esta propiedad se deduce de la linealidad, pues:
$$
T(u-v)=T(u+(-v))=T(u)+T(-v)= T(u)+T((-1)v) = T(u)-T(v).
$$

---



### c) Preservación de combinaciones lineales

Para cualesquiera vectores $v_1,\dots,v_n \in V$ y escalares $\alpha_1,\dots,\alpha_n \in \mathbb{R}$, se cumple:

$$
T(\alpha_1 v_1 + \alpha_2 v_2 + \cdots + \alpha_n v_n)
=
\alpha_1 T(v_1) + \alpha_2 T(v_2) + \cdots + \alpha_n T(v_n).
$$

Esta propiedad caracteriza a las transformaciones lineales: transforman combinaciones lineales en las correspondientes combinaciones lineales de las imágenes.

---



Estas tres propiedades se derivan directamente de la definición de transformación lineal:

$$
T(u+v)=T(u)+T(v),
\qquad
T(\lambda v)=\lambda T(v),
\quad \forall, u,v\in V,; \lambda\in\mathbb{R}.
$$


# Observación  

Es importante notar que en la parte a del teorema anterior, el cero del  lado izquierdo
de la igualdad es el cero de $V$ y el del lado derecho es el de $W$.  

Además la parte c del teorema es una generalización de las partes a y b.  

En general, cuando se define una función de $V$ en $W$, ésta se especifica mediante una
regla que asigna a cada elemento de $V$ un único elemento de $W$, ya que sería imposible decir para cada uno de los elementos  de $V$ cuál es el asignado en $W$ debido a que se trata de la asignación de una infinidad de elementos.   

Sin embargo, cuando se trata de una transformación lineal, es posible saber cómo está definida $T$ en todo el espacio vectorial $V$, conociendo lo que hace $T$ a una base de $V$.   

Así que en un espacio de dimensión finita, es posible describir $T$ proporcionando sólo las imágenes de un conjunto finito de vectores.

Sea $T: V \to W$ una transformación lineal, donde $V$ es un espacio vectorial de dimensión $n$.

**Teorema (Determinación de una transformación lineal por su acción sobre una base).**
Sea $B=\{v_1,\dots,v_n\}$ una base de $V$. Entonces, la transformación lineal $T$ queda completamente determinada por los valores $T(v_1),\dots,T(v_n)$.

Es decir, para todo vector $v\in V$, existe una única combinación lineal
$$
v=\alpha_1 v_1+\alpha_2 v_2+\cdots+\alpha_n v_n,
$$
con $\alpha_1,\dots,\alpha_n\in\mathbb{R}$, y por la linealidad de $T$ se tiene:

$$
T(v)=\alpha_1 T(v_1)+\alpha_2 T(v_2)+\cdots+\alpha_n T(v_n).
$$

En consecuencia, conocer la imagen de los vectores de una base de $V$ determina de manera única la imagen de cualquier vector del espacio.

Equivalentemente: si $S:V\to W$ es otra transformación lineal tal que $S(v_i)=T(v_i)$ para todo $i=1,\dots,n$, entonces $S=T$.


Veamos un ejemplo concreto que ilustra que una transformación lineal queda completamente determinada por su acción sobre una base.

### Ejemplo para $T:\mathbb{R}^2 \to \mathbb{R}^3$

Tomemos como base canónica de $\mathbb{R}^2$:
$$
B=\{e_1=(1,0), e_2=(0,1)\}.
$$



Definimos la transformación lineal $T$ indicando únicamente las imágenes de los vectores de la base:
$$
T(e_1)=(1,2,0), \qquad T(e_2)=(0,-1,3).
$$

Con esta información, la transformación queda totalmente determinada.

---



### Cómo se obtiene $T(v)$ para cualquier vector

Sea un vector cualquiera $v=(x,y)\in\mathbb{R}^2$.
Como $B$ es base,
$$
v = x e_1 + y e_2.
$$



Por linealidad,
$$
T(v)=T(xe_1+ye_2)=xT(e_1)+yT(e_2).
$$



Sustituyendo las imágenes definidas:
$$
T(v)=x(1,2,0)+y(0,-1,3).
$$



Operando:
$$
T(v)=(x,2x-y,3y).
$$

---



### Interpretación

* Conocer solo $T(e_1)$ y $T(e_2)$ permite calcular $T(v)$ para cualquier vector de $\mathbb{R}^2$.
* No es necesario definir $T$ en todos los vectores: basta con hacerlo sobre una base.
* La transformación queda completamente fijada por esas dos imágenes.

En forma matricial, la matriz asociada a $T$ (respecto a la base canónica) es:
$$
A=
\begin{pmatrix}
1 & 0\\ 
2 & -1\\ 
0 & 3
\end{pmatrix},
\quad \text{y} \quad
T(x,y)=A
\begin{pmatrix}
x\\ y
\end{pmatrix}.
$$



Esto ejemplifica el teorema: la acción de $T$ sobre la base determina su acción sobre todo $\mathbb{R}^2$.


Sea $T: V \to W$ una transformación lineal entre espacios vectoriales sobre $\mathbb{R}$.

## Núcleo (kernel) de $T$



El **núcleo** de $T$, denotado por $\ker(T)$ o $\operatorname{Nuc}(T)$, es el conjunto de todos los vectores de $V$ que son enviados al vector cero de $W$:

$$
\ker(T)={,v\in V: T(v)=0_W,}.
$$



**Interpretación:**

* Son los vectores que la transformación “anula”.
* Mide cuánta información se pierde al aplicar $T$.

**Propiedad importante:**
$\ker(T)$ es un subespacio vectorial de $V$.

---



## Imagen (rango) de $T$

La **imagen** de $T$, denotada por $\operatorname{Im}(T)$ o $T(V)$, es el conjunto de todos los vectores de $W$ que se obtienen como imagen de algún vector de $V$:

$$
\operatorname{Im}(T)=\{T(v)\in W : v \in V\}.
$$

**Interpretación:**

* Es el conjunto de todos los valores que realmente alcanza la transformación.
* Describe “hacia dónde llega” $T$ dentro de $W$.

**Propiedad importante:**
$\operatorname{Im}(T)$ es un subespacio vectorial de $W$.

---



## Relación conceptual

* El núcleo describe los vectores que se colapsan a cero.
* La imagen describe los vectores que sí se producen como salida.
* Ambas nociones son fundamentales para el estudio de transformaciones lineales y se conectan mediante el teorema de rango-nulidad:
  $$
  \dim(V)=\dim(\ker T)+\dim(\operatorname{Im} T).
  $$


# Definición de nulidad y rango de una transformación $T$  

Sea $T: V \to W$ una transformación lineal entre espacios vectoriales de dimensión finita.

## Nulidad de $T$

La **nulidad** de $T$ es la dimensión del núcleo de la transformación:

$$
\operatorname{nul}(T)=\dim(\ker T).
$$

Es decir, mide cuántos grados de libertad tiene el conjunto de vectores de $V$ que son enviados al vector cero de $W$.

**Interpretación:**
Indica cuánta información “se pierde” al aplicar la transformación.

---



## Rango de $T$

El **rango** de $T$ es la dimensión de la imagen de la transformación:

$$
\operatorname{rang}(T)=\dim(\operatorname{Im} T).
$$

Es el número de dimensiones del subespacio de $W$ que realmente es alcanzado por $T$.

**Interpretación:**
Mide la cantidad de direcciones independientes que produce la transformación en el espacio de llegada.

---



## Relación fundamental (teorema rango–nulidad)

Si $\dim(V)=n$, entonces:

$$
n=\operatorname{nul}(T)+\operatorname{rang}(T).
$$

Este resultado establece que la dimensión del espacio de partida se descompone en:

* la parte que se anula (nulidad), y
* la parte que se preserva como imagen independiente (rango).


# Ejemplo  

Consideremos una transformación lineal
$$
T:\mathbb{R}^3 \to \mathbb{R}^2
$$
definida por
$$
T(x,y,z)=(x+y, y+z).
$$



Su matriz asociada en las bases canónicas es
$$
A=
\begin{pmatrix}
1 & 1 & 0\\ 
0 & 1 & 1
\end{pmatrix}.
$$

---



# 1) Espacio nulo de (T)

Buscamos los vectores $(x,y,z)\in\mathbb{R}^3$ tales que $T(x,y,z=0$:

$$
\begin{cases}
x+y=0\\ 
y+z=0
\end{cases}
$$



De aquí:

$$
x=-y,\qquad z=-y.
$$

Sea $y=t$. Entonces:

$$
(x,y,z)=(-t,t,-t)=t(-1,1,-1).
$$



Por tanto,

$$
\ker(T)=\operatorname{span}{(-1,1,-1)}.
$$
Es una recta en $$\mathbb{R}^3$$.

### Nulidad

$$
\operatorname{nul}(T)=\dim(\ker T)=1.
$$

---



# 2) Imagen de (T)

La imagen está generada por las columnas de la matriz $A$:

$$
\text{col}_1=(1,0),\quad
\text{col}_2=(1,1),\quad
\text{col}_3=(0,1).
$$



Observamos que:

$$
(1,1)=(1,0)+(0,1),
$$

por lo que los generadores independientes son:

$$
(1,0),\ (0,1).
$$

Entonces

$$
\operatorname{Im}(T)=\operatorname{span}{(1,0),(0,1)}=\mathbb{R}^2.
$$



### Rango

$$
\operatorname{rang}(T)=\dim(\operatorname{Im}T)=2.
$$

---



# 3) Verificación del teorema rango–nulidad

Como $\dim(\mathbb{R}^3)=3$:

$$
\operatorname{nul}(T)+\operatorname{rang}(T)
=1+2
=3.
$$

Se cumple el teorema:

$$
\dim(V)=\operatorname{nul}(T)+\operatorname{rang}(T).
$$

---



# 4) Lectura geométrica

* El **núcleo** es una recta en $\mathbb{R}^3$ formada por los vectores que la transformación “colapsa” al cero.
* La **imagen** es todo $\mathbb{R}^2$: la transformación alcanza cualquier vector del plano.
* La transformación “pierde” una dimensión (nulidad 1) y conserva dos direcciones independientes (rango 2).


Consideremos la transformación lineal
$$
T:\mathbb{R}^2 \to \mathbb{R}^2
$$
definida por la matriz

$$
A=
\begin{pmatrix}
2 & 1\\ 
1 & 2
\end{pmatrix}.
$$



Entonces,  
$$
T(x,y)=A\begin{pmatrix}x \\ y\end{pmatrix}
=\begin{pmatrix}
2x+y\\ 
x+2y
\end{pmatrix}.
$$

---



# 1) Imagen del cuadrado unitario

El cuadrado unitario en (\mathbb{R}^2) está dado por los vértices:

$$
(0,0),\ (1,0),\ (1,1),\ (0,1).
$$



Aplicamos $T$ a cada vértice:

$$
T(0,0)=(0,0)
$$
$$
T(1,0)=(2,1)
$$
$$
T(1,1)=(3,3)
$$
$$
T(0,1)=(1,2)
$$



Los puntos imagen son:

$$
(0,0),\ (2,1),\ (3,3),\ (1,2).
$$



Este conjunto forma un **paralelogramo**, porque:

$$
(2,1)+(1,2)=(3,3),
$$
lo que muestra que los lados opuestos son paralelos y de igual longitud.



Geométricamente:

* los vectores columna de $A$, $(2,1)$ y $(1,2)$, son los lados del paralelogramo;
* el cuadrado unitario se transforma en el paralelogramo generado por esos vectores.

---



# 2) Script en matplotlib

El siguiente código dibuja:

* a la izquierda: el cuadrado unitario;
* a la derecha: su imagen bajo $T$.



```python
import numpy as np
import matplotlib.pyplot as plt

# Matriz de la transformación lineal
A = np.array([[2, 1],
              [1, 2]])

# Vértices del cuadrado unitario
cuadrado = np.array([
    [0, 0],
    [1, 0],
    [1, 1],
    [0, 1],
    [0, 0]  # cerrar la figura
])

# Aplicar la transformación
imagen = (A @ cuadrado.T).T

# Crear figura
fig, axes = plt.subplots(1, 2, figsize=(10,5))

# --- Cuadrado unitario ---
axes[0].plot(cuadrado[:,0], cuadrado[:,1], linewidth=2)
axes[0].set_title("Cuadrado unitario")
axes[0].axhline(0)
axes[0].axvline(0)
axes[0].set_aspect('equal')
axes[0].grid(True)

# --- Paralelogramo imagen ---
axes[1].plot(imagen[:,0], imagen[:,1], linewidth=2, color='orange')
axes[1].set_title("Imagen bajo T(x)=Ax")
axes[1].axhline(0)
axes[1].axvline(0)
axes[1].set_aspect('equal')
axes[1].grid(True)
plt.savefig(r"C:\Users\marco\Downloads\cuadrado_paralelogramo.png")
plt.show()
```


# 3) Lectura geométrica

* El cuadrado unitario está generado por los vectores base $(1,0)$ y $(0,1)$.
* La transformación lineal envía esos vectores en:
  $$
  T(1,0)=(2,1), \quad T(0,1)=(1,2).
  $$
* Por tanto, la imagen del cuadrado es el paralelogramo generado por esos dos vectores.
* Esto ilustra una propiedad general: **toda transformación lineal envía cuadrados en paralelogramos (o figuras degeneradas si el determinante es cero).**


# Ejemplo 4  

Un modo sencillo de transformar la esfera unitaria en un elipsoide es usar una transformación lineal representada por una matriz diagonal con factores de escala distintos en cada eje.

---



# Ejemplo de transformación lineal

Sea
$$
T:\mathbb{R}^3 \to \mathbb{R}^3,\qquad T(x)=Ax,
$$
con

$$
A=
\begin{pmatrix}
2 & 0 & 0\\ 
0 & 1 & 0\\ 
0 & 0 & 0.5
\end{pmatrix}.
$$



Esta transformación:

* estira en dirección $x$ por factor 2,
* deja igual la dirección $y$,
* comprime en dirección $z$ por factor $0.5$.



La esfera unitaria:

$$
x^2+y^2+z^2=1
$$



se transforma en el elipsoide:

$$
\frac{x^2}{2^2}+\frac{y^2}{1^2}+\frac{z^2}{0.5^2}=1.
$$

---



# Script en matplotlib

El siguiente código dibuja:

* a la izquierda: la esfera unitaria
* a la derecha: el elipsoide imagen bajo $T$



```python
import numpy as np
import matplotlib.pyplot as plt

# =========================
# Matriz de la transformación lineal
# =========================
A = np.array([[2, 0, 0],
              [0, 1, 0],
              [0, 0, 0.5]])

# =========================
# Parametrización de la esfera unitaria
# =========================
u = np.linspace(0, 2*np.pi, 60)
v = np.linspace(0, np.pi, 60)

x = np.outer(np.cos(u), np.sin(v))
y = np.outer(np.sin(u), np.sin(v))
z = np.outer(np.ones(np.size(u)), np.cos(v))

# =========================
# Aplicar la transformación lineal
# =========================
points = np.stack([x, y, z], axis=0).reshape(3, -1)
transformed = A @ points

x_e = transformed[0].reshape(x.shape)
y_e = transformed[1].reshape(y.shape)
z_e = transformed[2].reshape(z.shape)

# =========================
# Graficar
# =========================
fig = plt.figure(figsize=(12,5))

# --- Esfera unitaria ---
ax1 = fig.add_subplot(121, projection='3d')
ax1.plot_surface(x, y, z, alpha=0.6)
ax1.set_title("Esfera unitaria")
ax1.set_box_aspect([1,1,1])

# --- Elipsoide imagen ---
ax2 = fig.add_subplot(122, projection='3d')
ax2.plot_surface(x_e, y_e, z_e, alpha=0.6)
ax2.set_title("Imagen bajo T(x)=Ax (Elipsoide)")
ax2.set_box_aspect([2,1,0.5])
plt.savefig(r"C:\Users\marco\Downloads\esfera_elipsoide.png")
plt.show()
```


# Interpretación geométrica

* Cada punto de la esfera se transforma multiplicando por $A$.
* Los ejes del elipsoide coinciden con los ejes coordenados.
* Las longitudes de sus semiejes son los factores de escala de la matriz: $2$, $1$ y $0.5$.
* En general, **toda transformación lineal envía la esfera unitaria en un elipsoide** (posiblemente rotado si la matriz no es diagonal).


# Ejemplo 5  

Sea $T:M_{nn} \to M_{nn}$ tal que  $T(A)  =  A^{T} + A$.  Pruebe que $T$ es una transformación
lineal y determine núcleo e imagen.

Sea $T:M_{nn}\to M_{nn}$ definida por $T(A)=A^{T}+A$. Se quiere probar que $T$ es lineal y determinar su núcleo e imagen.

---

## 1. Linealidad de $T$

Para demostrar que $T$ es lineal debemos verificar:

* $T(A+B)=T(A)+T(B)$
* $T(\lambda A)=\lambda T(A)$

para toda $A,B\in M_{nn}$ y todo escalar $\lambda$.

### Aditividad

$$
T(A+B)=(A+B)^T+(A+B)
$$

Usando propiedades de la transpuesta:

$$
(A+B)^T=A^T+B^T
$$

Entonces:

$$
T(A+B)=A^T+B^T+A+B
$$

Agrupando:

$$
T(A+B)=(A^T+A)+(B^T+B)=T(A)+T(B)
$$

---

### Homogeneidad

$$
T(\lambda A)=(\lambda A)^T+\lambda A
$$

Como $(\lambda A)^T=\lambda A^T$:

$$
T(\lambda A)=\lambda A^T+\lambda A=\lambda(A^T+A)=\lambda T(A)
$$

Por lo tanto, $T$ es una transformación lineal.

---



## 2. Núcleo de $T$

El núcleo está dado por:

$$
\ker(T)={A\in M_{nn}:T(A)=0}
$$



Esto implica:

$$
A^T+A=0
$$



Equivalente a:

$$
A^T=-A
$$

Estas son precisamente las matrices antisimétricas (o skew-symmetric).



Un elemento típico del núcleo es:

$$
A=
\begin{pmatrix}
0 & a_{12} & a_{13} & \cdots & a_{1n} \\ 
-a_{12} & 0 & a_{23} & \cdots & a_{2n} \\ 
-a_{13} & -a_{23} & 0 & \cdots & a_{3n} \\ 
\vdots & \vdots & \vdots & \ddots & \vdots \\ 
-a_{1n} & -a_{2n} & -a_{3n} & \cdots & 0
\end{pmatrix}
$$



Por tanto:

$$
\ker(T)={A\in M_{nn}:A^T=-A}
$$

---



## 3. Imagen de $T$

Sea $B\in \operatorname{Im}(T)$. Entonces existe $A\in M_{nn}$ tal que:

$$
B=T(A)=A^T+A
$$



Observemos que:

$$
B^T=(A^T+A)^T=A+A^T=B
$$

Por lo tanto, $B$ es una matriz simétrica.



Recíprocamente, si $B$ es simétrica ($B^T=B$), podemos tomar:

$$
A=\frac{1}{2}B
$$



y entonces:

$$
T(A)=A^T+A=\frac{1}{2}B+\frac{1}{2}B=B
$$



Luego, toda matriz simétrica pertenece a la imagen.

Así:

$$
\operatorname{Im}(T)={B\in M_{nn}:B^T=B}
$$

---



## Conclusión

* $T(A)=A^T+A$ es una transformación lineal.
* Su núcleo es el subespacio de matrices antisimétricas:

$$
\ker(T)={A\in M_{nn}:A^T=-A}
$$



* Su imagen es el subespacio de matrices simétricas:

$$
\operatorname{Im}(T)=\{ B\in M_{nn}:B^T=B \}
$$

Es decir, la transformación descompone cualquier matriz en su parte simétrica y antisimétrica.

# Línea argumentativa (Marco teórico)



El desarrollo del **pensamiento computacional** constituye hoy un eje fundamental en la educación matemática. Como lo plantea **Wing (2006)**, esta forma de razonamiento debe considerarse una competencia transversal y básica, al nivel de la lectura, la escritura y las matemáticas. 

En consecuencia, fomentar su aprendizaje desde los niveles iniciales hasta la educación superior resulta indispensable para formar profesionales capaces de enfrentar los retos de la sociedad contemporánea.



En el campo de la didáctica de las matemáticas, **Papert (1980, 1996)** había anticipado esta necesidad al concebir las computadoras no como simples herramientas de cálculo, sino como **“objetos para pensar”**, capaces de favorecer la construcción activa de significados matemáticos a través de la programación. 

Esta visión se alinea con el constructivismo de **Piaget**, que resalta el valor de la experimentación en el aprendizaje, y con la perspectiva socio-constructivista de **Vygotsky**, que otorga un papel central a las herramientas culturales en la mediación del conocimiento.



En este marco, el uso de entornos como **Jupyter Notebooks** y lenguajes como **Python, LaTeX, Markdown e inglés** no debe interpretarse como una rivalidad con la enseñanza tradicional de la matemática, sino como una **ampliación de los registros de representación** que describió **Duval (1993, 2006)**. 

Según su teoría semiótica, el aprendizaje significativo se potencia cuando el estudiante logra traducir y articular distintos sistemas de representación: gráfico, algebraico, numérico, verbal y, en nuestro tiempo, digital.



De igual manera, el **Enfoque Ontosemiótico de Godino (2002, 2017)** muestra que los significados matemáticos emergen de prácticas diversas, lo que justifica la incorporación de estos lenguajes como recursos que enriquecen las configuraciones de práctica matemática, en sintonía con las exigencias de la investigación y la práctica profesional.



En el ámbito internacional, organizaciones como el **NCTM (2000, 2014)** han subrayado que la tecnología es esencial para la enseñanza y el aprendizaje de la matemática, dado que facilita la experimentación, la validación de conjeturas y la comunicación de ideas. 

Asimismo, instancias como la **ACM** y el **IEEE** han señalado la importancia de desarrollar habilidades computacionales como requisito fundamental en la formación universitaria, tanto en el sector público como privado.



Por tanto, enseñar matemáticas integrando Python, LaTeX, Markdown e inglés en ambientes como Jupyter Notebooks contribuye a la formación de profesionales con **competencias científicas, comunicativas y tecnológicas**, capaces de comprender la matemática de manera interactiva y experimental, y de responder a las demandas del mundo laboral. 

Esta integración, lejos de ser inadecuada, constituye un **puente entre la enseñanza académica y las necesidades de la sociedad**, fortaleciendo la pertinencia de los programas universitarios.
