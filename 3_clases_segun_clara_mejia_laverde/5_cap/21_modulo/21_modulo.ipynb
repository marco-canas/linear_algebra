# **Idea matemática**

El cono (doble cono) se modela como:

$$
z^2 = x^2 + y^2
$$

o equivalentemente:

$$
z = \pm \sqrt{x^2 + y^2}
$$



Las secciones se obtienen al intersectar con planos:

* Circunferencia → plano horizontal
* Elipse → plano inclinado (menos inclinado que la generatriz)
* Parábola → plano paralelo a la generatriz
* Hipérbola → plano que corta ambas hojas

---



# **Código completo en Plotly (3D interactivo)**


````python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Crear figura con dos paneles
fig = make_subplots(
    rows=1, cols=2,
    specs=[[{"type": "scene"}, {"type": "scene"}]],
    subplot_titles=("Cono de Apolonio", "Secciones cónicas")
)

# =========================
# CONO (doble)
# =========================
u = np.linspace(0, 2*np.pi, 100)
v = np.linspace(-5, 5, 100)
U, V = np.meshgrid(u, v)

X = V * np.cos(U)
Y = V * np.sin(U)
Z = V

fig.add_trace(go.Surface(
    x=X, y=Y, z=Z,
    opacity=0.5,
    showscale=False,
), row=1, col=1)

fig.add_trace(go.Surface(
    x=X, y=Y, z=-Z,
    opacity=0.5,
    showscale=False,
), row=1, col=1)

# =========================
# PLANOS Y SECCIONES
# =========================

t = np.linspace(0, 2*np.pi, 200)

# Circunferencia (plano horizontal)
z0 = 2
x_circ = z0 * np.cos(t)
y_circ = z0 * np.sin(t)
z_circ = np.full_like(t, z0)

fig.add_trace(go.Surface(
    x=X, y=Y, z=np.full_like(Z, z0),
    opacity=0.4,
    colorscale="Reds",
    showscale=False
), row=1, col=2)

fig.add_trace(go.Scatter3d(
    x=x_circ, y=y_circ, z=z_circ,
    mode='lines',
    line=dict(color='red', width=6),
    name="Circunferencia"
), row=1, col=2)

# Elipse (plano inclinado)
x_el = 3*np.cos(t)
y_el = 2*np.sin(t)
z_el = 2 + 0.3*x_el

fig.add_trace(go.Surface(
    x=X, y=Y, z=2 + 0.3*X,
    opacity=0.4,
    colorscale="Blues",
    showscale=False
), row=1, col=2)

fig.add_trace(go.Scatter3d(
    x=x_el, y=y_el, z=z_el,
    mode='lines',
    line=dict(color='blue', width=6),
    name="Elipse"
), row=1, col=2)

# Parábola (plano paralelo a generatriz)
x_par = np.linspace(-3, 3, 200)
y_par = x_par**2 / 4
z_par = y_par

fig.add_trace(go.Surface(
    x=X, y=Y, z=Y,
    opacity=0.4,
    colorscale="Greens",
    showscale=False
), row=1, col=2)

fig.add_trace(go.Scatter3d(
    x=x_par, y=y_par, z=z_par,
    mode='lines',
    line=dict(color='green', width=6),
    name="Parábola"
), row=1, col=2)

# Hipérbola (plano paralelo al eje del cono)
x_hip = np.linspace(-4, 4, 200)
y_hip = np.sqrt(np.abs(x_hip**2 - 1))
z_hip = x_hip

fig.add_trace(go.Surface(
    x=X, y=Y, z=X,
    opacity=0.4,
    colorscale="Purples",
    showscale=False
), row=1, col=2)

fig.add_trace(go.Scatter3d(
    x=x_hip, y=y_hip, z=z_hip,
    mode='lines',
    line=dict(color='purple', width=6),
    name="Hipérbola"
), row=1, col=2)

# =========================
# AJUSTES FINALES
# =========================

fig.update_layout(
    title="Cono de Apolonio y Secciones Cónicas",
    height=700,
    showlegend=True
)

fig.show()
fig.write_image(r"C:\Users\marco\Downloads\cono_de_apolonio.png")
````

````python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Malla
r = np.linspace(0, 5, 100)
theta = np.linspace(0, 2*np.pi, 100)

R, T = np.meshgrid(r, theta)

X = R * np.cos(T)
Y = R * np.sin(T)
Z1 = R
Z2 = -R

# Crear subplots
fig = make_subplots(
    rows=1, cols=2,
    specs=[[{'type': 'surface'}, {'type': 'surface'}]],
    subplot_titles=("Cono de Apolonio", "Secciones cónicas en el cono")
)

# CONO (izquierda)
fig.add_trace(go.Surface(x=X, y=Y, z=Z1, opacity=0.6), row=1, col=1)
fig.add_trace(go.Surface(x=X, y=Y, z=Z2, opacity=0.6), row=1, col=1)

# CONO (derecha)
fig.add_trace(go.Surface(x=X, y=Y, z=Z1, opacity=0.3, showscale=False), row=1, col=2)
fig.add_trace(go.Surface(x=X, y=Y, z=Z2, opacity=0.3, showscale=False), row=1, col=2)

# -------------------------
# PLANOS DE CORTE
# -------------------------

xx = np.linspace(-5,5,50)
yy = np.linspace(-5,5,50)
XX, YY = np.meshgrid(xx, yy)

# Circunferencia (plano horizontal)
Z_circ = np.ones_like(XX)*2

fig.add_trace(go.Surface(
    x=XX, y=YY, z=Z_circ,
    opacity=0.5,
    showscale=False,
    name="Circunferencia"
), row=1, col=2)

# Elipse (plano inclinado suave)
Z_ellipse = 0.5*XX + 2

fig.add_trace(go.Surface(
    x=XX, y=YY, z=Z_ellipse,
    opacity=0.5,
    showscale=False,
    name="Elipse"
), row=1, col=2)

# Parábola (plano paralelo a generatriz)
Z_parab = XX + 1

fig.add_trace(go.Surface(
    x=XX, y=YY, z=Z_parab,
    opacity=0.5,
    showscale=False,
    name="Parábola"
), row=1, col=2)

# Hipérbola (plano vertical)
Z_hyper = YY

fig.add_trace(go.Surface(
    x=XX, y=YY, z=Z_hyper,
    opacity=0.5,
    showscale=False,
    name="Hipérbola"
), row=1, col=2)

# -------------------------
# Layout
# -------------------------

fig.update_layout(
    title="Cono de Apolonio y sus secciones cónicas",
    height=700
)

fig.show()
fig.write_image(r"C:\Users\marco\Downloads\cono_y_secciones.png")
````

# **21.1 Secciones cónicas**

La ecuación general de segundo grado es:

$$
Ax^2 + Bxy + Cy^2 + Dx + Ey + F = 0
$$

La gráfica de esta ecuación es una sección cónica o un caso degenerado de éstas como un punto, una recta, un par de rectas o el conjunto vacío.
Cuando la constante B es cero, o sea cuando la ecuación no tiene término en xy, los ejes de simetría de la cónica son paralelos a los ejes coordenados.


Las cónicas no degeneradas están en posición canónica si sus gráficas y ecuaciones son como se indica en la figura siguiente. 

La ecuación está en forma canónica.


### Clasificación:

* Si $B = 0$ → ejes alineados con los ejes coordenados. 
* Si $B \neq 0$ → cónica rotada
* Si hay términos lineales → cónica trasladada



````python
import numpy as np
import matplotlib.pyplot as plt

fig, axs = plt.subplots(3, 3, figsize=(12, 12))

# =========================
# 1. Circunferencia
# =========================
a = 3
t = np.linspace(0, 2*np.pi, 400)

x = a * np.cos(t)
y = a * np.sin(t)

axs[0,0].plot(x, y, label=r"$x^2 + y^2 = {}^2$".format(a))
axs[0,0].scatter([0], [0], color="red", label="Centro")
axs[0,0].set_title("Circunferencia")

# =========================
# 2. Elipse eje mayor en x
# =========================
a, b = 4, 2
c = np.sqrt(a**2 - b**2)

x = a * np.cos(t)
y = b * np.sin(t)

axs[0,1].plot(x, y, label=r"$\frac{x^2}{%d} + \frac{y^2}{%d} = 1$"%(a**2,b**2))
axs[0,1].scatter([c,-c],[0,0],color="red",label="Focos")
axs[0,1].scatter([a,-a],[0,0],color="red",marker="x",label="Vértices")
axs[0,1].set_title("Elipse eje x")

# =========================
# 3. Elipse eje mayor en y
# =========================
a, b = 2, 4
c = np.sqrt(b**2 - a**2)

x = a * np.cos(t)
y = b * np.sin(t)

axs[0,2].plot(x, y, label=r"$\frac{x^2}{%d} + \frac{y^2}{%d} = 1$"%(a**2,b**2))
axs[0,2].scatter([0,0],[c,-c],color="red",label="Focos")
axs[0,2].scatter([0,0],[b,-b],color="red",marker="x",label="Vértices")
axs[0,2].set_title("Elipse eje y")

# =========================
# Parábolas
# =========================

x = np.linspace(-5,5,400)

# Arriba
y = x**2/4
axs[1,0].plot(x,y,label=r"$y=\frac{x^2}{4}$")
axs[1,0].scatter([0],[1],color="red",label="Foco")
axs[1,0].scatter([0],[0],color="red",marker="x",label="Vértice")
axs[1,0].set_title("Parábola ↑")

# Abajo
y = -x**2/4
axs[1,1].plot(x,y,label=r"$y=-\frac{x^2}{4}$")
axs[1,1].scatter([0],[-1],color="red",label="Foco")
axs[1,1].scatter([0],[0],color="red",marker="x",label="Vértice")
axs[1,1].set_title("Parábola ↓")

# Derecha
y = np.linspace(-5,5,400)
x = y**2/4
axs[1,2].plot(x,y,label=r"$x=\frac{y^2}{4}$")
axs[1,2].scatter([1],[0],color="red",label="Foco")
axs[1,2].scatter([0],[0],color="red",marker="x",label="Vértice")
axs[1,2].set_title("Parábola →")

# Izquierda
x = -y**2/4
axs[2,0].plot(x,y,label=r"$x=-\frac{y^2}{4}$")
axs[2,0].scatter([-1],[0],color="red",label="Foco")
axs[2,0].scatter([0],[0],color="red",marker="x",label="Vértice")
axs[2,0].set_title("Parábola ←")

# =========================
# Hipérbolas
# =========================

# Eje x
a, b = 2, 1
c = np.sqrt(a**2 + b**2)

x = np.linspace(-5,-a,200)
y = b*np.sqrt(x**2/a**2 - 1)
axs[2,1].plot(x,y)
axs[2,1].plot(x,-y)
x = np.linspace(a,5,200)
y = b*np.sqrt(x**2/a**2 - 1)
axs[2,1].plot(x,y,label=r"$\frac{x^2}{4}-\frac{y^2}{1}=1$")
axs[2,1].plot(x,-y)
axs[2,1].scatter([c,-c],[0,0],color="red",label="Focos")
axs[2,1].scatter([a,-a],[0,0],color="red",marker="x",label="Vértices")
axs[2,1].set_title("Hipérbola eje x")

# Eje y
x = np.linspace(-5,5,400)
y = np.sqrt((x**2 + 1))

axs[2,2].plot(x,y)
axs[2,2].plot(x,-y,label=r"$\frac{y^2}{4}-\frac{x^2}{1}=1$")
axs[2,2].scatter([0,0],[np.sqrt(5),-np.sqrt(5)],color="red",label="Focos")
axs[2,2].scatter([0,0],[2,-2],color="red",marker="x",label="Vértices")
axs[2,2].set_title("Hipérbola eje y")

# =========================
# Ajustes
# =========================

for ax in axs.flat:
    ax.axhline(0)
    ax.axvline(0)
    ax.legend(fontsize=8)
    ax.set_aspect('equal')

plt.tight_layout()
plt.savefig(r"C:\Users\marco\Downloads\9_secciones_conicas.png")
plt.show()
````

# **21.2 Formas cuadráticas en $\mathbb{R}^2$**

Una forma cuadrática es:

$$
F(x,y) = Ax^2 + Bxy + Cy^2
$$



### Forma matricial

$$
F(x,y) = \begin{bmatrix} x & y \end{bmatrix}
\begin{bmatrix} A & \frac{B}{2} \\  \frac{B}{2} & C \end{bmatrix}
\begin{bmatrix} x \\ y \end{bmatrix}
$$

---



## Ejemplo con pandas


````python
import pandas as pd

R = pd.DataFrame(
    [[3,1],
     [1,1]],
    columns=["x","y"],
    index=["x","y"]
)

print(R)
````

## Interpretación en Ciencia de Datos

* $R$ puede ser una **matriz de covarianza**
* $F(x,y)$ mide dispersión
* Direcciones principales → **componentes principales**

---



# **21.3 Teorema de los ejes principales**

## Idea clave

Toda forma cuadrática se puede diagonalizar:

$$
R = P D P^T
$$

y haciendo:

$$
X' = P^T X
$$

se obtiene:

$$
F(x',y') = \lambda_1 x'^2 + \lambda_2 y'^2
$$

---



## Rotación

Si $P$ es ortogonal con $\det(P)=1$:

$$
P = \begin{bmatrix}
\cos\theta & -\sin\theta \\ 
\sin\theta & \cos\theta
\end{bmatrix}
$$

---



## Visualización con seaborn (distribución rotada)


````python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

mean = [0,0]
cov = [[3,1],[1,1]]

data = np.random.multivariate_normal(mean, cov, 500)

sns.scatterplot(x=data[:,0], y=data[:,1])
plt.title("Datos con correlación (forma cuadrática)")
plt.savefig(r"C:\Users\marco\Downloads\datos_correlacion_forma_cuadratica.png")
plt.show()
````

## PCA con Python (Machine Learning)


````python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
pca.fit(data)

print("Valores propios:", pca.explained_variance_)
print("Vectores propios:", pca.components_)
````

```

- **Interpretación:**

* Valores propios → forma de la cónica
* Vectores propios → rotación óptima
* Esto es exactamente el **Teorema de los ejes principales**

# **Clasificación usando valores propios**

Dada:

$$
F(x,y) = \lambda_1 x'^2 + \lambda_2 y'^2
$$

* $\lambda_1, \lambda_2 > 0$ → elipse
* signos opuestos → hipérbola
* uno cero → parábola

# **21.4 Formas cuadráticas en $\mathbb{R}^n$**

Definición:

$$
F(X) = X^T R X
$$

donde:

$$
X =
\begin{bmatrix}
x_1 \\ x_2 \\ \vdots \\ x_n
\end{bmatrix}
\quad
R \in \mathbb{R}^{n \times n}
$$

---

````python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact

def plot_quad(a=1, b=0, c=1):
    x = np.linspace(-5,5,200)
    y = np.linspace(-5,5,200)
    X, Y = np.meshgrid(x,y)
    Z = a*X**2 + b*X*Y + c*Y**2
    
    plt.contour(X,Y,Z)
    plt.title("Forma cuadrática interactiva")
    plt.savefig(r"C:\Users\marco\Downloads\forma_cuadratica_interactiva.png")
    plt.show()

interact(plot_quad, a=(-3,3), b=(-3,3), c=(-3,3))
````

## **Problema 3 (Computacional con Python)**

Considere la matriz de covarianza:

$$
R =
\begin{bmatrix}
5 & 3 \\ 
3 & 2
\end{bmatrix}
$$

### a) Simule 1000 datos con distribución normal multivariada

### b) Grafique los datos

### c) Aplique PCA

### d) Interprete geométricamente el resultado



### Código esperado

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

mean = [0,0]
cov = [[5,3],[3,2]]

data = np.random.multivariate_normal(mean, cov, 1000)

plt.scatter(data[:,0], data[:,1], alpha=0.3)
plt.title("Datos correlacionados")
plt.grid(alpha = 0.3)
plt.savefig(r"C:\Users\marco\Downloads\datos_correlacionados.png")
plt.show()

pca = PCA()
pca.fit(data)

print(pca.explained_variance_)
print(pca.components_)
````
